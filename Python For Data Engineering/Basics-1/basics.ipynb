## 2️⃣ Comments in Python

Comments are used to explain code and are **ignored** by the Python interpreter.

| Type | Syntax | Description |
|---|---|---|
| Single-line | `# comment` | Starts with `#`, extends to end of line |
| Multi-line | `""" ... """` | Triple-quoted string literal (not assigned to a variable) |
| Docstring | `""" ... """` | Triple-quoted string as the **first statement** in a function/class/module |

> 💡 Use `Ctrl + /` in VS Code to quickly toggle comments on selected lines.

> ⚠️ Multi-line "comments" using `"""..."""`  are technically string literals that are not assigned to any variable and thus ignored by the interpreter.

````python
# ============================================
# 2️⃣ Comments in Python
# ============================================

# This is a single-line comment

"""
This is a multi-line comment.
It can span multiple lines.

But technically, it's a multi-line string literal that is not
assigned to any variable and thus ignored by the interpreter.
"""

# Docstrings — used inside functions/classes for documentation
def add(a, b):
    """
    This function adds two numbers
    and returns the result.
    """
    return a + b

result = add(5, 3)
print(f"The sum is: {result}")

# Comments are ignored by the Python interpreter
# They are meant for humans to read and understand the code better
````

## 4️⃣ Identifiers

Identifiers are **names** used in a Python program — variable names, function names, class names, etc.

### ✅ Rules for Identifiers

| Rule | Description |
|---|---|
| Allowed characters | Letters (A–Z, a–z), digits (0–9), underscore (`_`) |
| Must start with | A letter or underscore |
| Cannot start with | A digit |
| No special characters | `@`, `#`, `$`, `%`, `-`, spaces, etc. are **not allowed** |
| No keywords | Python keywords cannot be used as identifiers |

### ✔ Valid Examples
```python
name = "Rajan"
_age = 21
total_marks = 450
```

### ❌ Invalid Examples
```python
1name = "Rajan"      # Cannot start with digit
my-name = "Rajan"    # Hyphen not allowed
class = 10           # 'class' is a keyword
name with space = 1  # Spaces not allowed
name@domain = 1      # Special characters not allowed
```

````python
# ============================================
# 4️⃣ Identifiers — Examples
# ============================================

# ✔ Valid identifiers
name = "Rajan"
_age = 21
total_marks = 450
print(f"name: {name}, _age: {_age}, total_marks: {total_marks}")

# ❌ Invalid identifiers (uncommenting these will raise SyntaxError)
# 1name = "Rajan"       # Cannot start with a digit
# my-name = "Rajan"     # Hyphen not allowed
# class = 10            # 'class' is a keyword
# name with space = 1   # Spaces not allowed
# name@domain = 1       # Special characters not allowed
````

## 5️⃣ Variables

A variable is a **name** that refers to a value stored in memory. Python variables are **dynamically typed** — you don't need to declare the type.

### Memory Allocation

In Python, variables are **references** (pointers) to objects in memory. When two variables hold the same small integer value, Python may **reuse the same memory address** (integer caching for -5 to 256).

> 💡 Use `id()` to check the memory address of a variable.

````python
# ============================================
# 5️⃣ Variables — Memory Allocation
# ============================================

# Python reuses memory for small integers (-5 to 256)
a = 10
b = 10
c = 10
print(f"Memory address of a: {id(a)}")
print(f"Memory address of b: {id(b)}")
print(f"Memory address of c: {id(c)}")
# → All three point to the SAME memory address!

# Changing the value creates a NEW object in memory
a = 20
print(f"\nAfter a = 20:")
print(f"Value of b: {b}")  # b is still 10
print(f"Value of c: {c}")  # c is still 10
print(f"New memory address of a: {id(a)}")
````

### Multiple Variable Assignment

Python allows assigning values to multiple variables in a single line.

````python
# ============================================
# 5️⃣ Variables — Multiple Assignment
# ============================================

# Assigning different values to multiple variables in one line
x, y, z = 1, 2, 3
print(f"x: {x}, y: {y}, z: {z}")

# Assigning the same value to multiple variables
m = n = p = 50
print(f"m: {m}, n: {n}, p: {p}")
````

### Naming Conventions & Case Sensitivity

| Rule | Example |
|---|---|
| Can start with a letter or `_` | `valid_name`, `_validName` |
| Can contain letters, digits, `_` | `validName123` |
| Cannot start with a digit | ❌ `1name = 400` |
| Cannot contain hyphens or spaces | ❌ `name-with-hyphen`, `name with space` |
| Cannot contain special characters | ❌ `name@domain` |

> ⚠️ Python variables are **case-sensitive** — `var` and `Var` are two different variables.

````python
# ============================================
# 5️⃣ Variables — Naming Conventions & Case Sensitivity
# ============================================

# ✔ Valid variable names
valid_name = 100
_validName = 200
validName123 = 300
print(f"valid_name: {valid_name}, _validName: {_validName}, validName123: {validName123}")

# ❌ Invalid variable names (uncommenting these will raise errors)
# 1name = 400             # Cannot start with a number
# name-with-hyphen = 500  # Hyphens not allowed
# name with space = 600   # Spaces not allowed
# name@domain = 700       # Special characters not allowed

# Case sensitivity — var and Var are different variables
var = 10
Var = 20
print(f"var: {var}, Var: {Var}")
````

## 6️⃣ The `id()` Function — Object Identity

`id(object)` returns the **unique memory address** (identity) of an object.

Since Python objects like `int`, `float`, and `str` are **immutable**, modifying a value creates a **new object** with a new `id`.

| Type | Behavior on Modification |
|---|---|
| `int` | New object created → new `id` |
| `float` | New object created → new `id` |
| `str` | New object created → new `id` |

> 💡 Immutable objects **cannot be changed in place**. Any "modification" produces a brand new object.

````python
# ============================================
# 6️⃣ The id() Function — Object Identity
# ============================================

# --- Integer Example ---
x = 10
print("Integer Example:")
print(f"Initial ID of x: {id(x)}")
x = x + 1
print(f"New Value of x: {x}")
print(f"New ID of x: {id(x)}")
# → id changes because int is immutable!
print("-" * 30)

# --- Float Example ---
y = 10.5
print("Float Example:")
print(f"Initial ID of y: {id(y)}")
y = y + 1.1
print(f"New Value of y: {y}")
print(f"New ID of y: {id(y)}")
# → id changes because float is immutable!
print("-" * 30)

# --- String Example ---
z = "Hello"
print("String Example:")
print(f"Initial ID of z: {id(z)}")
z = z + " World"
print(f"New Value of z: {z}")
print(f"New ID of z: {id(z)}")
# → id changes because str is immutable!
````

## 7️⃣ The `print()` Function & `end` Parameter

By default, Python’s `print()` function appends a **newline character** (`\n`) after each output, automatically moving the cursor to the next line.

**Syntax:** `print(*objects, sep=' ', end='\n', file=sys.stdout)`

| Parameter | Default | Description |
|---|---|---|
| `sep` | `' '` (space) | Separator between multiple arguments |
| `end` | `'\n'` (newline) | What to print at the end of the output |

> 💡 You can customize the `end` parameter to control what happens after `print()` finishes.

````python
# ============================================
# 7️⃣ print() — Printing on the Same Line
# ============================================

# Use Case 1: Printing on the same line with spaces
print("Hello,", end=" ")
print("World!")

print()  # blank line for clarity

# Use Case 2: Printing without any separator (empty end)
print("H", end="")
print("e", end="")
print("l", end="")
print("l", end="")
print("o")
````

````python
# ============================================
# 7️⃣ print() — Loop Printing with Separators
# ============================================

# Use Case 3: Loop printing with comma separator
for i in range(5):
    print(i, end=", ")
print()  # Move to the next line after the loop

# Use Case 4: Custom character as end value
print("Item 1", end=" - ")
print("Item 2", end=" - ")
print("Item 3")
````

````python
# ============================================
# 7️⃣ print() — Tab Separators for Aligned Columns
# ============================================

# Use Case 5: Tab separators for aligned columns
print("Name", end="\t")
print("Age", end="\t")
print("City")
````

````python
# ============================================
# 7️⃣ print() — Progress Bar Simulation
# ============================================

import time

# Use Case 6: Progress bar simulation using \r (carriage return)
for i in range(11):
    # \r moves cursor to start, end="" prevents new line
    print(f"\rProgress: {i * 10}%", end="")
    time.sleep(0.3)
print(" Done!")
````

````python
# ============================================
# 7️⃣ print() — CSV-Like Output
# ============================================

# Use Case 7: CSV-like output using nested loops
data = [
    ["Name", "Age", "City"],
    ["Alice", "25", "NYC"],
    ["Bob", "30", "LA"]
]

for row in data:
    for item in row:
        print(item, end=",")
    print()  # Newline for next row
````

## 8️⃣ Control Characters (Escape Sequences)

Control characters are special sequences that control the **output format**.

| Escape Sequence | Description | Example Output |
|---|---|---|
| `\n` | New line | Moves to next line |
| `\t` | Tab space | Adds horizontal tab |
| `\\` | Backslash | Prints `\` |
| `\"` | Double quote inside string | Prints `"` |
| `\'` | Single quote inside string | Prints `'` |
| `\r` | Carriage return | Moves cursor to start of line |

````python
# ============================================
# 8️⃣ Control Characters (Escape Sequences)
# ============================================

# \n — New line
print("Hello\nWorld")
print("-" * 20)

# \t — Tab space
print("Name:\tRajan")
print("-" * 20)

# \\ — Backslash
print("File path: C:\\Users\\Rajan")
print("-" * 20)

# \" — Double quote inside string
print("He said \"Hello\"")
print("-" * 20)

# \' — Single quote inside string
print('It\'s Python!')
````

## 9️⃣ Line Joining Methods

Sometimes a statement is too long and we want to write it across multiple lines.

### Implicit Line Joining

Python automatically joins lines when they are inside:
- **Parentheses** `()`
- **Square brackets** `[]`
- **Curly braces** `{}`

> ✔ No backslash needed. This is the **cleaner and recommended method**.

### Explicit Line Joining

Uses a **backslash** `\` at the end of a line to continue the statement on the next line.

> ⚠️ **No space allowed** after the backslash `\`.

````python
# ============================================
# 9️⃣ Line Joining — Implicit (Recommended)
# ============================================

# Using Parentheses
total = (10 +
         20 +
         30)
print(f"Total (parentheses): {total}")

# Using List (Square Brackets)
numbers = [
    10,
    20,
    30,
    40
]
print(f"Numbers list: {numbers}")

# Using Dictionary (Curly Braces)
student = {
    "name": "Rajan",
    "age": 21,
    "course": "CSE"
}
print(f"Student dict: {student}")
````

````python
# ============================================
# 9️⃣ Line Joining — Explicit (Backslash)
# ============================================

# Using backslash \ to continue on next line
total = 10 + \
        20 + \
        30
print(f"Total (backslash): {total}")

# ⚠️ No space allowed after backslash!
# Backslash tells Python that the statement continues on the next line
````

## 🔟 Python Execution Model

Understanding how Python executes your code is crucial for debugging.

```
Step 1: Python checks the ENTIRE file for syntax errors
Step 2: If syntax is valid, Python compiles to bytecode (.pyc)
Step 3: Bytecode is executed line by line by the PVM (Python Virtual Machine)
```

| Step | What happens | Error Type |
|---|---|---|
| Step 1 | Grammar check of entire file | **SyntaxError** (no code runs) |
| Step 2 | Compilation to bytecode (.pyc) | — |
| Step 3 | Line-by-line execution | **RuntimeError** (partial execution possible) |

---
## 1️⃣1️⃣ Syntax Error vs Runtime Error

### Syntax Errors

A `SyntaxError` occurs when Python **cannot understand the structure** (grammar) of the code.

| Property | Detail |
|---|---|
| **When** | Happens **BEFORE** program execution |
| **Scope** | The **entire file** is checked first |
| **Execution** | **NO** line of code runs |
| **Bytecode** | `.pyc` is **NOT** generated |
| **Catchable** | **Cannot** be caught using `try-except` |

**Common causes:**
- Missing colon `:`
- Unclosed quotes or brackets
- Invalid indentation
- Incorrect keywords
- Broken expressions

```python
# Examples (all cause SyntaxError):
if True                    # missing colon
    print("Hello")

print("Hello)              # unclosed string

for i in range(5))         # extra parenthesis
    print(i)
```

````python
# ============================================
# 1️⃣1️⃣ Syntax Error — Demonstration
# ============================================

# These lines run successfully (no syntax errors in this cell)
print("There is no Error in this line")
print("There is no Error in this line too")

# ❌ Uncommenting the line below will cause a SyntaxError
# → The ENTIRE cell will fail, including the print statements above
# Prin("There is error in this line")
````

### Runtime Errors

A `RuntimeError` occurs when Python **understands the code** but **fails while executing** it.

| Property | Detail |
|---|---|
| **When** | Happens **DURING** execution |
| **Scope** | Code runs **until** the error line is reached |
| **Execution** | Partial execution is possible |
| **Bytecode** | `.pyc` **IS** generated |
| **Catchable** | **Can** be handled using `try-except` |

**Common causes:**
- Using undefined variables or functions
- Division by zero
- Invalid type operations
- Index out of range
- File or module not found

**Common error types:**
`NameError`, `TypeError`, `ZeroDivisionError`, `IndexError`, `ValueError`, `ModuleNotFoundError`

````python
# ============================================
# 1️⃣1️⃣ Runtime Error — Demonstrations
# ============================================

# Example 1: ZeroDivisionError
try:
    x = 10 / 0
except ZeroDivisionError as e:
    print(f"ZeroDivisionError: {e}")

# Example 2: NameError (misspelled function)
try:
    print("Hello")  # 'Prin' is not defined
except NameError as e:
    print(f"NameError: {e}")

# Example 3: IndexError
try:
    lst = [1, 2, 3]
    print(lst[5])  # Index 5 doesn't exist
except IndexError as e:
    print(f"IndexError: {e}")

# Example 4: TypeError
try:
    result = "Hello" + 5  # Can't add string and int
except TypeError as e:
    print(f"TypeError: {e}")
````

### Key Differences — One-Look Comparison

| Feature | Syntax Error | Runtime Error |
|---|---|---|
| **Type of problem** | Grammar / structure issue | Logic / data issue |
| **Detected** | **Before** execution | **During** execution |
| **Code runs?** | ❌ No code runs at all | ✅ Partial execution possible |
| **Catchable?** | ❌ Cannot be caught with `try-except` | ✅ Can be caught with `try-except` |
| **Bytecode (.pyc)** | ❌ NOT generated | ✅ IS generated |

---

### 🏆 Golden Rule (Remember Forever!)

> **If Python runs even ONE line of code, then the entire file is syntactically correct.**
>
> A `SyntaxError` anywhere in the file means **nothing** runs at all.
