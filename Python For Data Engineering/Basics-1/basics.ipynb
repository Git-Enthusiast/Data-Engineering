## 2️⃣ Comments, Code Annotation & Docstrings (PEP 257)

Comments are non-executable annotations used to document logic, explain algorithms, and improve code maintainability. They are completely ignored by the Python interpreter during bytecode compilation.

| Type | Syntax | Purpose / Placement |
|---|---|---|
| **Single-line Comment** | `# comment text` | Inline explanation or single-line note starting with `#` |
| **Multi-line String** | `\'\'\' ... \'\'\'` or `\"\"\" ... \"\"\"` | Unassigned string literals ignored by the interpreter at runtime |
| **Docstring (PEP 257)** | `\"\"\" ... \"\"\"` | First statement in a module, class, or function; accessible via `__doc__` |

> 💡 **Productivity Shortcut**: Press `Ctrl + /` (Windows/Linux) or `Cmd + /` (Mac) in VS Code to toggle comments on selected lines.
> 
> ⚠️ **Docstring vs Comment**: A true docstring is assigned to the object's `__doc__` attribute and retained in memory for documentation tools (like `help()`), whereas `#` comments are discarded during parsing.


````python
# ============================================================
# 2️⃣ Comments, Multi-line Strings & Docstrings
# ============================================================

# This is a standard single-line comment

"""
This is a multi-line string literal.
When it is not assigned to a variable, the Python interpreter
treats it as a multi-line comment.
"""

def calculate_pipeline_throughput(records_processed, time_in_seconds):
    """
    Calculates the throughput of a data pipeline in records per second.
    
    Parameters:
        records_processed (int): Total records processed
        time_in_seconds (float): Execution time in seconds
        
    Returns:
        float: Throughput rate (records/sec)
    """
    return records_processed / time_in_seconds

# Invoking the function and inspecting docstring
rate = calculate_pipeline_throughput(50000, 12.5)
print(f"Pipeline Throughput: {rate:.2f} records/sec")

print("\n--- Inspecting Function Docstring (__doc__) ---")
print(calculate_pipeline_throughput.__doc__)
````

## 3️⃣ Python Keywords, Reserved Identifiers & `keyword` Module

Keywords are **reserved words** that have special syntactic meanings in Python. They form the core vocabulary of the language and **cannot be used as variable names, function names, or any other identifier**.

### 🔍 Keyword Introspection
Python provides the built-in `keyword` module to programmatically inspect all keywords supported by the active Python runtime.


````python
# ============================================================
# 3️⃣ Python Keywords & The keyword Module
# ============================================================

import keyword

print("--- All Python Keywords (keyword.kwlist) ---")
print(keyword.kwlist)
print(f"\nTotal Keywords in Python {keyword.__file__ if hasattr(keyword, '__file__') else ''}: {len(keyword.kwlist)}")

# Checking if a string is a reserved keyword
test_words = ["for", "while", "lambda", "data_engineering", "True", "class", "spark"]

print("\n--- Keyword Validation Test ---")
for word in test_words:
    is_kw = keyword.iskeyword(word)
    print(f"'{word}': {'❌ Reserved Keyword' if is_kw else '✅ Valid Custom Identifier'}")
````

## 4️⃣ Identifiers & Naming Rules (PEP 8 Conventions)

An **identifier** is a user-defined name given to entities such as variables, functions, classes, modules, and other objects in Python.

### 📋 Lexical Rules for Identifiers

| Rule | Description | Valid Example | Invalid Example |
|---|---|---|---|
| **Allowed Characters** | Letters (`A-Z`, `a-z`), Digits (`0-9`), Underscore (`_`) | `user_name_01`, `_config` | `user-name` (No hyphens), `user name` (No spaces) |
| **Starting Character** | Must start with a letter or underscore (`_`) | `_total`, `batch_size` | `1name = 10` (Cannot start with a digit) |
| **No Special Symbols** | Symbols like `@`, `#`, `$`, `%`, `&`, `!` are forbidden | `amount_usd` | `amount$ = 100`, `name@domain = 5` |
| **No Keywords** | Cannot be a reserved Python keyword | `class_name` | `class = "A"`, `def = 10` |
| **Case Sensitivity** | Identifiers are strictly case-sensitive | `total` $\ne$ `Total` $\ne$ `TOTAL` | Accidental shadowing |


````python
# ============================================================
# 4️⃣ Identifiers & Case Sensitivity Examples
# ============================================================

# ✅ Valid Identifiers
dataset_name = "BigQuery_Orders"
_batch_id = 101
total_records_2025 = 500000

print("--- Valid Identifiers ---")
print(f"dataset_name       : {dataset_name}")
print(f"_batch_id          : {_batch_id}")
print(f"total_records_2025 : {total_records_2025}")

# Case Sensitivity Demonstration
count = 100
Count = 200
COUNT = 300

print("\n--- Case Sensitivity Check ---")
print(f"count: {count} | Count: {Count} | COUNT: {COUNT}")
print(f"Are 'count' and 'Count' identical? {count is Count}")

# ❌ Invalid Identifiers (Uncommenting any line below will raise a SyntaxError):
# 1st_record = "Alice"     # SyntaxError: invalid decimal literal
# record-id = 500          # SyntaxError: cannot assign to expression here
# user name = "Bob"        # SyntaxError: invalid syntax
# user@domain = "test"     # SyntaxError: cannot assign to expression
# class = "Engineering"    # SyntaxError: invalid syntax (keyword!)
````

## 5️⃣ Variables, Dynamic Typing & Multiple Assignment Paradigms

A **variable** in Python is a named reference that points to an object stored in computer memory. Unlike languages like C++ or Java, Python variables are **dynamically typed**:
* Variables do not possess a fixed data type.
* The data type belongs to the **object**, not the variable name.
* No type declaration (`int x;`) is required.

---

### 🔄 Multiple Variable Assignment Paradigms

1. **Parallel Assignment (Tuple Unpacking)**: Assigns distinct values to distinct variables in a single line.
   ```python
   x, y, z = 10, 20, 30
   ```
2. **Chained Assignment (Shared Reference)**: Assigns the same value to multiple variable names simultaneously.
   ```python
   a = b = c = 100
   ```


````python
# ============================================================
# 5️⃣ Variables & Multiple Assignment Paradigms
# ============================================================

# 1. Dynamic Typing: Re-binding a variable to different types

dynamic_var = 100
print(f"dynamic_var: {dynamic_var} | Type: {type(dynamic_var)}")

dynamic_var = "Now a String"
print(f"dynamic_var: {dynamic_var} | Type: {type(dynamic_var)}")


# 2. Multiple Variable Assignment (Parallel Unpacking)

batch_id, table_name, row_count = 101, "raw_telemetry", 45000

print("\n--- Parallel Assignment ---")
print(f"batch_id   : {batch_id}")
print(f"table_name : {table_name}")
print(f"row_count  : {row_count}")


# 3. Chained Assignment (Shared Reference)

initial_state = default_state = backup_state = "PENDING"

print("\n--- Chained Assignment ---")
print(f"initial: {initial_state}, default: {default_state}, backup: {backup_state}")
print(f"All point to same memory address: "
      f"{id(initial_state) == id(default_state) == id(backup_state)}")


# 4. Tuple Unpacking with * (Starred Unpacking)

a, *b = 2, 3, 4, 5, 6

print("\n--- Tuple Unpacking with * ---")
print(f"a : {a}")
print(f"b : {b}")
print(f"Type of b: {type(b)}")
````

## 6️⃣ Object Identity & Memory Model (`id()` & Small Integer Caching)

Every object created in Python is assigned an identity: a unique integer representing its memory address during its lifetime. The built-in `id(object)` function returns this address.

```
       VARIABLES ARE REFERENCES (LABELS)                  CPYTHON INTEGER CACHING (-5 to 256)
     ┌──────────────┐                                    ┌───────────────────────────────────┐
     │      a       ├──┐                                 │ Pre-allocated integer objects     │
     └──────────────┘  │                                 │ [ -5, -4, ..., 0, 1, ..., 256 ]   │
                       ├──> [ Integer Object: 10 ] <─────┤                                   │
     ┌──────────────┐  │    (Memory: 0x7FFE...)          │ Both a and b share same memory!   │
     │      b       ├──┘                                 └───────────────────────────────────┘
     └──────────────┘
```

### 🧠 Immutability & Re-Binding
Because numbers and strings are **immutable**, any modification creates a **new object in memory** rather than changing the existing object in place.


````python
# ============================================================
# 6️⃣ Object Identity & Memory Address Inspection
# ============================================================

# 1. CPython Small Integer Caching (-5 to 256)
a = 10
b = 10
c = 10

print("--- Small Integer Caching (-5 to 256) ---")
print(f"id(a): {id(a)}")
print(f"id(b): {id(b)}")
print(f"id(c): {id(c)}")
print(f"a is b: {a is b} (Pointing to the same pre-allocated cached object!)")

# 2. Immutability in Action: Re-binding creates a NEW object
a = a + 5  # a is now 15
print("\n--- After a = a + 5 ---")
print(f"Value of a: {a} | New id(a): {id(a)}")
print(f"Value of b: {b} | id(b) is still: {id(b)} (Unaltered!)")

# 3. Float Immutability
f1 = 10.5
print("\n--- Float Immutability ---")
print(f"Initial f1: {f1} | id(f1): {id(f1)}")
f1 = f1 + 1.5
print(f"Updated f1: {f1} | id(f1): {id(f1)} (New float object allocated)")

# 4. String Immutability
s1 = "Data"
print("\n--- String Immutability ---")
print(f"Initial s1: '{s1}' | id(s1): {id(s1)}")
s1 = s1 + " Engineer"
print(f"Updated s1: '{s1}' | id(s1): {id(s1)} (New string object allocated)")
````

## 7️⃣ The Deep Dive on `print()` — Complete 5-Parameter Anatomy

The `print()` function is one of the most versatile built-in tools in Python. To master it, you must understand its full internal signature:

$$\mathbf{\text{print}(*values, \; sep=\text{' '}, \; end=\text{'\\n'}, \; file=sys.stdout, \; flush=False) \rightarrow None}$$

### 📊 The 5 Core Parameters of `print()`

| Parameter | Type | Default | Detailed Description |
|---|---|---|---|
| `*values` | `tuple` (Variable-length) | — | Any number of positional arguments of any type. Converted internally using `str()`. |
| `sep` | `str` or `None` | `' '` (Space) | String inserted between multiple values when printed. |
| `end` | `str` or `None` | `'\n'` (Newline) | String appended at the very end after all values are printed. |
| `file` | Stream object | `sys.stdout` | Target output stream (console or any file object with a `.write()` method). |
| `flush` | `bool` | `False` | Whether to forcefully flush the I/O buffer immediately (`True`) or allow buffering. |

---

### ⚠️ Common `print()` Traps & `TypeError` Rules
1. **`sep` and `end` MUST be strings or `None`**: Passing an integer like `print("A", "B", sep=5)` raises `TypeError: sep must be None or a string, not int`.
2. **`file` must have a `.write()` method**: Passing a raw string filename like `print("Hi", file="log.txt")` raises `AttributeError: 'str' object has no attribute 'write'`.
3. **`print()` always returns `None`**: It performs I/O and returns `NoneType`.


````python
# ============================================================
# 7️⃣ Basic print() Usage, Multiple Values & sep Parameter
# ============================================================

# 1. Printing single and heterogeneous multiple values
print("--- 1. Printing Heterogeneous Values ---")
print("Rajan", "Raj", 2025, 45.8, True, None)

# 2. Default Separator vs Custom Separators (sep)
print("\n--- 2. The 'sep' Parameter ---")
print("Python", "Java", "C++")                 # Default space separator
print("Python", "Java", "C++", sep=", ")       # Comma space separator
print(1, 2, 3, 4, 5, sep=" | ")                # Pipeline delimiter
print("2025", "12", "31", sep="-")             # Date format YYYY-MM-DD
print("data", "warehouse", "layer", sep="/")   # File path simulation
````

````python
# ============================================================
# 7️⃣ The 'end' Parameter & Combining sep + end
# ============================================================

print("--- 1. The 'end' Parameter ---")
# Default end is '\n' (moves to next line)
print("Line 1")
print("Line 2")

# Custom end with space (stays on same line)
print("Hello", end=" ")
print("World!")

# Custom end with empty string (concatenates output)
print("D", end="")
print("A", end="")
print("T", end="")
print("A")

# 2. Combining sep and end together
print("\n--- 2. Combining 'sep' and 'end' ---")
print("Batch_01", "INGESTION", "COMPLETED", sep=" | ", end=" [STATUS: SUCCESS] ✅\n")

# 3. CSV formatting using loops
print("\n--- 3. CSV Formatted Table Output ---")
headers = ["ID", "Sensor", "Value"]
rows = [
    [101, "Temperature", 24.5],
    [102, "Pressure", 101.3],
    [103, "Humidity", 65.0]
]

print(*headers, sep=",")
for row in rows:
    print(*row, sep=",")
````

## 8️⃣ Control Characters & Escape Sequences

Escape sequences begin with a backslash (`\`) and allow you to insert characters that are otherwise difficult or impossible to type directly into string literals.

| Escape Sequence | Name | Effect / Output |
|:---:|---|---|
| `\n` | Newline | Moves the cursor to the beginning of the next line |
| `\t` | Horizontal Tab | Inserts a standard tab space (column alignment) |
| `\\` | Backslash Literal | Prints a literal backslash `\` character |
| `\"` | Double Quote | Prints `"` inside a double-quoted string |
| `\'` | Single Quote | Prints `'` inside a single-quoted string |
| `\r` | Carriage Return | Moves the cursor back to the start of the current line (in-place rewriting) |


````python
# ============================================================
# 8️⃣ Control Characters & Escape Sequences
# ============================================================

# 1. Newline (\n) and Tab (\t)
print("--- 1. Newline (\\n) and Tab (\\t) ---")
print("Column1\tColumn2\tColumn3")
print("DataA\tDataB\tDataC")
print("\nMulti\nLine\nOutput")

# 2. Quotes (\", \') and Backslashes (\\)
print("\n--- 2. Quotes & Backslashes ---")
print("Windows Path: C:\\Users\\Rajan\\DataEngineering\\Basics")
print("He said, \"Data is the new oil!\"")
print('It\'s a Python tutorial for Engineers.')

# 3. Carriage Return (\r) for Live In-Place Progress Bar
print("\n--- 3. Progress Bar Simulation with \\r ---")
for percent in [0, 25, 50, 75, 100]:
    bar = "█" * (percent // 10) + "-" * (10 - (percent // 10))
    print(f"\rProgress: [{bar}] {percent}% Complete", end="", flush=True)
    time.sleep(0.1)
print(" Done! 🚀")
````

## 9️⃣ Line Continuation & Joining Methods (Implicit vs Explicit)

PEP 8 recommends limiting all lines to a maximum of 79 characters for clarity. Python provides two ways to split long statements across multiple lines:

### 1. Implicit Line Continuation (✅ Recommended)
Python automatically continues statements that are wrapped inside **parentheses `()`**, **brackets `[]`**, or **braces `{}`**.
* Cleanest and most Pythonic method.
* No special escape characters needed.

### 2. Explicit Line Continuation (`\`)
Uses a **backslash `\`** at the end of the line to signal continuation to the interpreter.
* ⚠️ **Strict Rule**: No characters or spaces are permitted after the backslash `\` on the same line!


````python
# ============================================================
# 9️⃣ Line Continuation Methods
# ============================================================

# 1. Implicit Line Continuation using Parentheses (Recommended)
total_computation = (
    100 +
    250 +
    375 +
    500
)
print(f"Total (Parentheses Implicit): {total_computation}")

# Implicit Continuation in Lists and Dictionaries
pipeline_stages = [
    "1_Extract_From_S3",
    "2_Validate_Schema",
    "3_Transform_Silver_Layer",
    "4_Load_To_BigQuery"
]
print(f"Pipeline Stages List: {pipeline_stages}")

# 2. Explicit Line Continuation using Backslash (\)
explicit_sum = 10 + \
               20 + \
               30
print(f"Explicit Sum (Backslash): {explicit_sum}")
````

## 🔟 User Input Mechanics (`input()` Function & Prompting)

The `input()` function pauses program execution and waits for the user to provide text via standard input (`sys.stdin`).

### 📝 Syntax
$$\mathbf{\text{input}([prompt]) \rightarrow \text{str}}$$

* **`prompt` (optional)**: A string written to standard output without a trailing newline before reading input.
* 🚨 **THE GOLDEN RULE OF `input()`**:
  > **`input()` ALWAYS returns a value of type `str` (String).**
  > Even if the user enters digits like `42`, the return value is `"42"`.
  > You **must explicitly cast** it using `int()` or `float()` to perform arithmetic!


````python
# ============================================================
# 🔟 User Input & The String Concatenation Trap
# ============================================================

# Simulating user input stream for automated execution
simulated_user_stream = iter(["Rajan", "25", "50", "12.75"])
def mock_input(prompt=""):
    val = next(simulated_user_stream)
    print(f"{prompt}{val}")
    return val

# 1. Reading basic string input
user_name = mock_input("Enter your name: ")
print(f"Welcome, {user_name}! | Type: {type(user_name)}")

# 2. The String Concatenation Trap (Un-cast Input)
val1_str = mock_input("Enter first number: ")   # "25"
val2_str = mock_input("Enter second number: ")  # "50"

trap_result = val1_str + val2_str  # "25" + "50" -> "2550" (String Concatenation!)
print(f"Without Casting: '{val1_str}' + '{val2_str}' = '{trap_result}' (String Concatenation Trap!)")

# 3. Correct Numeric Input Parsing (Explicit Casting)
num1 = int(val1_str)
num2 = int(val2_str)
correct_sum = num1 + num2
print(f"With int() Casting: {num1} + {num2} = {correct_sum} (Mathematical Addition!)")

# 4. Reading Float Input
price = float(mock_input("Enter item price: "))
print(f"Item Price: ${price:.2f} | Type: {type(price)}")
````

## 1️⃣1️⃣ Type Casting & Conversion (Implicit Coercion vs Explicit Constructors)

**Type Casting (Type Conversion)** is the process of converting an object from one data type into another.

```
                                  TYPE CASTING IN PYTHON
                                             │
                     ┌───────────────────────┴───────────────────────┐
                     ▼                                               ▼
         [IMPLICIT TYPE CASTING]                         [EXPLICIT TYPE CASTING]
          (Automatic by Python)                           (Manual by Programmer)
                     │                                               │
      Prevents data loss automatically              Uses built-in constructor functions:
      int ──> float ──> complex                     int(), float(), str(), bool(),
                                                    list(), tuple(), set(), dict()
```

---

### 1. Implicit Type Casting (Coercion)
Python automatically converts smaller data types to larger/wider data types during mixed operations to prevent loss of precision.
* `int + float` $\rightarrow$ `float` (e.g. `10 + 2.5 = 12.5`)
* `int + complex` $\rightarrow$ `complex` (e.g. `5 + (2+3j) = (7+3j)`)
* *Note: Python NEVER converts complex numbers down to float or int implicitly.*

---

### 2. Explicit Type Casting (Constructors)

| Function | Target Type | Behavior / Rules | Example |
|---|---|---|---|
| `int(x)` | `int` | **Truncates** decimals towards 0 (does NOT round!). String must contain integer digits only. | `int(10.9) -> 10`<br>`int("123") -> 123` |
| `float(x)` | `float` | Converts integers or decimal numeric strings to double-precision floats. | `float(5) -> 5.0`<br>`float("45.67") -> 45.67` |
| `str(x)` | `str` | Converts any object to its textual string representation. | `str(100) -> "100"` |
| `bool(x)` | `bool` | Converts to `True` or `False` based on truthiness rules. | `bool(0) -> False`<br>`bool("Data") -> True` |
| `list(x)` | `list` | Converts iterables (string, tuple, set) into an ordered mutable list. | `list("ABC") -> ['A', 'B', 'C']` |
| `tuple(x)` | `tuple` | Converts iterables into an immutable tuple. | `tuple([1, 2]) -> (1, 2)` |
| `set(x)` | `set` | Converts iterables into an unordered collection of unique elements. | `set([1, 2, 2, 3]) -> {1, 2, 3}` |

---

### ⚠️ Common Type Casting Errors & Invalid Literal Traps
1. **`int("12.5")` raises `ValueError`**: `int()` cannot parse strings containing decimal points directly. You must do **`int(float("12.5"))`**!
2. **`int("abc")` raises `ValueError`**: Non-numeric strings cannot be cast to numbers.
3. **`int()` Truncates, Does Not Round**: `int(9.9)` produces `9`, not `10`. To round, use `round(9.9)`.


````python
# ============================================================
# 1️⃣1️⃣ Implicit Type Casting (Automatic Promotion)
# ============================================================

# Example 1: int + float -> float
int_val = 10
float_val = 2.5
implicit_result1 = int_val + float_val

print("--- Implicit Casting: int + float ---")
print(f"int_val: {int_val} ({type(int_val).__name__}) + float_val: {float_val} ({type(float_val).__name__})")
print(f"Result : {implicit_result1} | Type: {type(implicit_result1)}")

# Example 2: int + complex -> complex
complex_val = 2 + 3j
implicit_result2 = int_val + complex_val

print("\n--- Implicit Casting: int + complex ---")
print(f"int_val: {int_val} + complex_val: {complex_val}")
print(f"Result : {implicit_result2} | Type: {type(implicit_result2)}")
````

````python
# ============================================================
# 1️⃣1️⃣ Explicit Type Casting: int(), float(), str(), bool()
# ============================================================

print("--- 1. Explicit int() Casting ---")
print(f"int(10.9)    = {int(10.9)} (Truncated decimal, NOT rounded!)")
print(f"int('123')   = {int('123')}")
print(f"int(True)    = {int(True)}, int(False) = {int(False)}")

print("\n--- 2. Explicit float() Casting ---")
print(f"float(5)     = {float(5)}")
print(f"float('45.67') = {float('45.67')}")

print("\n--- 3. Explicit str() Casting ---")
age = 22
print("Concatenation with str(age): " + "I am " + str(age) + " years old.")

print("\n--- 4. Explicit bool() Casting (Truthy vs Falsy) ---")
# Falsy values evaluate to False
print(f"bool(0)     : {bool(0)}")
print(f"bool(0.0)   : {bool(0.0)}")
print(f"bool('')    : {bool('')}")
print(f"bool(None)  : {bool(None)}")
print(f"bool([])    : {bool([])}")
print(f"bool({{}})    : {bool({})}")

# Truthy values evaluate to True
print(f"bool(1)     : {bool(1)}")
print(f"bool(-5)    : {bool(-5)} (Non-zero integers are Truthy)")
print(f"bool('Data'): {bool('Data')}")
print(f"bool([10])  : {bool([10])}")
````

````python
# ============================================================
# 1️⃣1️⃣ Collection Type Casting & Invalid Conversion Traps
# ============================================================

print("--- 1. Collection Type Conversions ---")
# String to List (splits into characters)
name_chars = list("Rajan")
print(f"list('Rajan')       : {name_chars}")

# Tuple to List (make mutable)
tup = (10, 20, 30)
lst_from_tup = list(tup)
lst_from_tup.append(40)
print(f"list((10, 20, 30))  : {lst_from_tup}")

# List to Set (auto-deduplication)
raw_duplicates = [1, 2, 2, 3, 3, 3, 4]
unique_set = set(raw_duplicates)
print(f"set({raw_duplicates}) : {unique_set} (Deduplicated!)")


print("\n--- 2. Common Type Casting Traps & Fixes ---")

# Trap 1: int("12.5") fails because string has decimal point
try:
    bad_int = int("12.5")
except ValueError as e:
    print(f"Trap 1 Caught: {e}")
    # Correct fix: Parse to float first, then int!
    fixed_int = int(float("12.5"))
    print(f"Fix for '12.5' -> int(float('12.5')) = {fixed_int}")

# Trap 2: Invalid text string to int
try:
    bad_str = int("abc")
except ValueError as e:
    print(f"Trap 2 Caught: {e}")
````

## 1️⃣2️⃣ Python Execution Architecture (Source -> Bytecode -> PVM)

Python is both a **compiled and interpreted language**. The execution lifecycle proceeds through 3 structured phases:

```
 ┌───────────────────────┐
 │   Source Code (.py)   │  Human-readable Python source code
 └───────────┬───────────┘
             │  Stage 1: Lexical Analysis, Parsing & Syntax Verification
             ▼
 ┌───────────────────────┐
 │  Bytecode (.pyc)      │  Platform-independent intermediate instructions
 └───────────┬───────────┘
             │  Stage 2: Python Virtual Machine (PVM) Execution
             ▼
 ┌───────────────────────┐
 │ Line-by-Line Machine  │  PVM translates bytecode to native CPU instructions
 │   Execution (CPU)     │  and streams output to stdout/stderr
 └───────────────────────┘
```

| Execution Stage | Key Operations | Error Category Produced |
|---|---|---|
| **1. Compiler Phase** | Syntax checking, tokenization, AST generation, bytecode compilation (`.pyc`) | **`SyntaxError`** (Blocks execution entirely) |
| **2. PVM Interpreter** | Executes compiled bytecode instructions line-by-line | **`RuntimeError` / Exceptions** (Fails during execution) |


---
## 1️⃣3️⃣ Error Taxonomy: Syntax Errors vs Runtime Errors

Understanding the distinction between Syntax Errors and Runtime Errors is fundamental to debugging and robust error handling.

| Feature | Syntax Error (`SyntaxError`) | Runtime Error / Exception |
|---|---|---|
| **Root Cause** | Grammar, spelling, or structural violations | Invalid operations on valid syntax (e.g. dividing by zero) |
| **When Detected** | **Before execution** (during parsing stage) | **During execution** (when the bad line is reached) |
| **Code Execution** | **Zero code runs** (entire file blocked) | Partial execution runs **until** the error line |
| **Bytecode (`.pyc`)** | ❌ **NOT generated** | ✅ **IS generated** |
| **`try-except` Catchable** | ❌ **Cannot be caught** inside the same module | ✅ **Can be caught and handled gracefully** |

---

### 🏆 The Golden Rule of Python Execution
> **"If Python runs even ONE line of your code, the entire file is 100% syntactically valid."**
> A `SyntaxError` anywhere in the file prevents Python from starting execution.


````python
# ============================================================
# 1️⃣3️⃣ Runtime Errors & Exception Handling (try-except)
# ============================================================

# 1. ZeroDivisionError
try:
    divisor = 0
    res = 100 / divisor
except ZeroDivisionError as e:
    print(f"1. Caught ZeroDivisionError : {e}")

# 2. NameError (Accessing undefined variable)
try:
    print("undefined_pipeline_variable")
except NameError as e:
    print(f"2. Caught NameError          : {e}")

# 3. IndexError (List index out of bounds)
try:
    data_list = [10, 20, 30]
    bad_access = data_list[10]
except IndexError as e:
    print(f"3. Caught IndexError         : {e}")

# 4. TypeError (Invalid type operation)
try:
    invalid_concat = "Records: " + 500  # String + Int without casting
except TypeError as e:
    print(f"4. Caught TypeError          : {e}")

# 5. KeyError (Accessing missing dictionary key)
try:
    config = {"env": "production"}
    db_port = config["port"]
except KeyError as e:
    print(f"5. Caught KeyError           : {e}")
````

````python
# ============================================================
# 1️⃣4️⃣ Practical Data Engineering Stream Parser & Audit Logger
# ============================================================

def parse_and_log_raw_stream(raw_events):
    '''
    Demonstrates practical Data Engineering foundational concepts:
    - Multiple assignment
    - Type casting with error handling
    - Custom separator and file stream logging
    - Truthiness filtering
    '''
    valid_records = []
    
    with open("pipeline_execution_audit.log", "w", encoding="utf-8") as audit_log:
        print("RECORD_ID", "SENSOR_TAG", "METRIC_VAL", "STATUS", sep="\t", file=audit_log)
        
        for raw_record in raw_events:
            # 1. Multiple assignment unpacking
            event_id, sensor_tag, raw_metric = raw_record
            
            # 2. Defensive type casting with exception handling
            try:
                metric_clean = float(raw_metric)
                status = "VALID"
                valid_records.append((event_id, sensor_tag, metric_clean))
            except (ValueError, TypeError) as e:
                metric_clean = 0.0
                status = f"INVALID ({type(e).__name__})"
            
            # 3. Writing formatted audit row to file stream
            print(event_id, sensor_tag or "UNTAGGED", metric_clean, status, sep="\t", file=audit_log)
            
    return valid_records

# Sample incoming raw batch with mixed dirty records
raw_stream_batch = [
    (101, "TEMP_ZONE_A", "24.8"),
    (102, "PRESS_ZONE_B", "101.3"),
    (103, "", "BAD_VAL"),          # Missing tag and corrupt metric
    (104, "HUMID_ZONE_A", "68.2"),
    (105, "VIBR_ZONE_C", None)      # None metric
]

clean_data = parse_and_log_raw_stream(raw_stream_batch)
print("=== Clean Processed Records ===")
for rec in clean_data:
    print(rec)

print("\n=== Audit Log File Output ===")
with open("pipeline_execution_audit.log", "r", encoding="utf-8") as log_f:
    print(log_f.read())
````
