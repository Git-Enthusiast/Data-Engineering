# Session 5: Secure programming and testing
**Objective:** validate input, handle errors safely and test systematically. *(LO3, LO4)*

### 5.1 Input validation

```python
def parse_quantity(value):
    """Return a positive whole number or raise ValueError."""
    quantity = int(value)           # raises ValueError for text such as 'abc'
    if quantity <= 0:
        raise ValueError("Quantity must be positive.")
    return quantity

for text in ["3", "1", "abc", "0", "-2"]:
    try:
        print(repr(text), "->", parse_quantity(text))
    except ValueError as error:
        print(repr(text), "-> rejected:", error)
```

**Predict** which inputs are rejected, then run the cell.

### 5.2 Why eval() is unsafe
`eval()` runs text as Python code. Never pass untrusted user input to it. Parse numbers with `int()` or `float()` and choose from allowed operations instead.

### 5.3 Automate your test cases with assert

```python
def is_valid_quantity(value):
    return isinstance(value, int) and value >= 1

assert is_valid_quantity(3)         # normal
assert is_valid_quantity(1)         # boundary
assert not is_valid_quantity(0)     # zero
assert not is_valid_quantity(-2)    # negative
print("All tests passed")
```

### 6.1 The idea of rejecting bad input

```python
def add_task(tasks, title):
    title = title.strip()
    if not title:
        raise ValueError("Task title cannot be empty.")
    tasks.append({"title": title, "done": False})
    return tasks

tasks = []
add_task(tasks, "Buy milk")
try:
    add_task(tasks, "   ")
except ValueError as error:
    print("Rejected:", error)
print(tasks)
```
