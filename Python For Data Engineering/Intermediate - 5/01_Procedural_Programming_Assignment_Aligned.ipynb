## 1️⃣ What is a Programming Paradigm?

A programming paradigm is a way of organising and thinking about a program.

It answers a simple question:

**How should we arrange the instructions so the computer can solve the problem?**

Three paradigms in this unit are:

| Paradigm | Main idea |
|---|---|
| **Procedural** | Organise the solution as a sequence of steps and functions |
| **Object-Oriented** | Organise the solution around objects that contain data and behaviour |
| **Event-Driven** | Organise the program around events and the code that responds to them |

A paradigm is not a programming language. Python can be used with more than one paradigm.

```python
print("A paradigm is a way of organising a solution.")
```

## 2️⃣ Start with an Algorithm

Before writing Python code, describe the steps needed to solve the problem.

For example, to find the total cost of two items:

1. Read the first price.
2. Read the second price.
3. Add the two prices.
4. Display the total.

The algorithm describes **what needs to happen**. The Python code gives those steps a form the computer can execute.

```python
price1 = 20
price2 = 35
total = price1 + price2
print("Total:", total)
```

## 3️⃣ Why Procedural Programming Feels Like a Recipe

Imagine a recipe:

- get the ingredients
- prepare them
- cook them
- serve the food

A procedural program often feels similar. We write the work as a sequence of steps.

For a small calculation, this is very easy to follow.

```python
# A simple procedural program

price = 120
quantity = 3

total = price * quantity

print("Total cost:", total)
```

## 4️⃣ Sequence: One Step After Another

The most basic procedural flow is **sequence**.

Python executes the statements from top to bottom unless a condition, loop or function changes that flow.

```python
name = "Ravi"
score = 91

print("Name:", name)
print("Score:", score)
print("Status: Passed")
```

## 5️⃣ Variables Hold the Information We Need

A variable gives a value a name.

In a procedural program, variables often represent the current state of the calculation:

- the amount entered by the user
- a running total
- a counter
- a selected option

```python
athlete_name = "Ravi"
weekly_fee = 30
competition_count = 2

print(athlete_name)
print(weekly_fee)
print(competition_count)
```

## 6️⃣ Common Data Types

The type of data affects what we can do with it.

| Type | Example | Typical use |
|---|---|---|
| `str` | `"Ravi"` | names and text |
| `int` | `2` | counts |
| `float` | `22.50` | measurements and money |
| `bool` | `True` | yes/no decisions |

A procedural solution normally chooses data types that match the problem.

```python
athlete = "Ravi"       # str
competitions = 2      # int
private_fee = 19.0    # float
can_compete = True    # bool

print(type(athlete))
print(type(competitions))
print(type(private_fee))
print(type(can_compete))
```

## 7️⃣ Input and Output

Many small programs follow this pattern:

**input → process → output**

The user gives data, the program processes it, and a result is shown.

```python
# The values are written directly here so the example is easy to run
name = "Ravi"
weight = 74

print("Athlete:", name)
print("Weight:", weight, "kg")
```

## 8️⃣ Decisions with `if`

Programs often need to choose between different actions.

Procedural programs use control structures such as `if`, `elif` and `else` for these decisions.

```python
score = 67

if score >= 40:
    print("Pass")
else:
    print("Fail")
```

## 9️⃣ More than One Condition

A real program may need several choices.

The program still follows a clear sequence: test one condition, then another when needed.

```python
plan = "Intermediate"

if plan == "Beginner":
    weekly_fee = 25
elif plan == "Intermediate":
    weekly_fee = 30
elif plan == "Elite":
    weekly_fee = 35
else:
    weekly_fee = 0

print("Weekly fee:", weekly_fee)
```

## 🔟 Repetition with Loops

When the same action must happen many times, a loop saves us from writing the same statements repeatedly.

A `for` loop is useful when we want to process each item in a collection.

```python
athletes = ["Ravi", "Asha", "John"]

for athlete in athletes:
    print("Processing:", athlete)
```

## 1️⃣1️⃣ Counting with a Loop

A loop can also keep a running value.

This is a common procedural pattern when calculating totals or counting items.

```python
fees = [25, 30, 35]
total = 0

for fee in fees:
    total = total + fee

print("Total:", total)
```

## 1️⃣2️⃣ Functions Give the Program Named Steps

As a program grows, putting every instruction in one long block becomes difficult to manage.

A function gives a group of related instructions a name.

For example, instead of repeating the calculation for every athlete, we can write one function and call it when needed.

```python
def monthly_training_cost(weekly_fee):
    return weekly_fee * 4

print(monthly_training_cost(25))
print(monthly_training_cost(35))
```

## 1️⃣3️⃣ Parameters Let Functions Work with Different Data

A parameter is a value supplied to a function.

This makes a function reusable without copying the calculation.

```python
def competition_cost(number_of_competitions):
    return number_of_competitions * 22

print(competition_cost(0))
print(competition_cost(3))
```

## 1️⃣4️⃣ Returning a Result

A function can calculate something and return the result to the part of the program that called it.

```python
def total_cost(training_cost, competition_cost, coaching_cost):
    return training_cost + competition_cost + coaching_cost

monthly_total = total_cost(120, 44, 38)
print("Monthly total:", monthly_total)
```

## 1️⃣5️⃣ Functions Can Be Combined

Procedural programs are often built from a set of small functions.

Each function handles one part of the work, while the main flow decides when to call them.

```python
def training_cost(weekly_fee):
    return weekly_fee * 4

def competition_cost(count):
    return count * 22

def coaching_cost(hours):
    return hours * 9.50

def monthly_cost(weekly_fee, competitions, coaching_hours):
    return (
        training_cost(weekly_fee)
        + competition_cost(competitions)
        + coaching_cost(coaching_hours)
    )

print(monthly_cost(30, 2, 4))
```

## 1️⃣6️⃣ Scope: Where a Variable Exists

A variable created inside a function normally belongs to that function.

This helps keep the data used by one part of the program separate from unrelated parts.

```python
rate = 22

def show_cost(count):
    total = count * rate
    print(total)

show_cost(3)
print(rate)
```

## 1️⃣7️⃣ Lists and Dictionaries

Procedural programs still need data structures.

A list is useful when we have many items.
A dictionary is useful when we want to store related values using names.

```python
athletes = ["Ravi", "Asha", "John"]

fees = {
    "Beginner": 25,
    "Intermediate": 30,
    "Elite": 35
}

print(athletes)
print(fees["Intermediate"])
```

## 1️⃣8️⃣ A Simple Data Record

For a small procedural application, a dictionary can represent one athlete.

```python
athlete = {
    "name": "Ravi",
    "plan": "Intermediate",
    "weight": 74,
    "competitions": 2,
    "coaching_hours": 3
}

print(athlete["name"])
print(athlete["plan"])
```

## 1️⃣9️⃣ Validation Belongs in the Flow

A program should not blindly accept every value.

For example, a negative number of competitions does not make sense, so the program can reject it before doing the calculation.

```python
competition_count = -1

if competition_count < 0:
    print("Invalid number of competitions")
else:
    print("Accepted")
```

## 2️⃣0️⃣ A Small Procedural Application

Now the ideas come together.

The application below calculates a monthly cost for one athlete using:

- sequence
- selection
- functions
- data

```python
TRAINING_FEES = {
    "Beginner": 25.00,
    "Intermediate": 30.00,
    "Elite": 35.00
}

COMPETITION_FEE = 22.00
PRIVATE_COACHING_FEE = 9.50

def calculate_monthly_cost(plan, competitions, coaching_hours):
    training = TRAINING_FEES[plan] * 4
    competitions_cost = competitions * COMPETITION_FEE
    coaching = coaching_hours * PRIVATE_COACHING_FEE
    total = training + competitions_cost + coaching
    return training, competitions_cost, coaching, total

name = "Ravi"
plan = "Intermediate"
competitions = 2
coaching_hours = 3

training, competitions_cost, coaching, total = calculate_monthly_cost(
    plan, competitions, coaching_hours
)

print("Athlete:", name)
print("Training:", f"£{training:.2f}")
print("Competitions:", f"£{competitions_cost:.2f}")
print("Private coaching:", f"£{coaching:.2f}")
print("Total:", f"£{total:.2f}")
```

## 2️⃣1️⃣ Weight Category Check

The same procedural style can handle a different part of the problem.

```python
WEIGHT_LIMITS = {
    "Flyweight": 66,
    "Lightweight": 73,
    "Light-Middleweight": 81,
    "Middleweight": 90,
    "Light-Heavyweight": 100
}

def compare_weight(weight, category):
    if category == "Heavyweight":
        return "Within category"
    limit = WEIGHT_LIMITS[category]
    if weight <= limit:
        return "Within category"
    return f"Above category limit by {weight - limit:.1f} kg"

print(compare_weight(74, "Light-Middleweight"))
```

## 2️⃣2️⃣ Making the Program Handle More Athletes

The client wants more than six athletes.

A list of dictionaries gives a simple way to store multiple records in a procedural program.

```python
athletes = [
    {
        "name": "Ravi",
        "plan": "Intermediate",
        "competitions": 2,
        "coaching_hours": 3
    },
    {
        "name": "Asha",
        "plan": "Beginner",
        "competitions": 0,
        "coaching_hours": 1
    }
]

for athlete in athletes:
    print(athlete["name"], athlete["plan"])
```

## 2️⃣3️⃣ What Makes This Procedural?

Look back at the code.

The main structure is:

1. Store the data.
2. Call functions.
3. Make decisions.
4. Repeat work with loops.
5. Produce the result.

The program is organised around **steps and functions** rather than around objects or events.

## 2️⃣4️⃣ Strengths of Procedural Programming

Procedural programming is often a good fit when:

- the problem can be described as a clear sequence of steps
- the application is small or moderate in size
- the flow is mostly predictable
- calculations and data processing are the main work

It is easy to start with because the control flow is visible.

## 2️⃣5️⃣ Where Procedural Code Can Become Difficult

As an application grows, one large collection of functions and shared data can become harder to manage.

For example, if many functions depend on the same data, changing that data structure may require changes in many places.

This does not make procedural programming bad. It simply means that another way of organising the same problem may become more useful.

## 2️⃣7️⃣ A Procedural Code Checklist

When you look at source code, ask:

- Is the main flow a sequence of steps?
- Are functions used to group actions?
- Are loops and conditions driving the work?
- Is data passed into functions and processed there?

When the answer is mostly yes, you are probably looking at a procedural solution.
