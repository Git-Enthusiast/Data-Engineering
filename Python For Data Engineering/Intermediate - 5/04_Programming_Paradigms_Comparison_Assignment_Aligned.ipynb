## 1️⃣ Start with the Same Problem

Suppose we need a program that receives:

- athlete name
- training plan
- current weight
- competition category
- number of competitions
- private coaching hours

The program must calculate costs and report the athlete's position against the competition weight category.

The problem has not changed.

Only the way we organise the code changes.

## 2️⃣ Three Ways to Think About the Same Program

| Paradigm | Main question |
|---|---|
| Procedural | What steps should the program perform? |
| OOP | What objects exist, and what should each object know/do? |
| Event-Driven | What events can happen, and what should happen in response? |

The same business rules can appear in all three.

## 3️⃣ Procedural View

The procedural design might start with functions:

```text
read_input()
validate_input()
calculate_training_cost()
calculate_competition_cost()
calculate_coaching_cost()
compare_weight()
display_result()
```

The program is built around actions.

## 4️⃣ OOP View

The OOP design might start with objects:

```text
Athlete
TrainingPlan
CostCalculator
```

The program is built around things that have data and behaviour.

## 5️⃣ Event-Driven View

The event-driven design might start with events:

```text
athlete_registered
training_plan_selected
competition_added
calculation_requested
calculation_completed
```

The program is built around things that happen and the handlers that respond.

## 6️⃣ Procedural Example

Here is a small version of the monthly calculation.

````python
def monthly_cost(plan, competitions, coaching_hours):
    fees = {
        "Beginner": 25.0,
        "Intermediate": 30.0,
        "Elite": 35.0
    }

    training = fees[plan] * 4
    competition = competitions * 22.0
    coaching = coaching_hours * 9.5

    return training + competition + coaching

print(monthly_cost("Intermediate", 2, 3))
````

## 7️⃣ OOP Example

The same calculation can be placed inside an `Athlete` object.

````python
class Athlete:
    FEES = {
        "Beginner": 25.0,
        "Intermediate": 30.0,
        "Elite": 35.0
    }

    def __init__(self, name, plan, competitions, coaching_hours):
        self.name = name
        self.plan = plan
        self.competitions = competitions
        self.coaching_hours = coaching_hours

    def monthly_cost(self):
        return (
            self.FEES[self.plan] * 4
            + self.competitions * 22.0
            + self.coaching_hours * 9.5
        )

ravi = Athlete("Ravi", "Intermediate", 2, 3)
print(ravi.monthly_cost())
````

## 8️⃣ Event-Driven Example

The calculation can remain a normal function while an event tells the application that a calculation is needed.

This is a useful reminder:

**Event-driven describes the flow of control. It does not remove functions or objects.**

````python
subscriptions = {}

def subscribe(event_name, handler):
    subscriptions.setdefault(event_name, []).append(handler)

def publish(event):
    for handler in subscriptions.get(event["type"], []):
        handler(event)

def calculate_requested(event):
    total = monthly_cost(
        event["plan"],
        event["competitions"],
        event["coaching_hours"]
    )
    print("Calculated total:", total)

subscribe("calculation_requested", calculate_requested)

publish({
    "type": "calculation_requested",
    "plan": "Intermediate",
    "competitions": 2,
    "coaching_hours": 3
})
````

## 9️⃣ The Key Relationship Between the Paradigms

These paradigms are not three separate worlds.

A single application can contain:

- procedural functions for calculations
- OOP classes for domain objects
- event-driven code for user actions and notifications

This is common in real software because different parts of a problem may benefit from different structures.

## 🔟 Control Flow Comparison

| Procedural | OOP | Event-Driven |
|---|---|---|
| Program follows steps | Objects collaborate through methods | Events trigger handlers |
| Flow is usually explicit | Flow moves between object methods | Flow depends on what happens |
| Easy to trace in small programs | Clear structure for large domains | Natural for interaction and messages |

## 1️⃣1️⃣ Where Is the State?

**Procedural:** state is often stored in variables, lists or dictionaries.

**OOP:** state usually lives inside objects.

**Event-Driven:** state may exist in objects/variables while events carry information about changes.

The same data can exist in each style; the difference is where we choose to organise it.

## 1️⃣2️⃣ A Small Example of State

The score starts at 0 and changes after an action.

````python
score = 0

score = score + 10
print(score)

score = score + 5
print(score)
````

## 1️⃣3️⃣ The Same State in an Object

Now the score belongs to a player object.

````python
class Player:
    def __init__(self):
        self.score = 0

    def add_points(self, points):
        self.score += points

player = Player()
player.add_points(10)
player.add_points(5)

print(player.score)
````

## 1️⃣4️⃣ The Same Change as an Event

The event-driven version can react whenever points are added.

````python
subscriptions = []

def subscribe(handler):
    subscriptions.append(handler)

def publish(event):
    for handler in subscriptions:
        handler(event)

def show_score(event):
    print("Score changed to:", event["score"])

score = 0
subscribe(show_score)

score += 10
publish({"type": "score_changed", "score": score})

score += 5
publish({"type": "score_changed", "score": score})
````

## 1️⃣5️⃣ What Changes When Requirements Change?

Consider the requirement:

**“Only Intermediate and Elite athletes can enter competitions.”**

Procedural code might add a validation step in a function.

OOP code might put the rule in an `Athlete` method.

Event-driven code might react to a `competition_added` event and reject or report the action.

The business rule is the same.

````python
def can_enter_competition(plan):
    return plan in {"Intermediate", "Elite"}

for plan in ["Beginner", "Intermediate", "Elite"]:
    print(plan, can_enter_competition(plan))
````

## 1️⃣6️⃣ Comparing Code Structure

When comparing source code, look at **how the code is organised**, not just what output it produces.

Ask:

- Are there mainly functions and steps?
- Are there classes and objects?
- Are there event registrations and handlers?
- How is data passed around?
- Who controls what happens next?

## 1️⃣7️⃣ Comparing Reuse

Procedural reuse commonly comes from functions.

OOP reuse can come from classes, composition and inheritance.

Event-driven reuse often comes from reusable handlers and shared event infrastructure.

A good comparison explains the mechanism instead of simply saying “all three are reusable”.

## 1️⃣8️⃣ Comparing Coupling

Coupling is about how strongly one part of a program depends on another.

A direct function call is usually easy to trace.

Object collaboration can create clearer boundaries around data and behaviour.

Event-driven communication can reduce direct knowledge between components, but the event relationships can become harder to trace when the system is large.

## 1️⃣9️⃣ Comparing Where Each Fits

| Problem | Often a good starting point |
|---|---|
| Simple calculation | Procedural |
| Data-processing script | Procedural |
| Many domain entities | OOP |
| Desktop/web user interaction | Event-Driven + OOP |
| Notifications/messages | Event-Driven |
| Large business application | Often a hybrid |

## 2️⃣1️⃣ A Hybrid Judo Example

This example deliberately combines the three ideas.

- `Athlete` is OOP.
- `calculate_total()` is procedural.
- `calculation_requested` is event-driven.

````python
class Athlete:
    def __init__(self, name, plan, competitions, coaching_hours):
        self.name = name
        self.plan = plan
        self.competitions = competitions
        self.coaching_hours = coaching_hours

def calculate_total(plan, competitions, coaching_hours):
    fees = {"Beginner": 25.0, "Intermediate": 30.0, "Elite": 35.0}
    return (
        fees[plan] * 4
        + competitions * 22.0
        + coaching_hours * 9.5
    )

subscriptions = {}

def subscribe(event_name, handler):
    subscriptions.setdefault(event_name, []).append(handler)

def publish(event):
    for handler in subscriptions.get(event["type"], []):
        handler(event)

def show_result(event):
    athlete = event["athlete"]
    total = calculate_total(
        athlete.plan,
        athlete.competitions,
        athlete.coaching_hours
    )
    print(athlete.name, f"£{total:.2f}")

subscribe("calculation_requested", show_result)

ravi = Athlete("Ravi", "Intermediate", 2, 3)

publish({
    "type": "calculation_requested",
    "athlete": ravi
})
````

## 2️⃣2️⃣ How to Identify the Paradigm in Given Source Code

A source-code comparison can use three questions:

**Procedural:** Do I mainly see steps, functions, conditions and loops?

**OOP:** Do I see classes, objects, attributes and methods?

**Event-Driven:** Do I see events, handlers, callbacks, subscriptions or code reacting to actions?

More than one answer can be “yes”.

## 2️⃣3️⃣ Critical Evaluation

For a stronger evaluation, move beyond:

> “This code uses OOP because it has classes.”

A better explanation is:

> “The application uses classes to keep athlete data and athlete behaviour together. This makes each athlete easier to model and allows the system to manage many athlete objects consistently.”

The second statement explains **why the structure matters**.

## 2️⃣4️⃣ Evaluating the Judo Application

A strong analysis can discuss:

- why procedural functions are suitable for straightforward calculations
- why OOP is useful for modelling athletes and training plans
- why event-driven code is useful for reacting to application actions
- where each style makes the code simpler
- where each style introduces extra complexity
- why a hybrid design may be more practical than forcing one paradigm everywhere

## 2️⃣5️⃣ Choosing a Paradigm

When deciding how to organise a new program, consider:

| Question | What it suggests |
|---|---|
| Is the work mainly a clear sequence of calculations? | Procedural |
| Are there many important entities with their own data and behaviour? | OOP |
| Is the program mainly reacting to clicks, messages or other actions? | Event-Driven |
| Are different parts of the problem asking for different structures? | Hybrid |

## 2️⃣6️⃣ A Better Way to Explain a Choice

Instead of:

> “I used OOP because OOP is good for large programs.”

Explain the actual problem:

> “The application manages many athletes, and each athlete has data and behaviours such as cost calculation and weight checking. Modelling athletes as objects keeps related information together and makes the application easier to extend.”

## 3️⃣0️⃣ One Final Example

The output does not tell us which paradigm was used.

The structure of the source code does.

The three programs below could all produce the same result.

````python
# A short procedural solution

price = 30
weeks = 4
print(price * weeks)
````

### OOP structure

````python
class Plan:
    def __init__(self, price):
        self.price = price

    def monthly_cost(self):
        return self.price * 4

plan = Plan(30)
print(plan.monthly_cost())
````

### Event-driven structure

````python
listeners = []

def subscribe(handler):
    listeners.append(handler)

def publish(event):
    for handler in listeners:
        handler(event)

def calculate(event):
    print(event["price"] * event["weeks"])

subscribe(calculate)
publish({"type": "calculate", "price": 30, "weeks": 4})
````

## 3️⃣0️⃣ North Sussex Judo — One Application, Three Styles

The assignment gives us one problem and asks us to explain how different paradigms can be used in that application.

The client needs:

- athlete details
- training plan
- current weight and competition category
- number of competitions
- private coaching hours
- monthly itemised costs
- total monthly cost
- weight comparison

The business rules include:

| Rule | Value |
|---|---|
| Beginner | £25 per week |
| Intermediate | £30 per week |
| Elite | £35 per week |
| Private coaching | £9.50 per hour |
| Competition | £22 per competition |
| Private coaching limit | 5 hours per week |
| Competition plans | Intermediate and Elite only |
| Month | 4 weeks |

### Procedural part

The calculation rules can be written as ordinary functions. Each function receives data, performs a clear calculation, and returns a result.

````python
TRAINING_FEES = {
    "Beginner": 25.00,
    "Intermediate": 30.00,
    "Elite": 35.00
}

COMPETITION_FEE = 22.00
COACHING_FEE = 9.50

def training_cost(plan):
    return TRAINING_FEES[plan] * 4

def competition_cost(plan, number_of_competitions):
    if plan not in {"Intermediate", "Elite"}:
        return 0.0
    return number_of_competitions * COMPETITION_FEE

def coaching_cost(hours):
    return hours * COACHING_FEE

def monthly_cost(plan, competitions, coaching_hours):
    return (
        training_cost(plan)
        + competition_cost(plan, competitions)
        + coaching_cost(coaching_hours)
    )

print(f"£{monthly_cost('Intermediate', 2, 3):.2f}")
````

### OOP part

Now the same athlete data is kept inside an object.

The calculation can use the object's own information.

````python
class Athlete:
    def __init__(self, name, plan, weight, category, competitions, coaching_hours):
        self.name = name
        self.plan = plan
        self.weight = weight
        self.category = category
        self.competitions = competitions
        self.coaching_hours = coaching_hours

    def monthly_cost(self):
        return monthly_cost(
            self.plan,
            self.competitions,
            self.coaching_hours
        )

ravi = Athlete(
    "Ravi",
    "Intermediate",
    74,
    "Light-Middleweight",
    2,
    3
)

print(ravi.name)
print(f"£{ravi.monthly_cost():.2f}")
````

### Event-driven part

The application can also react to something happening.

For example, when a calculation is requested, an event can tell the relevant handler to calculate and display the result.

````python
subscriptions = {}

def subscribe(event_name, handler):
    subscriptions.setdefault(event_name, []).append(handler)

def publish(event):
    for handler in subscriptions.get(event["type"], []):
        handler(event)

def calculation_requested(event):
    athlete = event["athlete"]
    print(
        f"{athlete.name}: "
        f"monthly cost = £{athlete.monthly_cost():.2f}"
    )

subscribe("calculation_requested", calculation_requested)

publish({
    "type": "calculation_requested",
    "athlete": ravi
})
````

### What is happening in this example?

The same application contains three different ideas:

| Code | Paradigm |
|---|---|
| `training_cost()` and `monthly_cost()` | Procedural |
| `Athlete` and its methods | OOP |
| `subscribe()`, `publish()` and the handler | Event-Driven |

The paradigms are working together rather than competing with each other.

## 3️⃣1️⃣ How to Explain This in the Presentation

A good explanation can follow the structure of the source code:

**Procedural:** The calculation functions break the business rules into clear steps and are easy to trace.

**OOP:** The `Athlete` class groups athlete data with behaviour, making it easier to represent and manage multiple athletes.

**Event-Driven:** The event mechanism allows the application to react when a calculation is requested without putting the response directly into the code that creates the event.

Then evaluate the trade-off: the hybrid version is more flexible, but it also introduces more structure than a small procedural program would need.

## ✅ Final Takeaway

Do not choose a paradigm because one is “better” in every situation.

Start with the problem.

Ask how the program naturally behaves:

- **steps** → procedural
- **objects and responsibilities** → OOP
- **events and reactions** → event-driven
- **a mixture of these** → hybrid

The strongest programming decision is the one that makes the problem easier to understand, maintain and extend.
