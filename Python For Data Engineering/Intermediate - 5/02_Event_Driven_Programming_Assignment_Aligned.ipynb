## 1️⃣ What is an Event?

An **event** is something that happens and that a program may need to react to.

Examples:

- a user clicks a button
- a key is pressed
- a file finishes loading
- an order is placed
- a timer reaches a certain point

The important idea is simple:

**Something happens → the program responds.**

## 2️⃣ An Everyday Example

Think about a doorbell.

- Someone presses the button.
- That action is the event.
- The bell reacts to the event.

The person who presses the button does not need to know how the bell works.

Event-driven programming follows the same basic idea.

## 3️⃣ The Three Parts

A simple event-driven system usually has three important pieces:

| Part | Meaning |
|---|---|
| **Event** | Something happened |
| **Handler** | Code that knows how to respond |
| **Event source** | The thing that caused the event |

For a button, the source is the button, the event is the click, and the handler is the function that should run.

## 4️⃣ Procedural and Event-Driven Code Start Differently

In a procedural program, we normally think:

**run this step → then this step → then this step**

In an event-driven program, we think:

**wait for something to happen → respond to it**

That is the main difference students need to remember first.

## 5️⃣ A Tiny Event Example in Python

Python lets us create our own small event system so that the idea is visible without needing a graphical application.

```python
def handle_click():
    print("Button was clicked")

# Pretend this line represents the user clicking the button.
handle_click()
```

## 6️⃣ The Function Becomes a Handler

The function above is a **handler** because it contains the response to an event.

In a real user interface, a framework would call this function when the click happens.

Here we call it directly so the idea is easy to see.

## 7️⃣ Registering a Handler

An event-driven system needs a way to say:

**When event X happens, run function Y.**

A simple dictionary can store that connection.

```python
handlers = {}

def on(event_name, handler):
    handlers[event_name] = handler

def publish(event_name):
    handler = handlers.get(event_name)
    if handler:
        handler()

def show_message():
    print("Hello from the handler")

on("click", show_message)
publish("click")
```

## 8️⃣ One Event Can Have Different Responses

The same kind of event may cause different actions in different parts of an application.

```python
handlers = []

def subscribe(handler):
    handlers.append(handler)

def publish(event_name):
    for handler in handlers:
        handler(event_name)

def update_screen(event_name):
    print("Screen updated for:", event_name)

def write_log(event_name):
    print("Log entry:", event_name)

subscribe(update_screen)
subscribe(write_log)

publish("save_clicked")
```

## 9️⃣ Events Can Carry Data

An event is often more useful when it tells us what happened.

For example, an order event might include:

- order ID
- customer name
- total amount

```python
event = {
    "type": "order_placed",
    "order_id": 101,
    "customer": "Ravi",
    "total": 220.00
}

print(event)
```

## 🔟 A Handler Uses the Event Data

The handler can read the information carried by the event and decide what to do.

```python
def email_customer(event):
    print(
        "Email sent to",
        event["customer"],
        "for order",
        event["order_id"]
    )

email_customer(event)
```

## 1️⃣1️⃣ A Simple Event Dispatcher

Now the pieces can be connected:

**publish event → find handlers → call handlers → pass event data**

```python
subscriptions = {}

def subscribe(event_name, handler):
    subscriptions.setdefault(event_name, []).append(handler)

def publish(event):
    event_name = event["type"]
    for handler in subscriptions.get(event_name, []):
        handler(event)

def send_email(event):
    print("Email customer:", event["customer"])

def audit_order(event):
    print("Audit order:", event["order_id"])

subscribe("order_placed", send_email)
subscribe("order_placed", audit_order)

publish({
    "type": "order_placed",
    "order_id": 101,
    "customer": "Ravi"
})
```

## 1️⃣2️⃣ What the Dispatcher Is Doing

The dispatcher is a small piece of infrastructure.

Its job is not to calculate the order.

Its job is to connect:

**event name → interested handlers**

This separation is one of the useful ideas behind event-driven systems.

## 1️⃣3️⃣ User Interface Events

The framework waits for events and calls the registered code.

## 1️⃣4️⃣ A Simple Button Example

A real GUI framework has more code behind the scenes, but the structure is still easy to understand.

```python
def save_clicked():
    print("Saving the athlete record...")

button_handlers = {
    "save": save_clicked
}

# Simulated event from the UI
event_name = "save"

button_handlers[event_name]()
```

## 1️⃣5️⃣ Event-Driven Does Not Mean “No Functions” or “No Classes”

This is important.

Event-driven programming describes **how the program flow is organised**.

The handlers can be ordinary functions.
The application can also contain classes and objects.

That is why real applications often use a **hybrid** approach.

## 1️⃣6️⃣ A Short Comparison

| Question | Procedural | Event-Driven |
|---|---|---|
| Who decides what runs next? | The program's step-by-step flow | The next event |
| Typical starting point | Main sequence | Event source / event loop |
| Common examples | Calculations, scripts | GUIs, notifications, UI actions |
| Main idea | Do these steps | React when something happens |

## 1️⃣7️⃣ Events in the North Sussex Judo Application

Even though the client did not require a GUI, we can still imagine events such as:

- `athlete_registered`
- `athlete_updated`
- `competition_added`
- `calculation_completed`

The calculation itself can remain procedural or be handled by an object.

The event part is about **what happens when an important action occurs**.

## 1️⃣8️⃣ A Small Judo Event Example

Here the calculation is ordinary Python, while notifications are handled through events.

```python
subscriptions = {}

def subscribe(event_name, handler):
    subscriptions.setdefault(event_name, []).append(handler)

def publish(event):
    for handler in subscriptions.get(event["type"], []):
        handler(event)

def send_confirmation(event):
    print("Confirmation for:", event["athlete"])

def write_audit(event):
    print("Audit:", event["type"], event["athlete"])

subscribe("athlete_registered", send_confirmation)
subscribe("athlete_registered", write_audit)

publish({
    "type": "athlete_registered",
    "athlete": "Ravi"
})
```

## 1️⃣9️⃣ When Event-Driven Programming Is Useful

Event-driven programming is especially useful when the program spends a lot of its time **waiting for actions or messages**.

Examples include:

- desktop and web interfaces
- games
- sensors
- notification systems
- messaging systems

The program reacts instead of following one fixed sequence from start to finish.

## 2️⃣0️⃣ When It Can Be More Difficult

Event-driven code can be harder to follow when many events and handlers interact.

A programmer may need to ask:

- Who creates this event?
- Who is listening for it?
- Which handler runs first?
- What happens if a handler fails?

For a beginner, the most useful idea is still simply:

**event happens → handler responds**

## 2️⃣1️⃣ Recognising Event-Driven Code

When you are given source code, look for clues such as:

- event names
- callbacks
- handler functions
- subscription or registration code
- dispatchers
- code that waits for user actions or messages
- a loop that waits for events

These clues help you explain why the program is event-driven.

## 2️⃣2️⃣ A Small Source-Code Reading Example

Read this code as a story:

1. A handler is registered.
2. An event is created.
3. The dispatcher looks for handlers.
4. The handler runs.

That is event-driven control flow.

```python
listeners = []

def subscribe(handler):
    listeners.append(handler)

def publish(event):
    for handler in listeners:
        handler(event)

def show_alert(event):
    print("ALERT:", event["message"])

subscribe(show_alert)

publish({
    "type": "alert",
    "message": "Registration completed"
})
```

## ✅ Quick Summary

An event-driven program is organised around **events**.

The basic pattern is:

**event source → event → handler**

The event says **what happened**.
The handler says **what the program should do about it**.

Procedural programming gives us a step-by-step flow.
Event-driven programming changes the next action according to what happens.

That is enough foundation to compare the three paradigms in the final notebook.
