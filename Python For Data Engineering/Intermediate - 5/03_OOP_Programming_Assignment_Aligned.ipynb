## 1️⃣ What is Object-Oriented Programming?

Object-Oriented Programming, or **OOP**, organises a program around **objects**.

An object brings related data and actions together.

For example, an athlete has data such as:

- name
- training plan
- weight

and can have actions such as:

- calculate monthly cost
- check weight category
- display details

Instead of keeping all of these ideas separate, OOP groups related data and behaviour together.

## 2️⃣ Think About a Real Object

A real athlete is one useful way to think about this.

The athlete has **information** and can take part in **actions**.

In OOP, an object models something important in the problem.

## 3️⃣ A Class Is a Blueprint

A **class** describes what an object should contain and what it can do.

The class is the blueprint.
The object is one actual item created from that blueprint.

````python
class Athlete:
    pass

ravi = Athlete()
asha = Athlete()

print(type(ravi))
print(type(asha))
````

## 4️⃣ Attributes Store Object Data

Attributes are values stored on an object.

We can use them to represent the state of one athlete.

````python
class Athlete:
    pass

ravi = Athlete()
ravi.name = "Ravi"
ravi.plan = "Intermediate"
ravi.weight = 74

print(ravi.name)
print(ravi.plan)
print(ravi.weight)
````

## 5️⃣ The Constructor `__init__`

Instead of creating the attributes one by one, a class can define how a new object should start.

The `__init__` method runs when the object is created.

````python
class Athlete:
    def __init__(self, name, plan, weight):
        self.name = name
        self.plan = plan
        self.weight = weight

ravi = Athlete("Ravi", "Intermediate", 74)

print(ravi.name)
print(ravi.plan)
print(ravi.weight)
````

## 6️⃣ What Does `self` Mean?

`self` means **this object**.

When we write:

```python
self.name = name
```

we are saying:

“Store this name inside the object we are currently working with.”

## 7️⃣ Different Objects Can Hold Different Data

The same class can create many different objects.

````python
class Athlete:
    def __init__(self, name, plan):
        self.name = name
        self.plan = plan

ravi = Athlete("Ravi", "Intermediate")
asha = Athlete("Asha", "Beginner")

print(ravi.name, ravi.plan)
print(asha.name, asha.plan)
````

## 8️⃣ Methods Represent Behaviour

A method is a function defined inside a class.

It describes something the object can do.

````python
class Athlete:
    def __init__(self, name, plan):
        self.name = name
        self.plan = plan

    def describe(self):
        print(self.name, "is on the", self.plan, "plan")

ravi = Athlete("Ravi", "Intermediate")
ravi.describe()
````

## 9️⃣ Data and Behaviour Stay Together

This is one of the biggest differences from a simple procedural design.

In a procedural program, we may have:

- an athlete dictionary
- separate functions that receive the dictionary

In OOP, the object can hold the data and provide the behaviour through its methods.

````python
class Athlete:
    def __init__(self, name, weight):
        self.name = name
        self.weight = weight

    def show_weight(self):
        print(self.name, "weighs", self.weight, "kg")

ravi = Athlete("Ravi", 74)
ravi.show_weight()
````

## 🔟 Calculating a Monthly Cost with a Method

The Judo application gives us a good example of object behaviour.

````python
class Athlete:
    def __init__(self, name, plan, competitions, coaching_hours):
        self.name = name
        self.plan = plan
        self.competitions = competitions
        self.coaching_hours = coaching_hours

    def monthly_cost(self):
        training_fees = {
            "Beginner": 25.00,
            "Intermediate": 30.00,
            "Elite": 35.00
        }

        training = training_fees[self.plan] * 4
        competitions = self.competitions * 22.00
        coaching = self.coaching_hours * 9.50

        return training + competitions + coaching

ravi = Athlete("Ravi", "Intermediate", 2, 3)
print(f"£{ravi.monthly_cost():.2f}")
````

## 1️⃣1️⃣ Encapsulation

Encapsulation means keeping related data and operations together and controlling how the data is changed.

In Python, the language does not force strict private fields like some languages, but we can still design classes so that the public interface is clear.

````python
class Athlete:
    def __init__(self, name):
        self._name = name

    def get_name(self):
        return self._name

ravi = Athlete("Ravi")
print(ravi.get_name())
````

## 1️⃣2️⃣ Properties Give Controlled Access

A property lets a class expose data while still controlling validation.

````python
class Athlete:
    def __init__(self, weight):
        self.weight = weight

    @property
    def weight(self):
        return self._weight

    @weight.setter
    def weight(self, value):
        if value <= 0:
            raise ValueError("Weight must be greater than 0")
        self._weight = value

ravi = Athlete(74)
print(ravi.weight)
````

## 1️⃣3️⃣ One Class, Many Objects

A real application normally works with a collection of objects.

````python
class Athlete:
    def __init__(self, name, plan):
        self.name = name
        self.plan = plan

athletes = [
    Athlete("Ravi", "Intermediate"),
    Athlete("Asha", "Beginner"),
    Athlete("John", "Elite")
]

for athlete in athletes:
    print(athlete.name, athlete.plan)
````

## 1️⃣4️⃣ Composition

Objects can contain other objects.

For example, an athlete may have a `TrainingPlan` object rather than storing every detail directly on the athlete.

````python
class TrainingPlan:
    def __init__(self, name, weekly_fee):
        self.name = name
        self.weekly_fee = weekly_fee

class Athlete:
    def __init__(self, name, plan):
        self.name = name
        self.plan = plan

plan = TrainingPlan("Intermediate", 30)
ravi = Athlete("Ravi", plan)

print(ravi.name)
print(ravi.plan.name)
print(ravi.plan.weekly_fee)
````

## 1️⃣5️⃣ Inheritance

Inheritance allows one class to reuse or specialise another class.

The child class starts with the behaviour of the parent class and can add or change behaviour.

It is useful when there is a genuine **is-a** relationship.

````python
class Person:
    def __init__(self, name):
        self.name = name

    def describe(self):
        print("Name:", self.name)

class Athlete(Person):
    def compete(self):
        print(self.name, "is competing")

ravi = Athlete("Ravi")
ravi.describe()
ravi.compete()
````

## 1️⃣6️⃣ Polymorphism

Polymorphism means that different objects can respond to the same method name in their own way.

This lets code work with different object types through a common interface.

````python
class Dog:
    def speak(self):
        return "Woof"

class Cat:
    def speak(self):
        return "Meow"

animals = [Dog(), Cat()]

for animal in animals:
    print(animal.speak())
````

## 1️⃣7️⃣ Why OOP Can Help Large Applications

OOP can make a large system easier to divide into meaningful parts.

For the Judo application, we might have:

- `Athlete`
- `TrainingPlan`
- `Competition`
- `CostCalculator`

Each class can have a clear responsibility.

The application then becomes a set of objects that work together.

## 1️⃣8️⃣ OOP and the Judo Application

A possible model is:

```text
Athlete
 ├── name
 ├── weight
 ├── training_plan
 ├── competitions
 └── calculate_monthly_cost()

TrainingPlan
 ├── name
 └── weekly_fee
```

This is a different way of thinking from “write the steps first”.

## 1️⃣9️⃣ A Small OOP Version of the Application

The class now keeps the data and calculation together.

````python
class Athlete:
    TRAINING_FEES = {
        "Beginner": 25.00,
        "Intermediate": 30.00,
        "Elite": 35.00
    }

    def __init__(self, name, plan, weight, competitions, coaching_hours):
        self.name = name
        self.plan = plan
        self.weight = weight
        self.competitions = competitions
        self.coaching_hours = coaching_hours

    def monthly_cost(self):
        training = self.TRAINING_FEES[self.plan] * 4
        competitions = self.competitions * 22.00
        coaching = self.coaching_hours * 9.50
        return training + competitions + coaching

ravi = Athlete("Ravi", "Intermediate", 74, 2, 3)
print(ravi.name)
print(f"£{ravi.monthly_cost():.2f}")
````

## 2️⃣0️⃣ When OOP Is Useful

OOP is often useful when:

- the system contains many important entities
- each entity has its own data and behaviour
- the same kind of entity appears many times
- the application is expected to grow
- responsibilities need to be separated clearly

Examples include:

- banking systems
- booking systems
- games
- customer management systems
- inventory systems

## 2️⃣1️⃣ When OOP May Be More Than We Need

A very small script that adds two numbers does not need a hierarchy of classes.

OOP introduces structure, and structure is useful when it solves a real problem.

For a tiny calculation, procedural code may be simpler.

````python
def add(a, b):
    return a + b

print(add(2, 3))
````

## 2️⃣2️⃣ Recognising OOP in Source Code

Look for clues such as:

- `class`
- objects created from classes
- `self`
- `__init__`
- methods
- inheritance
- objects passed between other objects

These features suggest that the code is organised around objects.

## 2️⃣3️⃣ Procedural and OOP Can Work Together

An OOP program can still contain ordinary functions.

A program does not have to be “100% OOP” to use object-oriented ideas.

Real systems often combine styles when that gives a clearer solution.
