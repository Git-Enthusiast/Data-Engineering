# 🟦 Level 01 — Files & File-System Fundamentals

Before using `open("data.txt")`, understand **files, directories, paths, current working directory, file types, text vs binary data, and a sensible data-science project structure**.


## 1.1 What Is a File?

A file is persistent data stored and managed by the operating system. Python asks the OS to access the file and works through a file object.

Conceptual flow:

**Application → Operating System → File System → Storage**

Typical analytics files include `.csv`, `.json`, `.xlsx`, `.parquet`, `.pkl`, `.txt`, `.py`, and `.ipynb`.


````python
from pathlib import Path

p = LAB / "data" / "file_concept.txt"
p.write_text("Hello, Data Science!\n", encoding="utf-8")

print("Path:", p)
print("Exists:", p.exists())
print("Size:", p.stat().st_size, "bytes")
print("Raw bytes:", p.read_bytes())
````

## 1.2 Directories

A directory organizes files and other directories.

Important terms: **root, parent, child, current, nested**.


````python
for p in sorted(LAB.rglob("*")):
    print(("DIR " if p.is_dir() else "FILE"), p.relative_to(LAB))
````

## 1.3 Absolute Paths

An absolute path is fully qualified from the filesystem root, for example:

- Windows: `C:\Users\Rajan\project\data\sales.csv`
- POSIX: `/home/rajan/project/data/sales.csv`

It does not depend on the CWD.


````python
from pathlib import Path

absolute = (LAB / "data" / "raw" / "sales.csv").resolve()
print(absolute)
print("Anchor/root:", absolute.anchor)
````

## 1.4 Relative Paths

A relative path is interpreted from the current working directory.

Example:

`data/raw/sales.csv`

Relative paths are convenient for project-based code, but only when the execution context is controlled.


````python
from pathlib import Path

relative = Path("data") / "raw" / "sales.csv"
print("Relative:", relative)
print("Resolved from this CWD:", relative.resolve())
````

## 1.7 Text vs Binary

**Text:** characters are decoded using an encoding such as UTF-8.

**Binary:** raw bytes whose interpretation is defined by a format.

Text examples: `.txt`, `.csv`, `.json`, `.xml`, `.py`

Binary examples: `.png`, `.jpg`, `.pdf`, `.parquet`, `.pkl`

Encoding matters for text; exact byte positions matter for binary data.


````python
text_path = LAB / "data" / "text_vs_binary.txt"
text_path.write_text("café — ₹100\n", encoding="utf-8")

with open(text_path, "r", encoding="utf-8") as f:
    text = f.read()

with open(text_path, "rb") as f:
    raw = f.read()

print("Text:", text, type(text).__name__)
print("Bytes:", raw, type(raw).__name__)
````

## 1.14 Symbolic Links

A symbolic link points to another path. `is_symlink()` detects a symlink and `resolve()` follows it. Symlink creation can depend on OS permissions.


````python
target = LAB / "symlink_target.txt"
link = LAB / "symlink_demo.txt"

target.write_text("target", encoding="utf-8")

try:
    if link.exists() or link.is_symlink():
        link.unlink()
    link.symlink_to(target)
    print("Is symlink:", link.is_symlink())
    print("Resolved:", link.resolve())
    print("Read through link:", link.read_text(encoding="utf-8"))
except (OSError, NotImplementedError) as exc:
    print("Symlink unavailable:", exc)
````

# 🟩 Level 02 — `open()` & File Objects


## 2.1 `open()` Returns a File Object

`open()` returns a **handle/file object**, not the contents.

Flow:

`open() → file object → read/write/seek → close`


````python
p = LAB / "open_demo.txt"
p.write_text("Hello from open()\n", encoding="utf-8")

f = open(p, "r", encoding="utf-8")
print(f)
print("Type:", type(f))
print(f.tell())
print("Content:", repr(f.read()))
print(f"Content: {f.read()}")
f.close()
````

## 2.2 File Object Properties

Common properties are `name`, `mode`, `closed`, and `encoding`. The object also maintains a current position.


````python
p = LAB / "properties.txt"
p.write_text("Rajan\nData Analytics\n", encoding="utf-8")

f = open(p, "r", encoding="utf-8")
print("name:", f.name)
print("mode:", f.mode)
print("closed:", f.closed)
print("encoding:", f.encoding)
print("position:", f.tell())
f.close()
print("closed after close:", f.closed)
````

## 2.3 Full `open()` Syntax

Key parameters are:

`file`, `mode`, `buffering`, `encoding`, `errors`, `newline`, `closefd`, `opener`.

Daily analytics code usually relies most on `file`, `mode`, `encoding`, and sometimes `errors/newline`.


````python
import inspect
print(inspect.signature(open))
````

## 2.4 File Lifecycle

**Open → Use → Close**

Opening acquires a resource, use performs I/O, and closing releases the resource and flushes buffered output.


````python
p = LAB / "lifecycle.txt"

f = open(p, "w", encoding="utf-8")
print("Closed after open:", f.closed)
f.write("lifecycle example\n")
print("Position:", f.tell())
f.close()
print("Closed after close:", f.closed)
````

## 2.5 Why Closing Matters

File resources are finite. Repeatedly opening files without cleanup can exhaust file descriptors/handles.

The professional default is `with open(...) as f:`.


````python
handles = []
paths = []

for i in range(10):
    p = LAB / f"small_{i}.txt"
    p.write_text(str(i), encoding="utf-8")
    paths.append(p)
    handles.append(open(p, encoding="utf-8"))

print("Open handles:", len(handles))
for f in handles:
    f.close()

print("All closed:", all(f.closed for f in handles))
````

## 2.6 File Object Types

Text mode commonly yields `TextIOWrapper`. Binary mode commonly yields a buffered binary object.


````python
p = LAB / "object_types.txt"
p.write_text("hello", encoding="utf-8")

with open(p, "r", encoding="utf-8") as a:
    print("Text:", type(a).__name__)

with open(p, "rb") as b:
    print("Binary:", type(b).__name__)
````

## 2.7 Buffering

`buffering=-1` uses the system default. `0` means unbuffered binary I/O. `1` can mean line-buffered in suitable text contexts. Values greater than 1 specify a buffer size.

Normally, leave the default unless you have a performance/streaming reason.


````python
p = LAB / "buffering.txt"
p.write_text("buffer example", encoding="utf-8")

with open(p, "r", encoding="utf-8", buffering=-1) as f:
    print("Text object:", type(f).__name__)
    print("Buffered layer:", type(f.buffer).__name__)
````

## 2.8 `fileno()`

`f.fileno()` returns the OS-level integer file descriptor for supported real files.


````python
p = LAB / "fileno.txt"
p.write_text("descriptor", encoding="utf-8")

with open(p, encoding="utf-8") as f:
    print("File descriptor:", f.fileno())
````

## 2.9 `closefd`

When wrapping an existing descriptor, `closefd=False` allows the Python wrapper to close without closing the underlying descriptor.


````python
import os

p = LAB / "closefd.txt"
p.write_text("descriptor example", encoding="utf-8")

fd = os.open(p, os.O_RDONLY)
try:
    wrapper = os.fdopen(fd, "r", encoding="utf-8", closefd=False)
    print(wrapper.read())
    wrapper.close()
    print("Descriptor still usable:", os.fstat(fd).st_size, "bytes")
finally:
    os.close(fd)
````

# 🟨 Level 03 — File Modes

Know modes deeply: they determine whether files are read, replaced, appended, exclusively created, opened as binary, or opened for update.


## 3.1 `r`, `w`, `a`, `x`

| Mode | Meaning | Existing file? |
|---|---|---|
| `r` | read | must exist |
| `w` | write/truncate | creates or replaces |
| `a` | append | creates if missing |
| `x` | exclusive create | fails if it exists |

`w` is destructive because truncation happens when the file is opened.


````python
p = LAB / "modes.txt"
p.write_text("ORIGINAL\n", encoding="utf-8")

with open(p, "r", encoding="utf-8") as f:
    print("r:", repr(f.read()))

with open(p, "w", encoding="utf-8") as f:
    f.write("NEW\n")
print("w:", repr(p.read_text(encoding="utf-8")))

with open(p, "a", encoding="utf-8") as f:
    f.write("APPEND\n")
print("a:", repr(p.read_text(encoding="utf-8")))

x = LAB / "exclusive.txt"
if x.exists():
    x.unlink()
with open(x, "x", encoding="utf-8") as f:
    f.write("created once\n")

try:
    open(x, "x", encoding="utf-8")
except FileExistsError:
    print("x correctly raised FileExistsError")
````

## 3.2 Binary Modes

`rb`, `wb`, and `ab` work with bytes rather than decoded strings.


````python
p = LAB / "modes.bin"

with open(p, "wb") as f:
    f.write(b"ABC")

with open(p, "ab") as f:
    f.write(b"XYZ")

with open(p, "rb") as f:
    print(f.read())
````

## 3.3 Update Modes

- `r+` — read/write, must exist, no truncation
- `w+` — read/write, truncates
- `a+` — read/write, appends writes
- binary versions combine `b` as well


````python
p = LAB / "update_modes.txt"
p.write_text("ABCDE", encoding="utf-8")

with open(p, "r+", encoding="utf-8") as f:
    print("r+ read:", f.read())
    f.seek(0)
    f.write("XYZ")
print("r+ result:", p.read_text(encoding="utf-8"))

with open(p, "w+", encoding="utf-8") as f:
    f.write("12345")
    f.seek(0)
    print("w+ result:", f.read())

with open(p, "a+", encoding="utf-8") as f:
    f.write("678")
    f.seek(0)
    print("a+ result:", f.read())
````

## 3.4 Text Is the Default

`open(path, "r")` is equivalent to `open(path, "rt")` for text reading.


````python
p = LAB / "text_default.txt"
p.write_text("text", encoding="utf-8")

with open(p, "r", encoding="utf-8") as a, open(p, "rt", encoding="utf-8") as b:
    print(type(a).__name__, repr(a.read()))
    print(type(b).__name__, repr(b.read()))
````

## 3.5 Mode Selection Mental Model

Ask:

**read? write? append? create only? binary? both read/write?**

This simple checklist prevents many file-mode mistakes.


````python
mode_map = {
    "read": "r",
    "replace": "w",
    "append": "a",
    "exclusive create": "x",
    "read/write": "r+",
    "replace/read/write": "w+",
    "append/read": "a+",
    "binary read": "rb",
    "binary write": "wb",
}
for purpose, mode in mode_map.items():
    print(f"{purpose:22} -> {mode}")
````

# 🟧 Level 04 — Reading Files


## 4.1 `read()`

`read()` returns the remaining content as a string in text mode. `read(n)` limits the amount read.


````python
p = LAB / "read_methods.txt"
p.write_text("ABCDEFGHIJ\nKLMNOPQRST\n", encoding="utf-8")

with open(p, encoding="utf-8") as f:
    print("First 5:", repr(f.read(5)))
    print("Remaining:", repr(f.read()))
````

## 4.2 `readline()`

Reads one line, usually including its newline. Each call advances the file pointer.


````python
p = LAB / "readline.txt"
p.write_text("Line 1\nLine 2\nLine 3\n", encoding="utf-8")

with open(p, encoding="utf-8") as f:
    print(repr(f.readline()))
    print(repr(f.readline()))
    print(repr(f.readline()))
    print("EOF:", repr(f.readline()))
````

## 4.3 `readlines()`

Returns a list of the remaining lines. It loads all those lines into memory.


````python
p = LAB / "readlines.txt"
p.write_text("A\nB\nC\n", encoding="utf-8")

with open(p, encoding="utf-8") as f:
    lines = f.readlines()

print(lines)
print(type(lines).__name__)
````

## 4.4 Iterating Over a File

`for line in f:` is memory-efficient and is the preferred pattern for many large line-oriented files.


````python
p = LAB / "iteration.log"
p.write_text("INFO start\nWARNING slow\nERROR failed\nINFO end\n", encoding="utf-8")

with open(p, encoding="utf-8") as f:
    for number, line in enumerate(f, start=1):
        print(number, line.rstrip("\n"))
````

## 4.5 Memory Impact

`read()` and `readlines()` materialize large amounts of content in memory. Iteration streams records and is much safer for large logs.


````python
p = LAB / "memory_patterns.txt"
p.write_text("\n".join(f"record-{i}" for i in range(100)) + "\n", encoding="utf-8")

with open(p, encoding="utf-8") as f:
    all_text = f.read()

with open(p, encoding="utf-8") as f:
    streamed_count = sum(1 for _ in f)

print("read() result type:", type(all_text).__name__)
print("Streamed line count:", streamed_count)
````

## 4.6 `strip()` vs `rstrip("\n")`

`strip()` removes leading and trailing whitespace. `rstrip("\n")` removes only the newline character.

For data where indentation matters, the narrower operation is safer.


````python
line = "  indented value  \n"

print("Original:", repr(line))
print("strip():", repr(line.strip()))
print("rstrip('\\n'):", repr(line.rstrip("\n")))
````

## 4.7 Reading Exact Bytes

Binary mode is appropriate when you need exact byte ranges or headers.


````python
p = LAB / "header.bin"
p.write_bytes(bytes(range(32)))

with open(p, "rb") as f:
    header = f.read(16)

print(header)
print(header.hex())
````

## 4.8 Empty File Behavior

For an empty file:

- `read()` → `''`
- `readline()` → `''`
- `readlines()` → `[]`


````python
p = LAB / "empty.txt"
p.write_bytes(b"")

with open(p, encoding="utf-8") as f:
    print("read():", repr(f.read()))
    f.seek(0)
    print("readline():", repr(f.readline()))
    f.seek(0)
    print("readlines():", f.readlines())
````

## 4.9 Large-File Safe Pattern

Process a line immediately rather than accumulating the entire file.


````python
p = LAB / "large_style.log"
p.write_text("INFO\nERROR bad row\nINFO\nERROR bad input\n", encoding="utf-8")

errors = []
with open(p, encoding="utf-8") as f:
    for line in f:
        if "ERROR" in line:
            errors.append(line.rstrip("\n"))

print("Errors:", errors)
````

## 4.10 Manual Chunk Reading

Fixed-size chunking is useful when data is not naturally record-oriented.


````python
p = LAB / "chunk_text.txt"
p.write_text("ABCDEFGHIJKLMNOPQRSTUVWXYZ", encoding="utf-8")

chunks = []
with open(p, encoding="utf-8") as f:
    while True:
        chunk = f.read(5)
        if not chunk:
            break
        chunks.append(chunk)

print(chunks)
print("Reconstructed:", "".join(chunks))
````

## 4.11 Explicit `readline()` Loop

This pattern provides explicit EOF control.


````python
p = LAB / "readline_loop.txt"
p.write_text("Ravi\nAmit\nNeha\n", encoding="utf-8")

with open(p, encoding="utf-8") as f:
    while True:
        line = f.readline()
        if not line:
            break
        print(line.rstrip("\n"))
````

## 4.12 `splitlines()`

`splitlines()` separates a text block into lines without retaining the line-ending characters.


````python
data = "A\r\nB\nC\rD"
print(data.splitlines())
````

## 4.13 Parsing After Reading

You can split simple controlled text, but **do not use `split(",")` as a general CSV parser** because quoted commas and other CSV rules require a real CSV parser.


````python
data = "101,Ravi,5000\n102,Amit,7000\n"
for line in data.splitlines():
    print(line.split(","))
````

# 🟥 Level 05 — Writing & Appending Files


## 5.1 `write()`

`write()` accepts a string in text mode and returns the number of characters written. It does not automatically append a newline.


````python
p = LAB / "write.txt"
with open(p, "w", encoding="utf-8") as f:
    print("Returned:", f.write("Hello\n"))
    print("Returned:", f.write("World\n"))
print(p.read_text(encoding="utf-8"))
````

## 5.2 Writing Numbers

Convert non-string values with `str()` or use formatted strings.


````python
p = LAB / "write_numbers.txt"
value = 100

with open(p, "w", encoding="utf-8") as f:
    f.write(str(value))
    f.write("\n")
    f.write(f"score={value}\n")

print(p.read_text(encoding="utf-8"))
````

## 5.3 `writelines()`

`writelines()` writes an iterable of strings but **does not add newlines**.


````python
p = LAB / "writelines.txt"
lines = ["Alice\n", "Bob\n", "Charlie\n"]

with open(p, "w", encoding="utf-8") as f:
    f.writelines(lines)

print(repr(p.read_text(encoding="utf-8")))
````

## 5.4 Overwrite vs Append

`w` replaces the old content. `a` preserves old content and adds new content to the end.


````python
p = LAB / "overwrite_vs_append.txt"
p.write_text("old\n", encoding="utf-8")

with open(p, "w", encoding="utf-8") as f:
    f.write("new\n")
print("After w:", repr(p.read_text(encoding="utf-8")))

with open(p, "a", encoding="utf-8") as f:
    f.write("extra\n")
print("After a:", repr(p.read_text(encoding="utf-8")))
````

## 5.5 Append for Logs

Append mode is useful for incremental records, logs and simple audit information.

# 🟪 Level 06 — Context Managers


## 6.1 The `with` Statement

`with` automatically closes the file when the block ends, including when an exception occurs.

This should be the default pattern for file I/O.

## 6.2 Multiple Files

## 6.3 Cleanup on Errors

The context manager protocol ensures cleanup even when the body raises.


````python
p = LAB / "cleanup.txt"
p.write_text("cleanup", encoding="utf-8")
ref = None

try:
    with open(p, encoding="utf-8") as f:
        ref = f
        raise RuntimeError("simulated failure")
except RuntimeError as exc:
    print("Caught:", exc)

print("Closed after exception:", ref.closed)
````

## 6.4 Context Manager Protocol

A context manager implements `__enter__()` and `__exit__()`.

`__exit__()` can inspect an exception and can optionally suppress it by returning `True`.


````python
class DemoContext:
    def __enter__(self):
        print("enter")
        return self

    def __exit__(self, exc_type, exc_value, traceback):
        print("exit")
        print("Exception:", exc_type.__name__ if exc_type else None)
        return False

with DemoContext():
    print("body")
````

## 6.5 `ExitStack`

`ExitStack` is useful when the number of context-managed resources is dynamic.


````python
from contextlib import ExitStack

paths = [LAB / "stack_a.txt", LAB / "stack_b.txt", LAB / "stack_c.txt"]
for p in paths:
    p.write_text(p.stem, encoding="utf-8")

with ExitStack() as stack:
    handles = [stack.enter_context(open(p, encoding="utf-8")) for p in paths]
    print([h.read() for h in handles])

print("All closed:", all(h.closed for h in handles))
````

# 🟫 Level 07 — File Pointer / Random Access


## 7.1 `tell()`

`tell()` reports the current position in the stream. For text streams, treat positions as values intended for compatible `seek()` operations because encodings/newline translation add complexity.


````python
p = LAB / "tell.txt"
p.write_text("ABCDEFGHIJ", encoding="utf-8")

with open(p, encoding="utf-8") as f:
    print("Start:", f.tell())
    print(f.read(5))
    print("After read:", f.tell())
````

## 7.2 `seek()`

`seek(0)` returns to the beginning. Saved positions from `tell()` can be reused.


````python
p = LAB / "seek.txt"
p.write_text("ABCDEFGHIJ", encoding="utf-8")

with open(p, encoding="utf-8") as f:
    f.read(3)
    saved = f.tell()
    print("Saved:", saved)
    print("Next:", f.read(2))
    f.seek(saved)
    print("Replay:", f.read(2))
````

## 7.3 Binary Random Access

Binary streams are ideal for exact byte offsets.


````python
p = LAB / "random.bin"
p.write_bytes(bytes(range(256)))

with open(p, "rb") as f:
    f.seek(100)
    data = f.read(20)

print(list(data))
````

## 7.4 `whence`

The main reference positions are:

- `0` — beginning
- `1` — current position
- `2` — end

Binary streams are the clearest place to reason about relative byte offsets.


````python
p = LAB / "whence.bin"
p.write_bytes(bytes(range(50)))

with open(p, "rb") as f:
    f.seek(0, 2)
    print("End:", f.tell())
    f.seek(-10, 2)
    print("Tail:", list(f.read()))
````

## 7.5 Tail Reading

Reading the tail of a file is a useful operational pattern, but for encoded text avoid starting in the middle of a multibyte character.


````python
p = LAB / "tail.log"
p.write_text("line-1\nline-2\nline-3\n", encoding="utf-8")

with open(p, "rb") as f:
    f.seek(-7, 2)
    tail = f.read()

print(tail)
print(tail.decode("utf-8"))
````

## 7.6 `r+` Overwrites Rather Than Inserts

Writing at a position replaces existing content at that position; it does not shift the remainder like a text editor insertion.


````python
p = LAB / "rplus.txt"
p.write_text("0123456789", encoding="utf-8")

with open(p, "r+", encoding="utf-8") as f:
    f.seek(4)
    f.write("ABC")

print(p.read_text(encoding="utf-8"))
````

## 9.3 Specific Exception Handlers

Different failures require different responses. Prefer targeted handlers instead of `except: pass`.


````python
def safe_read(path):
    try:
        with open(path, encoding="utf-8") as f:
            return {"ok": True, "data": f.read(), "error": None}
    except FileNotFoundError as exc:
        return {"ok": False, "data": None, "error": f"missing: {exc}"}
    except PermissionError as exc:
        return {"ok": False, "data": None, "error": f"permission: {exc}"}
    except UnicodeDecodeError as exc:
        return {"ok": False, "data": None, "error": f"encoding: {exc}"}

print(safe_read(LAB / "utf8.txt"))
print(safe_read(LAB / "definitely_missing.txt"))
````

## 9.4 Professional Validation Flow

A useful workflow is:

**Exists? → Accessible? → Decodable? → Valid? → Processable?**


````python
from pathlib import Path

def validate_file(path):
    path = Path(path)
    return {
        "exists": path.exists(),
        "is_file": path.is_file(),
        "size_bytes": path.stat().st_size if path.exists() else None,
    }

print(validate_file(LAB / "utf8.txt"))
print(validate_file(LAB / "missing.txt"))
````

## 9.5 Logging File Errors

Logging creates an audit trail and is preferable to relying only on interactive `print()` statements.


````python
import logging

logger = logging.getLogger("file_demo")
if not logger.handlers:
    logger.addHandler(logging.StreamHandler())
logger.setLevel(logging.INFO)

try:
    open(LAB / "missing_logging.csv", encoding="utf-8")
except FileNotFoundError as exc:
    logger.error("Missing input: %s", exc)
````

## 9.6 Retry Pattern

Retries can make sense for transient access failures, but not for permanent conditions such as a wrong path.


````python
import time

def read_with_retry(reader, attempts=3, delay=0.05):
    last_error = None
    for attempt in range(1, attempts + 1):
        try:
            return reader()
        except PermissionError as exc:
            last_error = exc
            time.sleep(delay)
    raise RuntimeError("Could not read after retries") from last_error

print(read_with_retry(lambda: "success"))
````

## 15.31 End-to-End CSV Workflow

**File exists → read → validate schema → handle encoding → handle missing values → set dtypes → parse dates → clean/transform → analyze → export**


````python
from pathlib import Path
import pandas as pd

input_path = sales_path
output_path = LAB / "data" / "processed" / "end_to_end_sales.csv"

if not input_path.exists() or not input_path.is_file():
    raise FileNotFoundError(input_path)

pipeline_df = pd.read_csv(
    input_path,
    encoding="utf-8",
    dtype={"customer_id": "string"},
    parse_dates=["order_date"],
)

required = {
    "customer_id", "name", "order_date",
    "region", "quantity", "price"
}
missing = required - set(pipeline_df.columns)
if missing:
    raise ValueError(f"Missing: {sorted(missing)}")

pipeline_df["name"] = pipeline_df["name"].str.strip()
pipeline_df["region"] = pipeline_df["region"].str.strip()
pipeline_df["sales_value"] = pipeline_df["quantity"] * pipeline_df["price"]

pipeline_df.to_csv(output_path, index=False, encoding="utf-8")

print(pipeline_df)
print("Saved:", output_path)
````

````python
import csv
from io import StringIO

sample = "id|name|comment\n1|Ravi|Senior analyst\n2|Neha|Works in R&D\n"
dialect = csv.Sniffer().sniff(sample, delimiters=",;|\t")
header = csv.Sniffer().has_header(sample)

print("Delimiter:", repr(dialect.delimiter))
print("Quotechar:", repr(dialect.quotechar))
print("Header likely:", header)
print(list(csv.reader(StringIO(sample), dialect=dialect)))
````

## Project 10 — End-to-End Pipeline


````python
import logging
import time
import pandas as pd

root = RAW / "project10"
outdir = PROCESSED / "project10"
root.mkdir(exist_ok=True)
outdir.mkdir(parents=True, exist_ok=True)

(root / "a.csv").write_text("customer_id,name,amount\n001,Ravi,1000\n002,Neha,1500\n", encoding="utf-8")
(root / "b.csv").write_text("customer_id,name,amount\n003,Amit,1750\n004,Zoe,900\n", encoding="utf-8")

logger = logging.getLogger("project10")
if not logger.handlers:
    logger.addHandler(logging.StreamHandler())
logger.setLevel(logging.INFO)

required = {"customer_id", "name", "amount"}
frames = []

for p in sorted(root.glob("*.csv")):
    start = time.perf_counter()
    df = pd.read_csv(p, encoding="utf-8", dtype={"customer_id": "string"})
    missing = required - set(df.columns)
    if missing:
        raise ValueError(f"{p.name}: missing {sorted(missing)}")

    df["name"] = df["name"].str.strip()
    df["amount"] = pd.to_numeric(df["amount"], errors="raise")
    frames.append(df)

    elapsed = (time.perf_counter() - start) * 1000
    logger.info("READ_OK file=%s rows=%d bytes=%d elapsed_ms=%.2f",
                p.name, len(df), p.stat().st_size, elapsed)

result = pd.concat(frames, ignore_index=True)
final = outdir / "sales_cleaned.csv"
tmp = final.with_suffix(".tmp")
result.to_csv(tmp, index=False, encoding="utf-8")
tmp.replace(final)

print(result)
print("Saved:", final)
````
