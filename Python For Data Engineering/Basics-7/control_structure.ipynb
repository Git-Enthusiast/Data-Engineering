## 1️⃣ Introduction to Python Control Structures

A **control structure** is a programming construct that dictates the execution order and control flow of program statements. By default, the Python Virtual Machine (PVM) executes instructions sequentially from top to bottom. Control structures allow us to alter this default flow through branching decisions and repetitive loops.

```
                         PYTHON CONTROL STRUCTURE TAXONOMY
                                         │
        ┌────────────────────────────────┼────────────────────────────────┐
        ▼                                ▼                                ▼
[SEQUENTIAL CONTROL]            [SELECTION CONTROL]             [ITERATIVE CONTROL]
(Default Top-to-Bottom)         (Branching & Decisions)         (Repetition & Loops)
  • Line-by-line execution        • if Statement                  • for Loop (Sequences)
  • Predictable linear flow       • if-else Statement             • while Loop (Condition)
                                  • if-elif-else Ladder           • range() Generator
                                  • Ternary Operator              • enumerate() Indexing
                                  • match-case (Python 3.10+)     • Nested Loops
                                                                  • break, continue, pass, else
```

### 📊 Summary Comparison of the 3 Types

| Control Structure | Execution Behavior | Keywords / Constructs | Primary Use Case |
|---|---|---|---|
| **Sequential** | Statements execute strictly in order of appearance | Default line-by-line flow | Linear processing, setups, I/O logging |
| **Selection** | Executes specific blocks based on boolean conditions | `if`, `elif`, `else`, `match`, `case` | Decision making, validation, routing |
| **Iterative** | Repeats a block multiple times while a condition holds | `for`, `while`, `range()`, `enumerate()` | Batch processing, collections, polling |


````python
# ============================================================
# 1️⃣ Sequential vs Selection vs Iterative Flow Overview
# ============================================================

print("--- 1. Sequential Control (Line-by-Line) ---")
step1 = "Step 1: Ingest Raw Payload"
step2 = "Step 2: Transform Data Schema"
step3 = "Step 3: Load to Data Warehouse"
print(step1)
print(step2)
print(step3)

print("\n--- 2. Selection Control (Branching Decision) ---")
record_count = 1500
if record_count > 1000:
    print(f"High-volume batch detected: {record_count} records -> Route to Spark cluster")
else:
    print(f"Low-volume batch: {record_count} records -> Process in-memory with Pandas")

print("\n--- 3. Iterative Control (Repetitive Execution) ---")
retry_count = 0
while retry_count < 3:
    retry_count += 1
    print(f"Ping ingestion server... Attempt #{retry_count} [OK]")
````

## 2️⃣ Selection Control: `if`, `if-else`, `if-elif-else` & Complex Conditions

Selection statements evaluate a test condition to a boolean (`True` or `False`). Based on the result, control branches to a dedicated code block.

```
       [if condition]
             │
      True ┌─┴─┐ False
   ┌───────┘   └────────┐
   ▼                    ▼
[Execute if-block]  [Execute else-block (or check elif)]
   │                    │
   └───────┬────────────┘
           ▼
[Continue program execution]
```

### 📋 The Decision Hierarchy
1. **Simple `if`**: Executes code only if the condition evaluates to `True`.
2. **`if-else`**: Provides an alternative fallback block when the condition evaluates to `False`.
3. **`if-elif-else` Ladder**: Evaluates multiple conditions sequentially from top to bottom.
   * **The First-Match Rule**: Once a condition evaluates to `True`, its corresponding block executes, and **the entire remaining ladder is immediately skipped**.
   * The `else` block is optional and acts as a default catch-all if none of the preceding conditions are satisfied.
4. **Nested `if-else`**: An `if` statement placed inside another `if` or `else` block, enabling multi-tier filtering.


````python
# ============================================================
# 2️⃣ Selection Statements: if, if-else, if-elif-else & Nested
# ============================================================

# 1. Simple if Statement
threshold_gb = 50.0
current_storage_gb = 62.5

print("--- 1. Simple if ---")
if current_storage_gb > threshold_gb:
    print(f"⚠️ Storage Alert: Current {current_storage_gb}GB exceeds threshold {threshold_gb}GB!")

# 2. if-else Statement
user_role = "data_engineer"
print("\n--- 2. if-else ---")
if user_role == "admin" or user_role == "data_engineer":
    print(f"Access GRANTED: Role '{user_role}' has warehouse write permissions.")
else:
    print(f"Access DENIED: Role '{user_role}' has read-only permissions.")

# 3. if-elif-else Ladder (SLA Response Time Classifier)
latency_ms = 185
print("\n--- 3. if-elif-else Ladder ---")
if latency_ms < 50:
    sla_tier = "EXCELLENT (Tier 1)"
elif latency_ms < 150:
    sla_tier = "ACCEPTABLE (Tier 2)"
elif latency_ms < 300:
    sla_tier = "WARNING (Tier 3 - Degraded)"
else:
    sla_tier = "CRITICAL (SLA Breach)"
print(f"Latency: {latency_ms}ms | Performance Tier: {sla_tier}")

# 4. Compound Conditions using Logical Operators (and, or, not)
pipeline_active = True
records_ingested = 45000
error_rate = 0.005  # 0.5%

print("\n--- 4. Compound Logical Conditions ---")
if pipeline_active and records_ingested >= 10000 and not (error_rate > 0.02):
    print("✅ Pipeline Health Check: PASSED (Active, High Throughput, Low Error Rate)")
else:
    print("❌ Pipeline Health Check: FAILED (Action Required)")

# 5. Nested if-else Statement (Multi-Level Security & Role Check)
auth_token_valid = True
user_clearance = 3

print("\n--- 5. Nested if-else ---")
if auth_token_valid:
    print("Auth Token: VALID")
    if user_clearance >= 3:
        print("Clearance Level 3: Full Production Database Access Granted")
    else:
        print("Clearance Level < 3: Staging Access Only")
else:
    print("Auth Token: INVALID -> Terminating Session")
````

## 3️⃣ Truth Value Testing (Truthy vs Falsy) & Short-Circuiting

In Python, every object has an inherent boolean value when evaluated in a conditional context (`if`, `while`, or `bool()`). You do not need to explicitly compare against `True`, `False`, `None`, or check `len(x) > 0`.

### 📊 Comprehensive Truth Value Matrix

| Category | **Falsy Values** (Evaluate to `False`) | **Truthy Values** (Evaluate to `True`) |
|---|---|---|
| **Constants** | `None`, `False` | `True` |
| **Numeric Zero** | `0`, `0.0`, `0j`, `Decimal(0)`, `Fraction(0, 1)` | Any non-zero number: `1`, `-5`, `0.0001`, `-3.14` |
| **Empty Sequences** | `""` (Empty string), `[]` (Empty list), `()` (Empty tuple), `range(0)` | Non-empty sequences: `"Data"`, `[0]`, `(False,)`, `range(1)` |
| **Empty Mappings / Sets** | `{}` (Empty dict), `set()` (Empty set), `frozenset()` | Non-empty mappings/sets: `{"k": 0}`, `{0}` |

> 💡 **Data Engineering Best Practice**: Write idiomatic Python!
> * ✅ `if not records:` instead of `if len(records) == 0:`
> * ✅ `if config_name:` instead of `if config_name != "":`
> * ✅ `if connection is None:` instead of `if connection == None:`

---

### ⚡ Short-Circuit Evaluation Mechanics
* **`A and B`**: If `A` is falsy, Python returns `A` immediately without ever evaluating `B` (because the entire expression cannot be true).
* **`A or B`**: If `A` is truthy, Python returns `A` immediately without ever evaluating `B` (because the condition is already satisfied).


````python
# ============================================================
# 3️⃣ Truth Value Testing & Short-Circuit Evaluation
# ============================================================

# 1. Falsy Values Demonstration
falsy_samples = [None, False, 0, 0.0, "", [], (), {}, set(), range(0)]

print("--- 1. Testing Built-in Falsy Objects ---")
for item in falsy_samples:
    print(f"Value: {repr(item):<12} | bool(item): {bool(item)}")

# 2. Truthy Values Demonstration (Note: Negative numbers and collections with 0 are Truthy!)
truthy_samples = [True, 1, -10, 0.001, " ", "0", [0], (None,), {"key": False}, {0}]

print("\n--- 2. Testing Truthy Objects (Common Pitfalls) ---")
for item in truthy_samples:
    print(f"Value: {repr(item):<16} | bool(item): {bool(item)}")

# 3. Defensive Ingestion Cleanliness using Truthiness
print("\n--- 3. Defensive Data Ingestion Validation ---")
raw_data_stream = [
    {"source_id": "SRC_01", "payload": "sensor_temp_24C", "schema_tag": "v1"},
    {"source_id": "SRC_02", "payload": "", "schema_tag": "v1"},                # Empty payload (Falsy)
    {"source_id": "SRC_03", "payload": "press_101kPa", "schema_tag": None},      # Missing schema tag (Falsy)
    {"source_id": "SRC_04", "payload": "humid_65pct", "schema_tag": "v2"}
]

for record in raw_data_stream:
    # Clean, Pythonic truthy guard
    if record["payload"] and record["schema_tag"]:
        print(f"✅ Ingesting Valid Event: {record['source_id']} ({record['payload']})")
    else:
        print(f"⚠️ Dropping Malformed Event: {record['source_id']} (Missing payload or schema tag)")

# 4. Short-Circuit Guard: Safe dictionary access without KeyError
config = {"host": "localhost"}
# If "port" is not in config, the second check is never evaluated, avoiding KeyError!
port = "port" in config and config["port"]
print(f"\nSafe Port Lookup: {port}")
````

## 4️⃣ Ternary Operator (Conditional Expressions) & One-Liners

Python provides a concise expression syntax for conditional branching known as the **Ternary Operator** (introduced in PEP 308).

### 📝 Syntax
$$\mathbf{\text{result} = \text{[value\_if\_true]} \;\mathbf{\text{if}}\; \text{[condition]} \;\mathbf{\text{else}}\; \text{[value\_if\_false]}}$$

```
                      TERNARY OPERATOR EVALUATION FLOW
                                     │
                             [Evaluate Condition]
                                     │
                             True ┌──┴──┐ False
                        ┌─────────┘     └─────────┐
                        ▼                         ▼
             [Return value_if_true]    [Return value_if_false]
```

### 🎯 Key Properties & Best Practices
* **Returns a Value**: Unlike a standard `if-else` statement block, a ternary is an **expression**, meaning it can be assigned directly to variables, used inside formatted f-strings, or passed as function arguments.
* **Chaining / Nesting**: Multiple ternary operators can be chained (`val1 if cond1 else val2 if cond2 else val3`), but excessive nesting should be avoided to preserve readability.


````python
# ============================================================
# 4️⃣ Ternary Operator & One-Liner Patterns
# ============================================================

# 1. Basic Ternary Assignment
row_count = 85000
batch_type = "LARGE_BATCH" if row_count >= 50000 else "SMALL_BATCH"
print("--- 1. Basic Ternary Expression ---")
print(f"Rows: {row_count} -> Classification: {batch_type}")

# 2. Ternary Inside String Formatting & Print
server_status_code = 200
print("\n--- 2. Inline Ternary in String Output ---")
print(f"Health Status: {'ONLINE ✅' if server_status_code == 200 else 'OFFLINE ❌'}")

# 3. Nested / Chained Ternary Operator
throughput_rps = 4200
cluster_size = (
    "4X-Large (16 Nodes)" if throughput_rps > 10000 else
    "2X-Large (8 Nodes)"  if throughput_rps > 5000 else
    "Standard (4 Nodes)"  if throughput_rps > 1000 else
    "Micro (1 Node)"
)

print("\n--- 3. Chained Ternary Scale Calculator ---")
print(f"Throughput: {throughput_rps} RPS -> Recommended Cluster: {cluster_size}")

# 4. Ternary in List Comprehensions (Transformation)
raw_temperatures_celsius = [24.5, -999.0, 26.1, -999.0, 25.8]  # -999.0 indicates sensor error
cleaned_temperatures = [temp if temp != -999.0 else None for temp in raw_temperatures_celsius]

print("\n--- 4. Ternary in Data Transformation ---")
print(f"Raw    : {raw_temperatures_celsius}")
print(f"Cleaned: {cleaned_temperatures}")
````

## 5️⃣ Structural Pattern Matching: Python 3.10+ `match-case` Masterclass

Introduced in **Python 3.10 (PEP 634/635/636)**, `match-case` provides **Structural Pattern Matching**. It is far more expressive and powerful than a simple C/Java `switch-case` because it can match shapes, unpack nested structures, perform type checks, and bind variables simultaneously.

```
                            MATCH-CASE ARCHITECTURE
                                       │
                             [match <expression>:]
                                       │
                ┌──────────────────────┼──────────────────────┐
                ▼                      ▼                      ▼
         [case <pattern1>:]     [case <pattern2>:]      [case _:]
           Exact Value Match      Sequence/Dict Match    Default Catch-All
           OR Pattern (|)         Guards (if cond)       (Runs if no match)
           Variable Binding       Class / Type Match
```

### 📊 Structural Pattern Matching Capabilities

| Feature | Syntax / Pattern | Explanation / Behavior |
|---|---|---|
| **Literal Match** | `case 200:` / `case "SUCCESS":` | Exact scalar equality match |
| **OR Pattern** | `case "sat" \| "sun":` | Matches if any alternative matches (`\|` operator) |
| **Variable Capture** | `case x:` | Binds matched object to variable `x` (matches everything!) |
| **Wildcard Default** | `case _:` | Catch-all wildcard equivalent to `else` (does not bind variable) |
| **Conditional Guard** | `case x if x > 100:` | Matches pattern **AND** evaluates secondary boolean condition |
| **Sequence Unpacking**| `case [first, *rest]:` | Unpacks list/tuple and captures remaining items with `*` |
| **Dictionary Matching**| `case {"role": "admin", "id": uid}:`| Matches dictionary keys & binds values (**extra keys allowed!**) |
| **Type Checking** | `case int() as n:` / `case str():`| Verifies object type like `isinstance()` |
| **Class / Object** | `case Event(tag="DB", status=s):` | Matches user-defined class attributes |

---

### 🚨 Critical `match-case` Execution Rules
1. **Top-to-Down Evaluation**: Patterns are tested in the exact order written.
2. **First Match Wins**: The first successful match executes, and the rest are ignored.
3. **No Fall-Through**: Unlike C/C++/Java, Python **never** falls through to the next case. **No `break` statements are needed or allowed!**
4. **`case _:` Must Be Last**: Placing a wildcard `case _:` or a raw variable `case x:` before other cases will swallow all inputs and make subsequent cases unreachable.
5. **Dictionary Matching Allows Extra Keys**: A dictionary pattern checks for the **presence** of specified keys; having extra keys in the dictionary will **not** cause a mismatch.


````python
# ============================================================
# 5️⃣ match-case: Value, OR Patterns, Guards & Variable Binding
# ============================================================

# 1. Literal Value Matching & OR (|) Patterns
def parse_http_status(status_code):
    match status_code:
        case 200 | 201:
            return "SUCCESS: Request completed successfully."
        case 301 | 302:
            return "REDIRECTION: Resource moved."
        case 400 | 401 | 403 | 404:
            return "CLIENT_ERROR: Check credentials or resource URI."
        case 500 | 502 | 503:
            return "SERVER_ERROR: Downstream service unavailable."
        case _:
            return f"UNKNOWN_STATUS: Unrecognized code {status_code}."

print("--- 1. HTTP Status Code Router ---")
print(f"200 -> {parse_http_status(200)}")
print(f"404 -> {parse_http_status(404)}")
print(f"503 -> {parse_http_status(503)}")
print(f"999 -> {parse_http_status(999)}")

# 2. Variable Binding & Conditional Guards (case ... if ...)
def evaluate_sensor_telemetry(reading):
    match reading:
        case temp if temp < 0:
            return f"Sub-zero freezing alert! Temp: {temp}°C"
        case temp if 0 <= temp <= 35:
            return f"Nominal operating range. Temp: {temp}°C"
        case temp if 35 < temp <= 75:
            return f"Warning: Elevated thermal load! Temp: {temp}°C"
        case temp:
            return f"🚨 CRITICAL DANGER: Meltdown risk! Temp: {temp}°C"

print("\n--- 2. Guards & Variable Binding ---")
print(evaluate_sensor_telemetry(-5.2))
print(evaluate_sensor_telemetry(24.0))
print(evaluate_sensor_telemetry(48.5))
print(evaluate_sensor_telemetry(95.0))
````

````python
# ============================================================
# 5️⃣ match-case: Sequence Unpacking, Dictionaries & Classes
# ============================================================

# 1. Sequence Matching & Starred Unpacking ([first, *rest])
def process_pipeline_command(command):
    match command:
        case ["INGEST", source]:
            return f"Initiating single-source ingestion from '{source}'"
        case ["INGEST", source, *options]:
            return f"Initiating ingestion from '{source}' with flags: {options}"
        case ["TRANSFORM", "SILVER", table]:
            return f"Applying Silver layer transformation on table '{table}'"
        case ["EXPORT", target, format_type]:
            return f"Exporting dataset to '{target}' in '{format_type}' format"
        case _:
            return f"Invalid command syntax: {command}"

print("--- 1. Sequence & Command Unpacking ---")
print(process_pipeline_command(["INGEST", "s3://raw-lake"]))
print(process_pipeline_command(["INGEST", "s3://raw-lake", "--dedup", "--gzip", "--threads=8"]))
print(process_pipeline_command(["TRANSFORM", "SILVER", "dim_customers"]))
print(process_pipeline_command(["EXPORT", "gcs://analytics-bucket", "PARQUET"]))

# 2. Dictionary / Mapping Pattern Matching
def authenticate_and_route_user(user_profile):
    match user_profile:
        # Note: Extra keys in user_profile are permitted!
        case {"username": name, "role": "admin", "mfa_enabled": True}:
            return f"Admin User '{name}' authenticated with MFA -> Full Console Access"
        case {"username": name, "role": "admin", "mfa_enabled": False}:
            return f"Security Alert: Admin '{name}' must enable MFA before accessing console!"
        case {"username": name, "role": "engineer", "department": dept}:
            return f"Engineer '{name}' ({dept}) -> Development Sandbox Access"
        case {"username": name}:
            return f"Standard User '{name}' -> Read-Only Access"
        case _:
            return "Anonymous or malformed user payload"

print("\n--- 2. Dictionary Pattern Matching ---")
user1 = {"username": "Rajan", "role": "admin", "mfa_enabled": True, "ip": "192.168.1.50"}
user2 = {"username": "Dave", "role": "admin", "mfa_enabled": False}
user3 = {"username": "Priya", "role": "engineer", "department": "Data Platform", "badge_id": 9921}

print(authenticate_and_route_user(user1))
print(authenticate_and_route_user(user2))
print(authenticate_and_route_user(user3))

# 3. Class / Object Attribute Matching
class PipelineJob:
    def __init__(self, job_id, stage, status, retries=0):
        self.job_id = job_id
        self.stage = stage
        self.status = status
        self.retries = retries

def inspect_pipeline_job(job):
    match job:
        case PipelineJob(job_id=jid, status="FAILED", retries=r) if r < 3:
            return f"Job {jid} FAILED at attempt {r} -> Triggering automatic retry #{r+1}..."
        case PipelineJob(job_id=jid, status="FAILED", retries=r):
            return f"🚨 Job {jid} DEAD-LETTERED after {r} failed attempts -> Alerting on-call engineer!"
        case PipelineJob(job_id=jid, status="SUCCESS"):
            return f"Job {jid} COMPLETED successfully ✅"
        case PipelineJob(job_id=jid, stage=stg):
            return f"Job {jid} currently IN_PROGRESS at stage '{stg}'..."

print("\n--- 3. Class / Object Pattern Matching ---")
print(inspect_pipeline_job(PipelineJob("JOB_101", "Extraction", "FAILED", retries=1)))
print(inspect_pipeline_job(PipelineJob("JOB_102", "Transformation", "FAILED", retries=3)))
print(inspect_pipeline_job(PipelineJob("JOB_103", "Loading", "SUCCESS", retries=0)))
````

## 6️⃣ The `range()` Function & Memory Mechanics

The `range()` function in Python is a built-in type that generates an **immutable sequence of integers** on demand.

### 📝 The 3 Signatures of `range()`
$$\begin{aligned}
&\mathbf{1.\; \text{range}(stop)} &\rightarrow \; [0, 1, 2, \dots, stop-1] &\quad (start=0, step=1) \\[4pt]
&\mathbf{2.\; \text{range}(start, stop)} &\rightarrow \; [start, start+1, \dots, stop-1] &\quad (step=1) \\[4pt]
&\mathbf{3.\; \text{range}(start, stop, step)} &\rightarrow \; [start, start+step, start+2\cdot step, \dots] &\quad (custom\;step)
\end{aligned}$$

```
                           RANGE STEP MECHANICS
      
      range(1, 10, 2)  ──>  [ 1 ,  3 ,  5 ,  7 ,  9 ]   (stop 10 is EXCLUDED)
                              ▲    ▲    ▲    ▲    ▲
                              └──┬─┴──┬─┴──┬─┴──┬─┘
                                 +2   +2   +2   +2
                                 
      range(10, 0, -2) ──>  [ 10,  8 ,  6 ,  4 ,  2 ]   (stop 0 is EXCLUDED)
                              ▲    ▲    ▲    ▲    ▲
                              └──┬─┴──┬─┴──┬─┴──┬─┘
                                 -2   -2   -2   -2
```

---

### 🧠 Memory Efficiency: Lazy Immutable Sequence
In Python 3, `range()` does **NOT** generate or store all numbers in memory at once. It calculates each number **on the fly (lazy evaluation)** using only start, stop, and step attributes.
* `sys.getsizeof(range(10))` $\approx$ **48 bytes**
* `sys.getsizeof(range(1_000_000_000))` $\approx$ **48 bytes** (Constant $O(1)$ memory!)
* Whereas `list(range(1_000_000))` consumes **~8 Megabytes** of RAM!

---

### ⚠️ Common Range Traps & Errors
1. **Float Step**: `range(0, 5, 0.5)` raises `TypeError: 'float' object cannot be interpreted as an integer`.
2. **Step = 0**: `range(1, 5, 0)` raises `ValueError: range() arg 3 must not be zero`.
3. **Mismatched Direction**: `range(5, 1, 2)` produces an **empty range** (silent no-op) because $start > stop$ but $step > 0$.


````python
# ============================================================
# 6️⃣ range() Anatomy, Lazy Evaluation & Edge Case Traps
# ============================================================

import sys

# 1. The 3 Signatures of range()
print("--- 1. range() Signature Variations ---")
print("range(5)          ->", list(range(5)))             # 0 to 4
print("range(2, 8)       ->", list(range(2, 8)))          # 2 to 7
print("range(1, 10, 2)   ->", list(range(1, 10, 2)))      # Odd numbers
print("range(10, 0, -2)  ->", list(range(10, 0, -2)))     # Decreasing even numbers

# 2. Memory Footprint: range() vs list()
r_small = range(10)
r_huge = range(1_000_000_000)
lst_sample = list(range(100_000))

print("\n--- 2. Memory Mechanics: O(1) Constant Space ---")
print(f"Memory for range(10)           : {sys.getsizeof(r_small)} bytes")
print(f"Memory for range(1,000,000,000): {sys.getsizeof(r_huge)} bytes (O(1) Lazy!)")
print(f"Memory for list(100,000 items) : {sys.getsizeof(lst_sample)} bytes (~{sys.getsizeof(lst_sample)//1024} KB)")

# 3. Slicing and Indexing on range Objects
r = range(10, 100, 5)
print("\n--- 3. Range Indexing & Slicing ---")
print(f"range object : {r}")
print(f"Length       : {len(r)}")
print(f"First element: {r[0]}")
print(f"Last element : {r[-1]}")
print(f"Membership 25 in range: {25 in r}")
print(f"Membership 26 in range: {26 in r}")

# 4. Range Traps & Exception Handling
print("\n--- 4. Common Range Traps & Error Handling ---")

# Trap 1: Float step
try:
    bad_range = range(1, 10, 0.5)
except TypeError as e:
    print(f"Trap 1 (Float Argument) Caught: {e}")

# Trap 2: Step of 0
try:
    zero_step = range(1, 10, 0)
except ValueError as e:
    print(f"Trap 2 (Zero Step) Caught     : {e}")

# Trap 3: Silent Empty Range (start > stop with positive step)
empty_r = list(range(10, 2, 1))
print(f"Trap 3 (Mismatched Bounds)    : list(range(10, 2, 1)) -> {empty_r} (Silent Empty!)")
````

## 7️⃣ The `for` Loop & Sequence Iteration

In Python, a `for` loop is a **definite collection iterator**. Instead of manually managing an index counter as in C/C++, Python's `for` loop iterates directly over any **iterable object** (lists, tuples, strings, dictionaries, sets, generators).

```
                      FOR LOOP ITERATION MECHANISM
                                   │
                      [iterable = iter(sequence)]
                                   │
                    ┌──────────────▼──────────────┐
                    │ next(iterable) yields item? │
                    └──────┬───────────────┬──────┘
                       Yes │               │ No (StopIteration)
            ┌──────────────┘               └──────────────┐
            ▼                                             ▼
  [Assign item to variable]                   [Terminate loop normally]
  [Execute loop body statements]              [Execute optional else block]
```

---

### 🚨 The Critical Pitfall: Modifying Collections While Iterating
Never add or remove elements from a `list` while iterating directly over it with a `for` loop!
* **Why it fails**: Removing an element shifts all subsequent elements left by one index. The internal iterator increments its index and skips the next item immediately following the deleted element.
* **The Solution**: Iterate over a shallow copy of the list (`for item in my_list[:]:`) or use a list comprehension.


````python
# ============================================================
# 7️⃣ for Loop: Iterables, Unpacking & The Mutation Pitfall
# ============================================================

# 1. Iterating Over Diverse Built-in Data Types
print("--- 1. Iterating Over Built-in Iterables ---")

# A. String
print("String characters:", end=" ")
for char in "PYSPARK":
    print(char, end=" ")
print()

# B. Tuple
print("Tuple items     :", end=" ")
for zone in ("us-east1", "us-central1", "europe-west1"):
    print(zone, end=" | ")
print()

# C. Dictionary (Keys, Values, Items)
schema_metadata = {"table": "fact_sales", "rows": 1200000, "partitioned": True}
print("\nDictionary items unpacking:")
for key, val in schema_metadata.items():
    print(f"  • {key:<12} => {val}")

# 2. Sequence Unpacking in for Loops
print("\n--- 2. Multi-Variable Unpacking in Loops ---")
batch_partitions = [
    (101, "Orders", 45000),
    (102, "Customers", 12000),
    (103, "Products", 8500)
]

for batch_id, tbl_name, count in batch_partitions:
    print(f"Batch #{batch_id} | Table: {tbl_name:<10} | Ingested: {count:>6} rows")

# 3. 🚨 The Mutation Pitfall: Modifying List While Iterating
print("\n--- 3. Mutating List During Iteration Trap & Fix ---")

# ❌ The Broken Way (Skipping elements due to index shifting):
buggy_list = [10, 20, 30, 30, 40, 50]
for num in buggy_list:
    if num == 30:
        buggy_list.remove(num)
print(f"Buggy deletion result (Skipped consecutive 30!): {buggy_list}")

# ✅ The Correct Way (Iterate over a shallow slice copy `[:]`):
clean_list = [10, 20, 30, 30, 40, 50]
for num in clean_list[:]:  # Slice copy!
    if num == 30:
        clean_list.remove(num)
print(f"Correct deletion result (Iterating over copy [:]): {clean_list}")

# ✅ Alternative Pythonic Fix: List Comprehension
comp_list = [10, 20, 30, 30, 40, 50]
filtered_list = [x for x in comp_list if x != 30]
print(f"List Comprehension filter                        : {filtered_list}")
````

## 8️⃣ The `enumerate()` Built-in Function

When traversing a sequence, you often need both the **item index (counter)** and the **item value**. The built-in `enumerate()` function solves this cleanly by returning an iterator of `(index, item)` tuples.

### 📝 Syntax & Parameters
$$\mathbf{\text{enumerate}(iterable, \; start=0) \rightarrow \text{iterator of } (index, value)}$$

* **`iterable`**: Any iterable object (string, list, tuple, etc.).
* **`start`** *(optional, default=0)*: The integer value at which the index counter begins.

```
                              ENUMERATE ARCHITECTURE
    
    my_list = ["apple", "banana", "cherry"]
    enumerate(my_list, start=1)
                │
                ├──> (1, "apple")
                ├──> (2, "banana")
                └──> (3, "cherry")
```

---

### 🆚 `enumerate()` vs Anti-Pattern `range(len())`

| Feature | `enumerate(sequence)` (✅ Pythonic) | `range(len(sequence))` (❌ Anti-Pattern) |
|---|---|---|
| **Readability** | Clean, readable tuple unpacking | Cluttered double-indexing (`seq[i]`) |
| **Generality** | Works on **any** iterable (even non-indexable streams/sets) | Works **only** on subscriptable sequences with `len()` |
| **Performance** | Fast internal C-level counter generator | Extra overhead of calling `len()` and repeated indexing |
| **Flexibility** | Built-in `start` parameter | Requires manual arithmetic `i + 1` |


````python
# ============================================================
# 8️⃣ enumerate() Anatomy, Custom Start & Generator Recreation
# ============================================================

# 1. Standard enumerate with Default (start=0) and Custom Start (start=1)
pipeline_stages = ["Raw Ingestion", "Schema Validation", "Deduplication", "Silver Transform", "Gold Aggregation"]

print("--- 1. Standard enumerate() with start=1 ---")
for stage_num, stage_name in enumerate(pipeline_stages, start=1):
    print(f"Stage {stage_num}/{len(pipeline_stages)}: {stage_name}")

# 2. Comparison: Pythonic enumerate vs range(len())
print("\n--- 2. Pythonic enumerate() vs range(len()) ---")
devices = ["router-01", "switch-04", "firewall-02"]

# ❌ Clunky anti-pattern
print("Anti-pattern range(len()):")
for i in range(len(devices)):
    print(f"  Index {i}: {devices[i]}")

# ✅ Pythonic enumerate
print("Pythonic enumerate():")
for i, dev in enumerate(devices):
    print(f"  Index {i}: {dev}")

# 3. Custom Under-the-Hood Generator Implementation of enumerate()
def custom_enumerate(iterable, start=0):
    # Recreating enumerate() using Python generators and yield.
    counter = start
    for element in iterable:
        yield counter, element
        counter += 1

print("\n--- 3. Custom Generator Implementation Test ---")
test_fruits = ["Apple", "Banana", "Cherry"]
for idx, fruit in custom_enumerate(test_fruits, start=100):
    print(f"Custom Index {idx}: {fruit}")
````

## 9️⃣ The `while` Loop & Condition-Based Iteration

A `while` loop is an **indefinite condition-controlled loop**. It repeatedly executes its body as long as its boolean condition remains `True`. When the condition evaluates to `False`, execution proceeds past the loop.

```
                         WHILE LOOP EXECUTION LIFECYCLE
                                       │
                              [1. Initialization]
                                       │
                         ┌─────────────▼─────────────┐
                         │   2. Evaluate Condition   │
                         └──────┬─────────────┬──────┘
                           True │             │ False
               ┌────────────────┘             └────────────────┐
               ▼                                               ▼
     [Execute Loop Body]                             [Terminate Loop]
     [3. Update / Increment]                         [Execute optional else]
               │                                               │
               └───────────────────────┬───────────────────────┘
                                       ▼
```

### 🏛️ The 3 Fundamental Pillars of a While Loop
1. **Initialization**: Establishing the loop control variable before entering the loop (e.g. `i = 0`).
2. **Condition**: A boolean expression evaluated before each iteration (e.g. `while i < 10:`).
3. **Update (Increment/Decrement)**: Modifying the control variable inside the loop so that the condition eventually becomes `False`.

---

### 🆚 `for` Loop vs `while` Loop: When to Use Which?

| Feature | `for` Loop | `while` Loop |
|---|---|---|
| **Paradigm** | Sequence-based / Definite iteration | Condition-based / Indefinite iteration |
| **Number of Iterations** | Known in advance (length of iterable) | Unknown in advance (depends on runtime state) |
| **Termination** | Automatic when sequence is exhausted | Must be managed manually via condition updates |
| **Primary Use Cases** | Processing collections, batch lists, fixed ranges | Waiting for network response, polling, user interaction |


````python
# ============================================================
# 9️⃣ while Loop: Counting, Reverse Countdown & Controlled Loops
# ============================================================

# 1. Basic Counting While Loop (Counting Up)
print("--- 1. Basic while Loop (Counting 1 to 5) ---")
counter = 1
while counter <= 5:
    print(f"Iteration step: {counter}")
    counter += 1  # Crucial update step!

# 2. Reverse Countdown While Loop
print("\n--- 2. Reverse Countdown (5 down to 1) ---")
countdown = 5
while countdown > 0:
    print(f"T-minus {countdown} seconds...", end=" ")
    countdown -= 1
print("🚀 BLAST OFF!")

# 3. Controlled Infinite Loop (while True + break)
# Simulating a polling consumer reading from a message buffer
print("\n--- 3. Controlled Ingestion Polling Loop ---")
message_queue = ["EVENT_01", "EVENT_02", "EVENT_03", "EMPTY_QUEUE"]
queue_idx = 0

while True:
    msg = message_queue[queue_idx]
    queue_idx += 1
    
    if msg == "EMPTY_QUEUE":
        print("Queue drained! Exiting polling loop.")
        break
    
    print(f"Processing message: {msg}")

# 4. Sentinel-Controlled Verification Loop
print("\n--- 4. Sentinel-Controlled Verification Loop ---")
simulated_inputs = iter(["wrong_pass", "admin123", "secret_token_99"])
authenticated = False
attempts = 0

while not authenticated and attempts < 3:
    attempts += 1
    entered_pass = next(simulated_inputs)
    print(f"Attempt #{attempts}: Trying token '{entered_pass}'...")
    if entered_pass == "secret_token_99":
        authenticated = True
        print("✅ Access Granted: Token Verified!")
    else:
        print("❌ Invalid token.")
````

## 🔟 Loop Control Statements: `break`, `continue`, and `pass`

Loop control statements alter the regular sequential flow of loop execution based on conditional triggers.

```
       [break]                         [continue]                          [pass]
          │                                │                                 │
   Immediately TERMINATES           SKIPS the remainder              NULL statement / No-op
   the entire loop and              of the current iteration         placeholder used when syntax
   jumps outside.                   and jumps to the NEXT.           requires a code block.
```

### 📊 Comparison of Loop Control Statements

| Statement | Effect on Current Iteration | Effect on Entire Loop | Effect on Loop `else` |
|---|---|---|---|
| **`break`** | Terminated immediately | **Terminated immediately** | **SKIPPED** |
| **`continue`** | Skipped immediately | Continues with next iteration | Normal execution when loop finishes |
| **`pass`** | Does nothing (No-op) | Continues sequentially | Normal execution |

---

### ⚠️ Common Mistakes & Traps
1. **`continue` Before Increment in `while` Loop**: If `continue` is placed before the loop increment (`i += 1`), the increment statement is skipped, the variable never changes, causing an **unintentional infinite loop**!
2. **`break` / `continue` Outside Loop**: Using `break` or `continue` outside a `for` or `while` loop raises `SyntaxError: 'break' outside loop`.


````python
# ============================================================
# 🔟 break, continue, pass & Execution Flow Comparison
# ============================================================

# 1. break in Search Optimization (Early Exit)
print("--- 1. break in Search Optimization ---")
partition_ids = [101, 102, 103, 104, 105, 106, 107]
target_partition = 104

for pid in partition_ids:
    print(f"Checking Partition ID {pid}...", end=" ")
    if pid == target_partition:
        print("FOUND! 🎯 Stopping search immediately.")
        break

# 2. continue in Data Cleansing (Skipping Corrupted / Dirty Records)
print("\n--- 2. continue in Data Filtering ---")
raw_measurements = [24.5, -999.0, 25.1, None, 26.8, -500.0, 27.2]

clean_measurements = []
for val in raw_measurements:
    if val is None or val < 0:
        print(f"  [SKIPPING invalid/corrupt record: {val}]")
        continue
    clean_measurements.append(val)
    print(f"  [RECORD KEPT: {val}]")

print(f"Final Cleaned Dataset: {clean_measurements}")

# 3. Side-by-Side Comparison of break vs continue
print("\n--- 3. Side-by-Side Execution Trace ---")
print("With break at 3    :", end=" ")
for i in range(1, 6):
    if i == 3:
        break
    print(i, end=" ")
print()

print("With continue at 3 :", end=" ")
for i in range(1, 6):
    if i == 3:
        continue
    print(i, end=" ")
print()

# 4. pass as a Syntactic Placeholder
print("\n--- 4. pass Statement Demonstration ---")
def future_etl_stage():
    pass  # Placeholder for upcoming feature implementation

for i in range(3):
    if i == 1:
        pass  # Do nothing for i == 1
    print(f"Step {i} executed")

# 5. Trap: Correct Placement of Increment with continue in while Loop
print("\n--- 5. Safe while-continue Pattern ---")
k = 0
while k < 5:
    k += 1  # Increment MUST happen before continue!
    if k == 3:
        print(f"Skipping {k}...")
        continue
    print(f"Value: {k}")
````

## 1️⃣1️⃣ Loop `else` Clause: `for-else` and `while-else` Architecture

Python features a unique syntax where an `else` block can be attached directly to a `for` or `while` loop.

### 🚨 The Golden Rule of Loop `else`
> **The `else` block executes IF AND ONLY IF the loop terminates NORMALLY (i.e. exhausts all iterations or the condition becomes `False`).**
> **If the loop is exited prematurely via a `break` statement, the `else` block is COMPLETELY BYPASSED!**

```
                       LOOP ELSE EXECUTION MECHANISM
                                     │
                             [Loop Iteration]
                                     │
                 ┌───────────────────┴───────────────────┐
                 ▼                                       ▼
       [Terminated by 'break']                 [Completed Normally]
                 │                                       │
        (else block SKIPPED)                    (else block RUNS ✅)
```

---

### 💡 Why Use Loop `else`? (Eliminating Boolean Flag Variables)
In languages like C or Java, searching a list requires a temporary flag variable (`found = false`). In Python, the `for-else` pattern cleanly replaces the flag:
* If the item is found $\rightarrow$ `break` (skips `else`).
* If the loop finishes without finding the item $\rightarrow$ `else` executes automatically!


````python
# ============================================================
# 1️⃣1️⃣ for-else & while-else Search Patterns (No Flag Variable!)
# ============================================================

# 1. for-else Search: Item FOUND (break skips else)
print("--- 1. for-else (Item FOUND -> else skipped) ---")
warehouse_tables = ["dim_customers", "dim_products", "fact_transactions", "dim_stores"]
search_target = "fact_transactions"

for tbl in warehouse_tables:
    if tbl == search_target:
        print(f"✅ Found target table: '{search_target}' -> Ready for query execution.")
        break
else:
    print(f"❌ Target table '{search_target}' not found in warehouse metadata.")

# 2. for-else Search: Item NOT Found (Loop completes -> else executes)
print("\n--- 2. for-else (Item NOT Found -> else executes) ---")
missing_target = "fact_inventory_daily"

for tbl in warehouse_tables:
    if tbl == missing_target:
        print(f"✅ Found target table: '{missing_target}'")
        break
else:
    print(f"❌ Target table '{missing_target}' NOT FOUND in warehouse metadata! (else executed)")

# 3. while-else Retry Logic
print("\n--- 3. while-else Ingestion Retry Architecture ---")
max_attempts = 3
attempt = 0
service_healthy = False  # Simulating a failing downstream service

while attempt < max_attempts:
    attempt += 1
    print(f"Attempt #{attempt}: Connecting to Lakehouse...")
    if service_healthy:
        print("Connected successfully! ✅")
        break
else:
    print("🚨 Ingestion Pipeline Failure: Max retry attempts exhausted! (else executed)")
````

## 1️⃣2️⃣ Nested Loops & Multi-Dimensional Processing

A **nested loop** is a loop situated inside another loop. 

### 🔄 The 4 Combinations of Nested Loops
1. **`for` inside `for`**: Traversing matrices, 2D grids, and cross-join combinations.
2. **`while` inside `while`**: Multi-level stateful polling loops.
3. **`for` inside `while`**: Iterating over fixed batch records inside a streaming loop.
4. **`while` inside `for`**: Running per-item retry loops across a list of endpoints.

```
                      NESTED LOOP EXECUTION ARCHITECTURE
    
    Outer Loop: i = 1  ──>  Inner Loop: j = 1, j = 2, j = 3  (Inner runs fully)
    Outer Loop: i = 2  ──>  Inner Loop: j = 1, j = 2, j = 3  (Inner runs fully)
    Outer Loop: i = 3  ──>  Inner Loop: j = 1, j = 2, j = 3  (Inner runs fully)
    
    Total Iterations = (Outer Iterations M) × (Inner Iterations N) = M × N
```

---

### 🚨 Critical Scope Rule for `break` and `continue` in Nested Loops
* `break` and `continue` apply **ONLY to the nearest innermost enclosing loop**.
* A `break` inside the inner loop terminates the inner loop, but the **outer loop continues running**!
* **How to break out of multiple loops**: Use a helper function with `return`, a boolean flag, or raise a custom control exception.


````python
# ============================================================
# 1️⃣2️⃣ Nested Loops: Matrices, Breakout Strategies & Patterns
# ============================================================

# 1. Nested for Loop: 2D Matrix Grid Traversal
print("--- 1. Nested for Loop (2D Matrix Coordinate Grid) ---")
matrix_2d = [
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
]

for row_idx, row in enumerate(matrix_2d):
    for col_idx, val in enumerate(row):
        print(f"Cell ({row_idx}, {col_idx}) = {val:<2}", end="  |  ")
    print()

# 2. Scope of break in Nested Loops (Only exits the INNER loop)
print("\n--- 2. break in Inner Loop (Outer Continues) ---")
for outer in range(1, 4):
    print(f"Outer Loop {outer}:", end=" ")
    for inner in range(1, 4):
        if inner == 2:
            break  # Exits ONLY the inner loop!
        print(f"inner={inner}", end=" ")
    print("-> (Inner broken, outer proceeding)")

# 3. Clean Multi-Level Breakout Strategy (Function with return)
print("\n--- 3. Multi-Level Loop Breakout via Function return ---")
def find_item_in_matrix(matrix, target):
    for r_idx, row in enumerate(matrix):
        for c_idx, val in enumerate(row):
            if val == target:
                return f"Target {target} FOUND at Matrix Coordinates ({r_idx}, {c_idx})! 🎯"
    return f"Target {target} NOT FOUND."

print(find_item_in_matrix(matrix_2d, 50))
print(find_item_in_matrix(matrix_2d, 999))

# 4. Nested Pattern Printing
print("\n--- 4. Nested Pattern Printing (Star Pyramid) ---")
rows = 4
for i in range(1, rows + 1):
    # Inner loop for spacing + inner loop for stars
    spaces = " " * (rows - i)
    stars = "*" * (2 * i - 1)
    print(f"{spaces}{stars}")
````

## 1️⃣3️⃣ Common Traps, Anti-Patterns & Error Taxonomy

Understanding common syntax and logical pitfalls is essential for writing robust and maintainable Python code.

### 📋 Control Structure Error Taxonomy

| Error Type | Trigger / Root Cause | Example Violation | Correct Solution |
|---|---|---|---|
| **`SyntaxError: expected ':'`** | Omitted colon at the end of a header statement | `if x > 5` | `if x > 5:` |
| **`IndentationError`** | Misaligned or missing indentation in a block | `if x > 5:`<br>`print(x)` | `if x > 5:`<br>`    print(x)` |
| **`SyntaxError: invalid syntax`** | Using assignment `=` instead of equality `==` | `if x = 10:` | `if x == 10:` |
| **`SyntaxError: 'break' outside loop`**| `break` placed outside `for` or `while` | `if condition: break` | Place `break` only within a loop |
| **`TypeError`** | Passing float arguments to `range()` | `range(0, 5, 0.5)` | Use integer math or `numpy.arange()` |
| **`ValueError`** | Specifying a step of `0` in `range()` | `range(1, 10, 0)` | Step must be non-zero integer |
| **Unintended Infinite Loop** | Missing increment before `continue` in `while` | `while i < 5:`<br>`  if i==2: continue`<br>`  i += 1` | Increment before `continue` |
| **Unreachable Case Trap** | Catch-all `case _:` or `case x:` placed first | `match v:`<br>`  case _:`<br>`  case 10:` | Keep `case _:` as the final case |


````python
# ============================================================
# 1️⃣3️⃣ Error Handling & Anti-Pattern Demonstrations
# ============================================================

# 1. Handling Invalid Assignment in Condition vs Comparison
print("--- 1. Comparison Equality Check (== vs =) ---")
status = "PENDING"
# Note: if status = "PENDING" is a SyntaxError in Python!
if status == "PENDING":
    print("Status correctly checked with '==' comparison operator.")

# 2. Defensive Range Type & Value Validation
print("\n--- 2. Defensive range() Error Handling ---")
def safe_custom_range(start, stop, step):
    try:
        return list(range(start, stop, step))
    except TypeError as te:
        return f"Caught TypeError: {te} (Arguments must be integers!)"
    except ValueError as ve:
        return f"Caught ValueError: {ve} (Step cannot be 0!)"

print("Valid Call (1, 6, 2)  :", safe_custom_range(1, 6, 2))
print("Float Step (1, 6, 0.5):", safe_custom_range(1, 6, 0.5))
print("Zero Step (1, 6, 0)   :", safe_custom_range(1, 6, 0))

# 3. Variable Leakage / Shadowing in match-case Warning
print("\n--- 3. match-case Variable Binding Scope ---")
# When a variable pattern is used in match-case, it binds to the global/local scope!
active_val = 999
match 42:
    case active_val:  # This binds active_val to 42!
        print(f"Captured: {active_val} (Notice active_val was re-bound from 999 to 42!)")
````

In real-world data engineering pipelines, control structures orchestrate **batch chunking**, **schema dispatching**, **defensive stream sanitization**, and **resilient network retries with exponential backoff**.

```
                         [Raw Multi-Source Telemetry]

                  │                                         │
                  ▼                                         ▼
        [3. Defensive Filtering]                  [4. Ingestion & Retries]
        (continue + truthiness)                   (while-else with Backoff)
```

### 🏆 Master Summary Reference Matrix

| Feature | Primary Purpose | Golden Rule / Critical Best Practice |
|---|---|---|
| **`if-elif-else`** | Multi-branch boolean decision ladder | First `True` condition executes; rest skipped |
| **Ternary Operator** | Single-line conditional value expression | `[on_true] if [cond] else [on_false]` |
| **`match-case`** | Structural pattern matching (Python 3.10+) | No `break` needed; `case _:` must be last |
| **`range()`** | Memory-efficient arithmetic integer generator | $O(1)$ lazy memory; step cannot be 0 or float |
| **`for` Loop** | Definite collection / sequence iteration | Never mutate a list while iterating over it! |
| **`enumerate()`** | Indexed collection traversal | Prefer over `range(len())`; returns `(index, item)` |
| **`while` Loop** | Indefinite condition-based iteration | Ensure loop variable updates towards termination |
| **`break` / `continue`** | Loop interruption / iteration skipping | Affects **only** the nearest enclosing inner loop |
| **Loop `else`** | Normal loop completion handler | Runs **only** if loop finished without `break` |


````python
# ============================================================
# 1️⃣4️⃣ Production Data Engineering Pipeline & Master Architecture
# ============================================================

import time
import json

# 1. Batch Chunking Engine (using range & enumerate)
def chunked_batch_processor(record_stream, chunk_size=3):
    print("=== 1. High-Throughput Batch Chunking Engine ===")
    total_records = len(record_stream)
    
    for batch_num, start_idx in enumerate(range(0, total_records, chunk_size), start=1):
        end_idx = min(start_idx + chunk_size, total_records)
        current_chunk = record_stream[start_idx:end_idx]
        print(f"Processing Batch #{batch_num} | Rows [{start_idx}:{end_idx}] ({len(current_chunk)} records)... [SUCCESS ✅]")

sample_records = [f"REC_{i:04d}" for i in range(1, 11)]
chunked_batch_processor(sample_records, chunk_size=3)

# 2. Defensive Data Sanitization (using continue & truthiness)
def clean_telemetry_stream(raw_events):
    print("\n=== 2. Defensive Telemetry Sanitizer ===")
    clean_events = []
    
    for event in raw_events:
        # Check truthiness of core attributes
        device_id = event.get("device_id")
        reading = event.get("reading")
        
        # Defensive skip using continue
        if not device_id or reading is None or reading < 0:
            print(f"⚠️ Dropping Dirty Record: {event}")
            continue
            
        clean_events.append({"device": device_id, "val": reading})
        print(f"✅ Ingested Clean Record: {device_id} -> {reading}")
        
    return clean_events

raw_stream = [
    {"device_id": "IOT_A", "reading": 24.2},
    {"device_id": "", "reading": 25.0},          # Empty ID (Falsy)
    {"device_id": "IOT_B", "reading": -999.0},    # Corrupted reading
    {"device_id": "IOT_C", "reading": None},       # Null reading
    {"device_id": "IOT_D", "reading": 28.5}
]
cleaned = clean_telemetry_stream(raw_stream)

# 3. JSON Event Dispatcher with match-case
def dispatch_lakehouse_event(event_payload):
    print("\n=== 3. Structural Pattern Matching Event Dispatcher ===")
    for evt in event_payload:
        match evt:
            case {"type": "SCHEMA_CHANGE", "table": tbl, "new_column": col}:
                print(f"[SCHEMA MIGRATION] Adding column '{col}' to table '{tbl}'")
            case {"type": "BATCH_LOAD", "table": tbl, "rows": r} if r >= 1000:
                print(f"[BULK INGESTION] Loaded {r} rows into '{tbl}' via Spark")
            case {"type": "BATCH_LOAD", "table": tbl, "rows": r}:
                print(f"[STREAM INGESTION] Loaded {r} rows into '{tbl}' via Microbatch")
            case {"type": "MAINTENANCE", "action": act}:
                print(f"[MAINTENANCE] Performing '{act}' on lakehouse cluster")
            case _:
                print(f"[UNRECOGNIZED EVENT] Forwarding to Dead-Letter Queue: {evt}")

lakehouse_events = [
    {"type": "SCHEMA_CHANGE", "table": "dim_customers", "new_column": "tax_id"},
    {"type": "BATCH_LOAD", "table": "fact_orders", "rows": 45000},
    {"type": "BATCH_LOAD", "table": "dim_stores", "rows": 50},
    {"type": "MAINTENANCE", "action": "VACUUM_DELTA_TABLES"},
    {"type": "UNKNOWN_SIGNAL", "payload": 0xDEAD}
]
dispatch_lakehouse_event(lakehouse_events)

# 4. Resilient Network API Poller with Exponential Backoff (while-else)
def fetch_api_with_exponential_backoff(max_retries=3):
    print("\n=== 4. Resilient API Ingestion with Exponential Backoff ===")
    attempt = 0
    base_delay = 0.05  # Simulated 50ms base delay
    
    # Simulate a transient failure that succeeds on attempt 3
    while attempt < max_retries:
        attempt += 1
        print(f"Connecting to REST API Endpoint (Attempt #{attempt}/{max_retries})...", end=" ")
        
        if attempt == 3:  # Succeeds on 3rd attempt
            print("HTTP 200 OK! Data stream acquired. ✅")
            break
        else:
            backoff_time = base_delay * (2 ** (attempt - 1))
            print(f"HTTP 503 Unavailable -> Backing off for {backoff_time:.2f}s...")
            time.sleep(backoff_time)
    else:
        print("🚨 CRITICAL: Ingestion failed after maximum retry attempts!")

fetch_api_with_exponential_backoff(max_retries=3)
````
