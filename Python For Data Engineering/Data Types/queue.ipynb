## 1️⃣ What is a Queue?

A **queue** is a data structure that holds items in an order. The most important rule is:

> **FIFO = First In, First Out**

The item that enters the queue **first** is the item that leaves **first**.

### An everyday example

Imagine students standing in a line at a college office. Rajan arrives first, then Asha, then Imran. The person at the front is served first. A new student joins at the back; they do not jump ahead of everyone.

A queue in a program follows the same idea.

| Student action | Queue meaning |
|---|---|
| Join the back of the line | **Enqueue** — add an item |
| The person at the front is served and leaves | **Dequeue** — remove the oldest item |
| Ask who is at the front | **Front / Peek** — inspect the next item |

A queue is useful whenever things should be handled in the order they arrive, such as print jobs, customer requests, or tasks waiting to run.

### A queue drawn as boxes

Suppose we enqueue `10`, then `20`, then `30`:

```text
Dequeue happens here                         Enqueue happens here
        ↓                                            ↓
      FRONT                                        REAR
        |                                            |
        v                                            v
     +------+     +------+     +------+
     |  10  | --> |  20  | --> |  30  |
     +------+     +------+     +------+
       oldest item                 newest item
```

If we dequeue once, `10` leaves first. The queue becomes `20, 30`. If we enqueue `40`, it joins at the rear: `20, 30, 40`.

**Remember:** In the diagram, the front is where items leave and the rear is where new items join.

## 2️⃣ Queue Vocabulary and Operations

| Term | Easy meaning | Example |
|---|---|---|
| **Queue** | The whole line of waiting items | `[10, 20, 30]` |
| **Front** | The oldest item; the next one to leave | `10` |
| **Rear** | The newest item; where the next item joins | `30` |
| **Enqueue** | Add an item at the rear | Add `40` |
| **Dequeue** | Remove and return the front item | Remove `10` |
| **Peek / Front** | See the front without removing it | Read `10` |
| **is_empty** | Check whether there are no items | `True` for an empty queue |
| **Size** | Number of items currently waiting | `3` for three items |

The names describe the idea, not one particular Python method. Different queue implementations use different method names.

## 3️⃣ Creating a Queue with a Python List

The simplest way to understand a queue is to use a normal Python list.

```python
queue = []
```

An empty list can represent an empty queue. For this first example, the **left side** of the list is the front and the **right side** is the rear:

```text
Queue: [10, 20, 30]
        ↑        ↑
      FRONT    REAR
```

We will use `append(item)` to add at the rear and `pop(0)` to remove from the front. This is easy to learn, but later we will see why `pop(0)` can be slow for large queues.

````python
# ============================================
# 3️⃣ Creating an empty queue
# ============================================

queue = []  # We will use this list as a queue

print("Queue:", queue)
print("Type:", type(queue))
print("Number of items:", len(queue))
````

### 4️⃣ Enqueue — adding items to the rear

**Enqueue** means adding a new item to the back (rear) of the queue.

When the queue is represented by a list, use:

**Syntax:** `queue.append(item)`

`append()` adds the new item at the right-hand end, which we are using as the rear.

````python
# ============================================
# 4️⃣ Enqueue — add at the REAR
# ============================================

queue = []
queue.append(10)  # First item arrives
queue.append(20)  # Second item joins at the rear
queue.append(30)  # Third item joins at the rear

print("Queue:", queue)  # [10, 20, 30]
print("Front item:", queue[0])
print("Rear item:", queue[-1])
````

### 5️⃣ Dequeue — removing the front item

**Dequeue** removes **and returns** the item at the front of the queue.

For our list-based queue, use:

**Syntax:** `item = queue.pop(0)`

The index `0` means “the first item.” After it is removed, the remaining items move one position to the left.

````python
# ============================================
# 5️⃣ Dequeue — remove from the FRONT
# ============================================

queue = [10, 20, 30]
removed_item = queue.pop(0)

print("Removed item:", removed_item)  # 10
print("Queue after dequeue:", queue)  # [20, 30]
````

### 6️⃣ Peek — view the front without removing it

**Peek** means checking the next item without changing the queue.

For a **non-empty list-based queue**, `queue[0]` returns the front item. Use `queue[-1]` to see the rear item.

| Expression | Meaning |
|---|---|
| `queue[0]` | Front / next item to be removed |
| `queue[-1]` | Rear / newest item |
| `queue[1]` | Second item, if it exists |

These index expressions raise an `IndexError` if the queue is empty, so check first when emptiness is possible.

````python
# ============================================
# 6️⃣ Peek at the front and rear
# ============================================

queue = [10, 20, 30]

print("Front (next to leave):", queue[0])
print("Rear (newest item):", queue[-1])
print("Queue remains unchanged:", queue)
````

### 7️⃣ Check whether a queue is empty and find its size

Use:

- `len(queue)` to count the items.
- `len(queue) == 0` to check whether it is empty.
- `not queue` as a short Pythonic empty check.

In Python, an empty list is considered `False` in a condition, while a non-empty list is considered `True`.

````python
# ============================================
# 7️⃣ is_empty and size
# ============================================

queue = []
print("Size:", len(queue))                 # 0
print("Empty using len:", len(queue) == 0) # True
print("Empty using not:", not queue)       # True

queue.append("Rajan")
print("\nAfter enqueue:", queue)
print("Size:", len(queue))
print("Empty now?:", not queue)             # False
````

## 8️⃣ See FIFO in Action — Step by Step

Follow each operation. The **leftmost item is the front** in our list-based examples.

| Operation | Queue after operation | Front |
|---|---|---|
| Start | `[]` | No front item |
| Enqueue 5 | `[5]` | `5` |
| Enqueue 8 | `[5, 8]` | `5` |
| Enqueue 12 | `[5, 8, 12]` | `5` |
| Dequeue | `[8, 12]` | `8` |
| Enqueue 20 | `[8, 12, 20]` | `8` |
| Dequeue | `[12, 20]` | `12` |

Notice that adding new items does not change who is at the front. The oldest item stays at the front until it is removed.

````python
# ============================================
# 8️⃣ FIFO walkthrough
# ============================================

queue = []

queue.append(5)
print("After enqueue 5:", queue)

queue.append(8)
print("After enqueue 8:", queue)

queue.append(12)
print("After enqueue 12:", queue)

served = queue.pop(0)
print("Dequeued:", served, "| Queue now:", queue)

queue.append(20)
print("After enqueue 20:", queue)

served = queue.pop(0)
print("Dequeued:", served, "| Queue now:", queue)
````

## 9️⃣ Important Edge Case — Dequeue from an Empty Queue

A queue may become empty after its last item is removed. Calling `pop(0)` on an empty list raises `IndexError` because there is no item at index `0`.

```python
queue = []
# queue.pop(0)  # Uncomment to see the IndexError
```

The right approach depends on the program:

- **Check first** when an empty queue is a normal situation.
- **Use `try`/`except`** when you want to respond to a failed removal.
- **Raise an error** if removing from an empty queue means the caller has made a programming mistake.

````python
# ============================================
# 9️⃣ Safe dequeue and safe peek examples
# ============================================

def safe_dequeue(queue):
    """Remove and return the front item, or None if the queue is empty."""
    if not queue:
        return None
    return queue.pop(0)


def safe_peek(queue):
    """Return the front item, or None if the queue is empty."""
    if not queue:
        return None
    return queue[0]

queue = []
print("Safe dequeue on empty queue:", safe_dequeue(queue))
print("Safe peek on empty queue:", safe_peek(queue))

queue.extend(["A", "B"])
print("Front before dequeue:", safe_peek(queue))
print("Removed:", safe_dequeue(queue))
print("Queue after removal:", queue)
````

**A small design warning:** These helper functions use `None` to mean “there was no item.” That can be confusing if `None` is itself a valid value in your queue. Larger programs may prefer a unique sentinel object or may raise a clear exception instead.

### 9.1 A `try`/`except` approach

Python raises `IndexError` when a list operation uses an invalid index. You can catch that error if your program needs to handle the empty case.

````python
# A try/except version

def dequeue_or_none(queue):
    try:
        return queue.pop(0)
    except IndexError:
        return None

print(dequeue_or_none([]))          # None
print(dequeue_or_none(["first"]))  # first
````

## 🔟 Useful Python Tricks with a List-Based Queue

| Task | Code | Explanation |
|---|---|---|
| Enqueue one item | `queue.append(x)` | Add to the rear |
| Dequeue one item | `queue.pop(0)` | Remove from the front |
| See the front | `queue[0]` | Inspect the next item, if present |
| See the rear | `queue[-1]` | Inspect the newest item, if present |
| Count items | `len(queue)` | Return the current size |
| Check empty | `not queue` | `True` when no items remain |
| Add several items | `queue.extend(items)` | Add each item at the rear |
| Process everything | `while queue:` | Keep going while items remain |

**Important:** `extend()` adds each element of an iterable. `queue.append([1, 2])` adds the whole list as one item, but `queue.extend([1, 2])` adds `1` and `2` separately.

````python
# ============================================
# 🔟 Useful queue patterns
# ============================================

queue = []
queue.append("print report")
queue.extend(["print photo", "print notes"])
print("Queued jobs:", queue)

# Process items in arrival order
while queue:
    current_job = queue.pop(0)
    print("Processing:", current_job)

print("Queue after processing everything:", queue)
````

### A note about copying and sharing a queue

Writing `other_queue = queue` does **not** create a new queue. Both variable names refer to the same list. If one name removes an item, the other name sees that change too.

Use `queue.copy()` to make a separate shallow copy when that is what you need.

````python
# Alias versus a separate shallow copy

original_queue = ["A", "B", "C"]
alias_queue = original_queue       # Same list object
copy_queue = original_queue.copy() # Separate list

alias_queue.append("D")
print("Original after changing alias:", original_queue)  # Includes D
print("Alias:", alias_queue)
print("Independent copy:", copy_queue)                  # Does not include D
````

## 1️⃣1️⃣ Why `list.pop(0)` Is Not Ideal for Large Queues

For a list like `[10, 20, 30, 40]`, `pop(0)` removes `10`. Python then shifts the remaining elements toward the front so the list stays contiguous. If the list has many items, this shifting work can be expensive.

Typical complexity:

- `append(x)` — **O(1) amortized**
- `pop(0)` — **O(n)**, because many remaining items may need to shift
- `queue[0]` — **O(1)**

This does not mean `pop(0)` is always wrong. It is easy to understand and is fine for small demonstrations. For general-purpose FIFO queues, use `collections.deque`.

---
## 1️⃣2️⃣ `collections.deque` — A Better General-Purpose Queue

A **deque** (pronounced like “deck”) is a double-ended queue. It efficiently supports adding and removing items at **both ends**.

Import it with:

```python
from collections import deque
```

For FIFO queue behaviour, use:

- `queue.append(item)` to add at the right/rear.
- `queue.popleft()` to remove from the left/front.
- `queue[0]` to inspect the front when the deque is not empty.
- `queue[-1]` to inspect the rear when the deque is not empty.

For normal enqueue and dequeue operations, `deque` supports approximately **O(1)** time per operation.

````python
# ============================================
# 1️⃣2️⃣ A queue using collections.deque
# ============================================

from collections import deque

queue = deque()  # Start empty
queue.append(10) # Enqueue
queue.append(20)
queue.append(30)

print("Queue:", queue)
print("Front:", queue[0])
print("Rear:", queue[-1])

removed = queue.popleft()  # Dequeue from the left/front
print("Dequeued:", removed)
print("Queue after dequeue:", queue)
````

### List queue vs `deque`

| Operation | List approach | `deque` approach | Typical time |
|---|---|---|---|
| Enqueue | `append(x)` | `append(x)` | O(1) amortized |
| Dequeue | `pop(0)` | `popleft()` | List O(n); deque O(1) |
| Peek front | `queue[0]` | `queue[0]` | O(1) |
| Peek rear | `queue[-1]` | `queue[-1]` | O(1) |
| Add left | `insert(0, x)` | `appendleft(x)` | List O(n); deque O(1) |

**Practical rule:** Learn the queue idea with a list if that feels simpler, but prefer `deque` for actual FIFO queue code that may grow.

### 12.1 Safe operations with `deque`

An empty deque also raises `IndexError` when you call `popleft()` or try to access `queue[0]`. The container does not guess what you want to happen when it is empty; your program must choose how to handle it.

````python
from collections import deque

queue = deque()

if queue:
    print("Front:", queue[0])
else:
    print("Queue is empty; there is no front item.")

if queue:
    print("Dequeued:", queue.popleft())
else:
    print("Nothing to dequeue.")
````

## 1️⃣3️⃣ Implementing a Reusable `Queue` Class

So far, we have used a list or deque directly. A **class** lets us bundle the queue's data and operations in one reusable object. This can make larger programs easier to read.

Our class will use a `deque` internally and provide clear methods:

- `enqueue(item)` — add at the rear
- `dequeue()` — remove and return the front
- `peek()` — inspect the front without removal
- `is_empty()` — check whether it contains no items
- `size()` — count the items

For this version, calling `dequeue()` or `peek()` on an empty queue raises a clear `IndexError`.

````python
# ============================================
# 1️⃣3️⃣ Define a reusable Queue class
# ============================================

from collections import deque

class Queue:
    """A FIFO queue implemented using collections.deque."""

    def __init__(self):
        self._items = deque()

    def enqueue(self, item):
        """Add item at the rear of the queue."""
        self._items.append(item)

    def dequeue(self):
        """Remove and return the front item."""
        if self.is_empty():
            raise IndexError("Cannot dequeue from an empty queue")
        return self._items.popleft()

    def peek(self):
        """Return the front item without removing it."""
        if self.is_empty():
            raise IndexError("Cannot peek at an empty queue")
        return self._items[0]

    def is_empty(self):
        return len(self._items) == 0

    def size(self):
        return len(self._items)

    def __len__(self):
        return len(self._items)

    def __repr__(self):
        return f"Queue({list(self._items)!r})"
````

### 13.1 Using the `Queue` class

After running the class definition above, create an object using `Queue()`. Each object gets its own internal deque.

````python
my_queue = Queue()
print(my_queue)
print("Empty?", my_queue.is_empty())

my_queue.enqueue("first")
my_queue.enqueue("second")
my_queue.enqueue("third")

print("Queue:", my_queue)
print("Front:", my_queue.peek())
print("Size:", my_queue.size())
print("Dequeued:", my_queue.dequeue())
print("After dequeue:", my_queue)
print("Length using len():", len(my_queue))
````

### 13.2 Handling an empty custom queue

The class raises a helpful `IndexError` when an invalid operation is attempted on an empty queue. You can catch it if an empty queue is possible during normal program flow.

````python
empty_queue = Queue()

try:
    empty_queue.dequeue()
except IndexError as error:
    print("Handled safely:", error)

try:
    empty_queue.peek()
except IndexError as error:
    print("Handled safely:", error)
````

### 13.3 Why use methods instead of modifying the internal data directly?

The leading underscore in `_items` is a Python convention meaning “internal implementation detail.” It is not absolute privacy, but users of the class are encouraged to call the public methods instead of changing `_items` themselves. That helps protect the queue's rules and makes it easier to change the implementation later.

## 1️⃣4️⃣ Time and Space Complexity

Let **n** represent the number of items in the queue.

| Operation | List with `append()` / `pop(0)` | `collections.deque` |
|---|---:|---:|
| Enqueue | O(1) amortized | O(1) |
| Dequeue | O(n) | O(1) |
| Peek front | O(1) | O(1) |
| Peek rear | O(1) | O(1) |
| Check empty | O(1) | O(1) |
| Count items | O(1) | O(1) |
| Store n items | O(n) space | O(n) space |

**What does O(n) mean?** If there are about ten times as many items, an O(n) operation may need roughly ten times as much work. O(1) means the operation takes roughly constant time as the number of items grows, ignoring implementation details and occasional resizing.

Python lists may resize on append, so list append is described as O(1) **amortized**, not guaranteed O(1) for every single call.

## 1️⃣5️⃣ `queue.Queue` — Queues for Threaded Programs

Python has another queue implementation in the `queue` module. It is especially useful when **multiple threads** need to safely pass work to one another.

```python
from queue import Queue
```

The naming can be confusing: `collections.deque` is often ideal for a normal queue in a single-threaded algorithm, while `queue.Queue` provides locking and optional blocking behaviour for producer-consumer programs.

Common methods:

- `put(item)` — add an item
- `get()` — remove and return an item
- `empty()` — check whether it appears empty (not a reliable coordination strategy across threads)
- `task_done()` / `join()` — track when queued work has been completed

For this notebook's main queue algorithms, we will mostly use `collections.deque`.

````python
# ============================================
# 1️⃣5️⃣ A simple queue.Queue example (single thread)
# ============================================

from queue import Queue as ThreadQueue

jobs = ThreadQueue()
jobs.put("send email")
jobs.put("create report")

print("First job:", jobs.get())
print("Second job:", jobs.get())
````

### 15.1 A safe producer-consumer pattern

A **producer** adds jobs. A **consumer** takes jobs and handles them. With `queue.Queue`, a consumer can wait for work to arrive. This example is deliberately single-threaded so it finishes immediately and is easy to understand.

The special marker `None` below means “there are no more jobs.” In a real application, choose a marker that cannot be confused with real work.

````python
from queue import Queue as ThreadQueue

jobs = ThreadQueue()

# Producer: add work
for job in ["resize image", "save file"]:
    jobs.put(job)
jobs.put(None)  # Sentinel: tell the consumer to stop

# Consumer: take one job at a time
while True:
    job = jobs.get()  # Waits if the queue is empty
    try:
        if job is None:
            print("No more jobs. Stopping consumer.")
            break
        print("Working on:", job)
    finally:
        jobs.task_done()  # Mark this retrieved item as handled

jobs.join()  # Returns once every put item has a matching task_done()
print("All queued items have been accounted for.")
````

## 1️⃣8️⃣ Application — Breadth-First Search (BFS)

**Breadth-First Search (BFS)** visits nodes in a graph in layers: first the starting node, then its nearby unvisited neighbours, then neighbours farther away.

BFS uses a queue because the nodes discovered first should be explored first. This is the core idea behind finding the shortest path in an **unweighted graph**.

### A small graph

```text
A ---- B ---- D
|      |
C ---- E
```

The exact visiting order depends on the order in which neighbours are stored. In the example below, we visit neighbours in the listed order.

````python
# ============================================
# 1️⃣8️⃣ Breadth-First Search (BFS)
# ============================================

from collections import deque

graph = {
    "A": ["B", "C"],
    "B": ["A", "D", "E"],
    "C": ["A", "E"],
    "D": ["B"],
    "E": ["B", "C"],
}

def bfs(graph, start):
    visited = {start}       # Mark when enqueued, so we do not enqueue it repeatedly
    queue = deque([start])
    order = []

    while queue:
        node = queue.popleft()
        order.append(node)

        for neighbour in graph.get(node, []):
            if neighbour not in visited:
                visited.add(neighbour)
                queue.append(neighbour)

    return order

print("BFS order from A:", bfs(graph, "A"))
````

### Why mark nodes as visited when enqueueing?

If a node has several neighbours, more than one node may discover it. Marking it as visited as soon as it enters the queue prevents duplicates from piling up and prevents unnecessary work. It also helps BFS process each reachable node only once.

### What if the graph is disconnected?

BFS from one starting node only visits nodes reachable from that starting node. It does not magically visit a separate disconnected component.

````python
# A disconnected graph example

disconnected_graph = {
    "A": ["B"],
    "B": ["A"],
    "X": ["Y"],
    "Y": ["X"],
}

print("Reachable from A:", bfs(disconnected_graph, "A"))
print("Reachable from X:", bfs(disconnected_graph, "X"))
````

## 1️⃣9️⃣ Application — Shortest Path in an Unweighted Graph

Because BFS explores nodes in increasing number of edges from the starting node, the first time it reaches a target it has found a shortest path measured by **number of edges**. This guarantee applies to unweighted graphs (or graphs where all edges have equal cost), not to arbitrary weighted graphs.

We store each node's parent so that we can rebuild the path after reaching the target.

````python
from collections import deque

def shortest_path_unweighted(graph, start, target):
    if start == target:
        return [start]

    queue = deque([start])
    parent = {start: None}

    while queue:
        node = queue.popleft()
        for neighbour in graph.get(node, []):
            if neighbour not in parent:  # Not visited yet
                parent[neighbour] = node
                if neighbour == target:
                    # Rebuild target -> start, then reverse the path
                    path = [target]
                    while path[-1] != start:
                        path.append(parent[path[-1]])
                    path.reverse()
                    return path
                queue.append(neighbour)

    return None  # Target was not reachable

print("Path A to D:", shortest_path_unweighted(graph, "A", "D"))
print("Path A to X:", shortest_path_unweighted(graph, "A", "X"))
````

### 19.1 Special cases to remember

- If `start == target`, the path is just `[start]`.
- If the target cannot be reached, this function returns `None`.
- The graph is expected to store neighbours in lists, such as `{"A": ["B", "C"]}`.
- If edges have different travel costs, use an algorithm designed for weighted graphs, such as Dijkstra's algorithm for non-negative edge weights, rather than relying on ordinary BFS.

## 2️⃣0️⃣ Implementing a Circular Queue

A **circular queue** is a queue stored in a fixed-size array-like structure. When the rear reaches the last position, it wraps around to the beginning if there is space there.

Circular queues are useful when a program wants a fixed capacity, such as a small buffer. We will implement one using a Python list plus indexes.

Key ideas:

- `front` stores the index of the next item to remove.
- `rear` stores the index where the next item will be inserted.
- `size` tracks how many items are currently stored.
- `% capacity` wraps an index back to the beginning.
- This implementation **raises `OverflowError` when full**; it does not silently discard an old item.

````python
# ============================================
# 2️⃣0️⃣ Circular Queue class
# ============================================

class CircularQueue:
    """A fixed-capacity FIFO queue using circular indexing."""

    def __init__(self, capacity):
        if capacity <= 0:
            raise ValueError("Capacity must be greater than zero")
        self._data = [None] * capacity
        self._capacity = capacity
        self._front = 0
        self._rear = 0
        self._size = 0

    def is_empty(self):
        return self._size == 0

    def is_full(self):
        return self._size == self._capacity

    def __len__(self):
        return self._size

    def enqueue(self, item):
        if self.is_full():
            raise OverflowError("Cannot enqueue: circular queue is full")
        self._data[self._rear] = item
        self._rear = (self._rear + 1) % self._capacity
        self._size += 1

    def dequeue(self):
        if self.is_empty():
            raise IndexError("Cannot dequeue from an empty circular queue")
        item = self._data[self._front]
        self._data[self._front] = None
        self._front = (self._front + 1) % self._capacity
        self._size -= 1
        return item

    def peek(self):
        if self.is_empty():
            raise IndexError("Cannot peek at an empty circular queue")
        return self._data[self._front]

    def __repr__(self):
        items = [self._data[(self._front + i) % self._capacity] for i in range(self._size)]
        return f"CircularQueue({items!r}, capacity={self._capacity})"
````

### 20.1 Use the circular queue

Notice that an item can be removed and a later item can reuse a slot near the beginning of the underlying list. The queue still returns items in FIFO order.

````python
queue = CircularQueue(3)
queue.enqueue("A")
queue.enqueue("B")
queue.enqueue("C")
print(queue, "| full:", queue.is_full())

print("Removed:", queue.dequeue())
queue.enqueue("D")  # Uses a slot after the index wraps around
print("After removing A and adding D:", queue)

print("Dequeue order:", queue.dequeue(), queue.dequeue(), queue.dequeue())
print("Empty now:", queue.is_empty())
````

### 20.2 Circular queue edge cases

A real implementation must distinguish **empty** from **full**. Here the `size` variable does that clearly. `None` is used to clear unused slots for display, but it does not determine whether the queue is empty; `size` does.

````python
queue = CircularQueue(2)

try:
    queue.dequeue()
except IndexError as error:
    print("Empty case:", error)

queue.enqueue(1)
queue.enqueue(2)
try:
    queue.enqueue(3)
except OverflowError as error:
    print("Full case:", error)
````

## 2️⃣1️⃣ Priority Queue — When Importance Matters More Than Arrival Time

A normal queue follows FIFO. A **priority queue** removes the item with the highest priority (or smallest priority number in the convention we choose) rather than always removing the oldest item.

Python's `heapq` module implements a min-heap, which efficiently returns the **smallest** item first. A common pattern is to store tuples such as `(priority, item)`.

In the example, priority `1` is more urgent than priority `3`. This is a different rule from FIFO, so use a priority queue only when importance should determine the next item.

````python
# ============================================
# 2️⃣1️⃣ Priority queue with heapq
# ============================================

import heapq

priority_jobs = []
heapq.heappush(priority_jobs, (3, "weekly report"))
heapq.heappush(priority_jobs, (1, "server outage"))
heapq.heappush(priority_jobs, (2, "reply to email"))

while priority_jobs:
    priority, job = heapq.heappop(priority_jobs)
    print(f"Priority {priority}: {job}")
````

### 21.1 A subtle priority-queue case: equal priorities

If tuples are `(priority, item)`, Python compares the second tuple element when priorities tie. That can raise a `TypeError` if those item values cannot be compared. A common solution is to add an increasing counter so every entry has a predictable tie-breaker.

````python
import heapq
from itertools import count

counter = count()
priority_jobs = []

# Entries are: priority, arrival counter, job description
heapq.heappush(priority_jobs, (2, next(counter), {"task": "write notes"}))
heapq.heappush(priority_jobs, (2, next(counter), {"task": "check email"}))
heapq.heappush(priority_jobs, (1, next(counter), {"task": "fix urgent bug"}))

while priority_jobs:
    priority, _order, job = heapq.heappop(priority_jobs)
    print(priority, job["task"])
````

## 2️⃣2️⃣ Advanced Queue Pattern — Sliding-Window Maximum

Suppose you have numbers and want the maximum for every group of `k` consecutive numbers. A simple approach scans each window from scratch, but that can be slow for large inputs.

A **monotonic deque** keeps candidate indexes in decreasing value order. It discards indexes that have left the window and removes smaller values from the rear because they cannot become a useful maximum while a larger, newer value remains ahead of them.

This technique can solve the problem in **O(n)** time.

````python
from collections import deque

def sliding_window_max(nums, k):
    if k <= 0:
        raise ValueError("k must be positive")
    if k > len(nums):
        return []

    candidates = deque()  # Stores indexes; values decrease from front to rear
    result = []

    for i, value in enumerate(nums):
        # Remove indexes that are no longer inside the current window
        while candidates and candidates[0] <= i - k:
            candidates.popleft()

        # Remove smaller or equal values from the rear
        while candidates and nums[candidates[-1]] <= value:
            candidates.pop()

        candidates.append(i)

        # A full window exists starting from index k - 1
        if i >= k - 1:
            result.append(nums[candidates[0]])

    return result

numbers = [1, 3, -1, -3, 5, 3, 6, 7]
print("Numbers:", numbers)
print("Window size 3 maximums:", sliding_window_max(numbers, 3))
````

### 22.1 Why do we store indexes instead of only values?

Indexes tell us **when** an item leaves the window. If we only store values, we cannot reliably determine whether a repeated value has moved outside the current window. Storing indexes lets us remove expired candidates precisely.

### 22.2 Edge cases

- `k <= 0` is invalid, so the function raises `ValueError`.
- If `k` is greater than the list length, no complete window exists, so the result is `[]`.
- Duplicate values are safe. The code removes older values that are less than or equal to the current value because the newer value lasts longer in future windows.

````python
print("k = 1:", sliding_window_max([4, 2, 8], 1))
print("Window larger than input:", sliding_window_max([4, 2], 3))
print("Repeated values:", sliding_window_max([5, 5, 4, 5], 2))
````

## 2️⃣3️⃣ Advanced Queue Pattern — First Non-Repeating Character in a Stream

Imagine characters arrive one at a time. After each character, you want to know the **first character that has appeared exactly once so far**.

We can combine:

- A dictionary (`counts`) to count character appearances.
- A queue to keep characters in arrival order.
- A loop that removes characters from the front while they have appeared more than once.

When the front character has count `1`, it is the first non-repeating character. If the queue becomes empty, there is no such character at that moment.

````python
from collections import deque, Counter

def first_non_repeating_after_each_char(text):
    counts = Counter()
    candidates = deque()
    answers = []

    for char in text:
        counts[char] += 1
        candidates.append(char)

        while candidates and counts[candidates[0]] > 1:
            candidates.popleft()

        answers.append(candidates[0] if candidates else None)

    return answers

stream = "aabc"
print("Stream:", stream)
print("First non-repeating after each character:", first_non_repeating_after_each_char(stream))
````

**Note:** This example treats uppercase and lowercase letters as different characters (`A` and `a` are not the same). It also treats spaces and punctuation as normal characters. You can normalize the text first if your task requires different behaviour.

## 2️⃣4️⃣ Queue Rules, Special Cases, and Common Mistakes

### Quick checklist

1. **FIFO:** The oldest queued item leaves first.
2. **Enqueue at the rear; dequeue from the front.** Keep the ends consistent.
3. **Empty queue:** Do not access `queue[0]`, call `pop(0)`, or call `popleft()` without handling the possibility that no item exists.
4. **Prefer `deque` for a normal queue.** A list's `pop(0)` may shift many elements.
5. **Do not confuse a normal queue with a priority queue.** FIFO is based on arrival order; priority queues use priority.
6. **Mark BFS nodes visited when enqueuing them** to avoid duplicate entries.
7. **Queue state changes after dequeue.** Save the returned item if you still need it.
8. **Bounded `deque(maxlen=...)` drops items from the opposite end when full.** That is useful for rolling history but is not the same as a queue that raises an error on overflow.
9. **`queue.Queue.empty()` is not a safe synchronization strategy across threads.** Another thread may change the queue immediately afterward; use appropriate blocking operations and task tracking.
10. **A sentinel such as `None` is only safe if it cannot be confused with real data.**

### ✅ Quick Summary — Python Queue

- **Queue rule:** FIFO — First In, First Out.
- **Enqueue:** Add an item at the rear.
- **Dequeue:** Remove and return the front item.
- **Peek:** Look at the front without removing it.
- **Recommended general-purpose queue:** `collections.deque`.
- **Common deque methods:** `append()`, `popleft()`, `appendleft()`, and `pop()`.
- **Avoid repeated `list.pop(0)` for large queues:** it may take O(n) time because items shift.
- **Thread coordination:** use `queue.Queue` when multiple threads need a synchronized work queue.
- **Circular queue:** uses a fixed capacity and wraps indexes around.
- **Priority queue:** processes by priority, not ordinary FIFO order.
- **BFS:** uses a queue to explore graphs in layers and find shortest paths in unweighted graphs.
- **Always think about edge cases:** empty queue, full capacity, disconnected graph, duplicate values, and what should happen when an operation is invalid.
