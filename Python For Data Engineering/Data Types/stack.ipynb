## 1️⃣ What is a Stack?

A **stack** is a data structure that stores items in a particular order. The most important rule is:

> **LIFO = Last In, First Out**

The item added **last** is the item removed **first**.

### Everyday example: a stack of plates

Imagine you put three plates on top of one another:

1. Put Plate A down.
2. Put Plate B on A.
3. Put Plate C on B.

Plate C is on the top, so you normally remove C first, then B, then A. This is exactly how a stack works.

A stack is not a special Python keyword. It is a **way of organising data**. In Python, we commonly use a list to implement it.

### A stack drawn as boxes

Suppose we push `10`, then `20`, then `30`:

```text
          TOP
       +--------+
       |   30   |  ← added last; removed first
       +--------+
       |   20   |
       +--------+
       |   10   |  ← added first; removed last
       +--------+
```

When we `pop()`, only the top item (`30`) is removed. The next top becomes `20`.

**Important:** A stack has a *top*. You should add and remove items at that same end.

---
## 2️⃣ Stack Vocabulary and Operations

| Term | Easy meaning | Example |
|---|---|---|
| **Stack** | The whole collection of items | `[10, 20, 30]` |
| **Top** | The item that would come out next | `30` |
| **Push** | Add an item to the top | Push `40` |
| **Pop** | Remove and return the top item | Pop returns `40` |
| **Peek / Top** | Look at the top without removing it | Peek returns `30` |
| **is_empty** | Check whether there are no items | Empty stack → `True` |
| **Size** | Number of items currently stored | Three items → `3` |
| **Underflow** | Trying to remove/read an item when empty | Pop an empty stack |

In Python lists, we usually use:
- `append(item)` to **push** an item.
- `pop()` to **pop** the last item.
- `stack[-1]` to **peek** at the last item without removing it.

## 3️⃣ Creating a Stack with a Python List

A normal Python list can act as a stack. For this notebook, we will keep the **top of the stack at the right-hand end** of the list.

```text
Python list:  [10, 20, 30]
                         ↑
                        TOP
```

This choice makes adding and removing from the top efficient because `append()` and `pop()` at the end are generally **O(1) amortized** operations.

An empty stack is simply an empty list: `stack = []`.

````python
# ============================================
# 3️⃣ Creating a Stack
# ============================================

stack = []  # An empty list that we will use as a stack
print("Stack:", stack)
print("Type:", type(stack))
print("Is it empty?", len(stack) == 0)
````

### 4️⃣ Push — adding items

**Push** means adding a new item to the top of the stack.

With our list-based stack, use `append()`:

**Syntax:** `stack.append(item)`

The item is placed at the right-hand end, which we have chosen as the top.

````python
# ============================================
# 4️⃣ Push — add items to the TOP
# ============================================

stack = []
stack.append(10)  # Push 10
stack.append(20)  # Push 20 on top of 10
stack.append(30)  # Push 30 on top of 20

print("Stack:", stack)   # [10, 20, 30]
print("Top item:", stack[-1])  # 30

# Read the list from left to right: 10 was added first, 30 was added last.
# Therefore, 30 is the next item that would be popped.
````

### 5️⃣ Pop — removing the top item

**Pop** removes **and returns** the top item.

**Syntax:** `item = stack.pop()`

For our stack, `pop()` without an index removes the rightmost list element. After popping, the stack becomes shorter by one item.

> `pop()` is different from just looking at an item: it changes the stack.

````python
# ============================================
# 5️⃣ Pop — remove and return the TOP item
# ============================================

stack = [10, 20, 30]
removed_item = stack.pop()

print("Removed item:", removed_item)  # 30
print("Stack after pop:", stack)      # [10, 20]

# We can store the returned value and use it later.
next_removed = stack.pop()
print("Removed next:", next_removed)  # 20
print("Stack now:", stack)            # [10]
````

### 6️⃣ Peek — view the top without removing it

**Peek** means looking at the top item while leaving the stack unchanged.

For a non-empty list-based stack, `stack[-1]` gives the last item.

| Expression | What it does |
|---|---|
| `stack[-1]` | Reads the top item; does not remove it |
| `stack.pop()` | Removes and returns the top item |

`stack[-1]` raises `IndexError` when the stack is empty, so check first when emptiness is possible.

````python
# ============================================
# 6️⃣ Peek — inspect without removing
# ============================================

stack = [10, 20, 30]
print("Top item:", stack[-1])
print("Stack remains:", stack)  # Still [10, 20, 30]

# Compare with pop():
removed = stack.pop()
print("pop() returned:", removed)
print("Stack after pop():", stack)  # Now [10, 20]
````

### 7️⃣ Check whether a stack is empty and find its size

Two common checks are:

- `len(stack) == 0` — checks whether the stack has zero items.
- `len(stack)` — returns the number of items.

Python also lets you use a list directly in an `if` condition: an empty list behaves as `False`, and a non-empty list behaves as `True`. The explicit `len(stack) == 0` version can be easier for beginners to read.

````python
# ============================================
# 7️⃣ is_empty and size
# ============================================

stack = []
print("Size:", len(stack))                 # 0
print("Empty?", len(stack) == 0)           # True

stack.append("Python")
stack.append("SQL")
print("Stack:", stack)
print("Size:", len(stack))                 # 2
print("Empty?", len(stack) == 0)           # False

# Python truth-value shortcut:
if stack:
    print("The stack contains at least one item.")
else:
    print("The stack is empty.")
````

## 8️⃣ See LIFO in Action — Step by Step

Let us follow each operation carefully. The **rightmost item is always the top** in our examples.

| Operation | Stack after operation | Top |
|---|---|---|
| Start | `[]` | No top |
| Push 5 | `[5]` | `5` |
| Push 8 | `[5, 8]` | `8` |
| Push 2 | `[5, 8, 2]` | `2` |
| Pop | `[5, 8]` | `8` |
| Push 9 | `[5, 8, 9]` | `9` |
| Pop | `[5, 8]` | `8` |

Notice that the most recently pushed value, `9`, is removed first. That is the LIFO rule.

````python
# ============================================
# 8️⃣ LIFO walkthrough
# ============================================

stack = []

stack.append(5)
print("After push 5:", stack)

stack.append(8)
print("After push 8:", stack)

stack.append(2)
print("After push 2:", stack)

print("Popped:", stack.pop())  # 2
print("Stack:", stack)

stack.append(9)
print("After push 9:", stack)

print("Popped:", stack.pop())  # 9
print("Final stack:", stack)
````

## 9️⃣ Important Edge Case — Popping an Empty Stack

A stack can become empty after one or more `pop()` operations. If we then call `pop()` again, Python raises an `IndexError` because there is no item to remove.

This situation is often called **stack underflow**.

Do not assume a stack always has an item. In real programs, the stack may be empty because of user input, a loop, or a previous operation.

````python
# ============================================
# 9️⃣ Empty-stack safety
# ============================================

empty_stack = []

# Unsafe if the stack may be empty:
# empty_stack.pop()  # Would raise IndexError

# Safe pattern 1: check before popping
if empty_stack:
    print("Removed:", empty_stack.pop())
else:
    print("Cannot pop: the stack is empty.")

# Safe pattern 2: handle the possible exception
try:
    item = empty_stack.pop()
    print("Removed:", item)
except IndexError:
    print("Caught IndexError: there was no item to pop.")
````

### 9.1 Safe peek and safe pop patterns

There is no single rule that fits every program:

- **Check first** when an empty stack is an expected possibility.
- **Use `try`/`except`** when you want to handle an operation that might fail.
- **Raise an error** when an empty stack means the caller has used your data structure incorrectly.

Avoid silently returning a fake value such as `0` for every failed pop. A valid stack item could itself be `0`, so the caller might not know whether a pop really happened.

````python
# ============================================
# Safe helper functions for a plain list
# ============================================

def safe_peek(stack):
    """Return the top item, or None when the stack is empty."""
    if not stack:
        return None
    return stack[-1]


def safe_pop(stack):
    """Return the top item, or None when the stack is empty."""
    if not stack:
        return None
    return stack.pop()

sample_stack = [10, 20]
print("Peek:", safe_peek(sample_stack))  # 20; does not remove it
print("After peek:", sample_stack)       # [10, 20]
print("Pop:", safe_pop(sample_stack))    # 20; removes it
print("After pop:", sample_stack)        # [10]
print("Empty peek:", safe_peek([]))      # None
````

**A small design warning:** These helper functions use `None` to represent “no item”. That is fine only when `None` is not also a valid value you need to distinguish from a missing item. In a larger program, raising an exception or returning a separate success flag may be clearer.

## 🔟 Useful Python Tricks with a List-Based Stack

| Task | Code | Explanation |
|---|---|---|
| Push | `stack.append(x)` | Add `x` at the top |
| Pop | `stack.pop()` | Remove and return the top |
| Peek | `stack[-1]` | Read the top without removing |
| Size | `len(stack)` | Count items |
| Is empty? | `not stack` | True when empty |
| Remove all items | `stack.clear()` | Empty the stack |
| See a copy | `stack.copy()` | Make a shallow copy of the list |
| Pop everything | `while stack:` | Continue until empty |

Remember: in this notebook, the **top is at the right**. If you change that convention, your operations and explanations must change too.

````python
# ============================================
# 🔟 Useful stack patterns
# ============================================

stack = [10, 20, 30]

# Pop all items — last pushed comes out first
print("Pop all items:")
while stack:
    print(stack.pop())  # 30, then 20, then 10

print("After while loop:", stack)  # []

# clear() removes all items at once
stack = [1, 2, 3]
stack.clear()
print("After clear:", stack)
````

### A note about copying and sharing a stack

A list variable refers to a list object. Writing `other = stack` does **not** make a new stack; both variable names refer to the same list. Use `stack.copy()` if you need a separate shallow copy of the outer list.

For ordinary stack items such as numbers and strings, a shallow copy is usually enough. If the items are themselves mutable objects, both copies can still refer to those same objects.

````python
# ============================================
# Alias versus a separate shallow copy
# ============================================

original_stack = [1, 2, 3]
alias_stack = original_stack       # Same list object!
copy_stack = original_stack.copy() # A separate list object

alias_stack.append(4)
copy_stack.append(99)

print("Original:", original_stack)  # [1, 2, 3, 4]
print("Alias:", alias_stack)        # [1, 2, 3, 4] — same object
print("Copy:", copy_stack)          # [1, 2, 3, 99] — separate list
````

## 1️⃣1️⃣ Implementing a Reusable `Stack` Class

So far, we have used a plain list directly. This is good for learning, but a **class** lets us package stack behaviour behind clear method names.

For example, we can write `my_stack.push(10)` instead of remembering that `append(10)` means push.

Our class will provide:
- `push(item)` — add an item.
- `pop()` — remove and return the top; raise `IndexError` if empty.
- `peek()` — read the top; raise `IndexError` if empty.
- `is_empty()` — return `True` or `False`.
- `size()` — return the number of items.
- `clear()` — remove all items.

We store items in a private-by-convention attribute named `_items`. The underscore signals that callers should normally use the class methods instead of changing this list directly.

````python
# ============================================
# 1️⃣1️⃣ Define a reusable Stack class
# ============================================

class Stack:
    """A simple stack using a Python list; the right end is the top."""

    def __init__(self):
        self._items = []

    def push(self, item):
        """Add an item to the top of the stack."""
        self._items.append(item)

    def pop(self):
        """Remove and return the top item; fail clearly if empty."""
        if self.is_empty():
            raise IndexError("pop from an empty stack")
        return self._items.pop()

    def peek(self):
        """Return the top item without removing it."""
        if self.is_empty():
            raise IndexError("peek from an empty stack")
        return self._items[-1]

    def is_empty(self):
        """Return True when the stack has no items."""
        return len(self._items) == 0

    def size(self):
        """Return the number of items in the stack."""
        return len(self._items)

    def clear(self):
        """Remove all items from the stack."""
        self._items.clear()

    def __len__(self):
        """Allow len(stack_object)."""
        return len(self._items)

    def __bool__(self):
        """Make bool(stack_object) False when empty."""
        return not self.is_empty()

    def __repr__(self):
        """Show the stored items when inspecting the object."""
        return f"Stack({self._items!r})"
````

### 11.1 Using the `Stack` class

After defining the class, create an object with `Stack()`. This object has its own independent list of items.

````python
# ============================================
# Use our Stack class
# ============================================

numbers_stack = Stack()
print(numbers_stack)                 # Stack([])
print(numbers_stack.is_empty())      # True

numbers_stack.push(10)
numbers_stack.push(20)
numbers_stack.push(30)

print(numbers_stack)                 # Stack([10, 20, 30])
print("Top:", numbers_stack.peek())  # 30
print("Size:", numbers_stack.size()) # 3
print("len():", len(numbers_stack))  # 3

print("Popped:", numbers_stack.pop())
print("After pop:", numbers_stack)   # Stack([10, 20])
````

### 11.2 Handling an empty custom stack

Our `Stack` class raises a clear `IndexError` when `pop()` or `peek()` is called on an empty stack. The caller can catch that error when the situation is expected.

````python
# ============================================
# Empty stack behaviour in the custom class
# ============================================

empty_custom_stack = Stack()

for operation_name, operation in [
    ("pop", empty_custom_stack.pop),
    ("peek", empty_custom_stack.peek),
]:
    try:
        operation()
    except IndexError as error:
        print(f"Cannot {operation_name}: {error}")

print("Still empty?", empty_custom_stack.is_empty())
````

### 11.3 Why methods are useful

A class hides the implementation detail. If we later change how the stack stores values, calling code can still use `push()`, `pop()`, and `peek()`.

**Beginner note:** `self` means “this particular stack object”. Two objects made from the same class have separate data.

````python
# ============================================
# Two different stack objects do not share data
# ============================================

stack_a = Stack()
stack_b = Stack()

stack_a.push("A")
stack_a.push("B")
stack_b.push(100)

print("stack_a:", stack_a)  # Stack(['A', 'B'])
print("stack_b:", stack_b)  # Stack([100])
````

## 1️⃣2️⃣ Using `collections.deque` as a Stack

Python's list is a good default for a stack. Another option is `collections.deque` (pronounced roughly “deck”). It supports efficient additions and removals at both ends.

Use `append()` to push and `pop()` to remove from the right, just as we did with a list. A `deque` is particularly handy when your program needs fast operations at **both ends**. For a stack used only at one end, a list is usually simple and suitable.

````python
# ============================================
# 1️⃣2️⃣ Stack using collections.deque
# ============================================

from collections import deque

deque_stack = deque()
deque_stack.append("first")
deque_stack.append("second")
deque_stack.append("third")

print("Stack:", deque_stack)
print("Pop:", deque_stack.pop())
print("After pop:", deque_stack)
print("Peek:", deque_stack[-1])
````

### List stack vs deque stack

| Feature | `list` | `collections.deque` |
|---|---|---|
| Push at right | `append()` | `append()` |
| Pop from right | `pop()` | `pop()` |
| Peek at right | `stack[-1]` | `stack[-1]` |
| Remove from left | `pop(0)` is O(n) | `popleft()` is O(1) |
| Typical simple stack | Excellent choice | Also works well |
| Need fast operations at both ends | Not ideal at the left end | Designed for this |

Do **not** use `list.pop(0)` for a normal stack. It removes the first item and shifts the other list elements, which is slower for large lists and does not follow our chosen top-at-the-right convention.

## 1️⃣3️⃣ Time and Space Complexity

Complexity describes how the amount of work grows as the number of items grows. Let **n** be the number of items in the stack.

| Operation | Typical list-stack cost | Why |
|---|---:|---|
| `append(item)` / push | O(1) amortized | Add at the right end |
| `pop()` | O(1) | Remove from the right end |
| Peek with `stack[-1]` | O(1) | Access one position directly |
| `len(stack)` | O(1) | Length is maintained by Python |
| Check empty | O(1) | A simple length/truth-value check |
| Search for a value | O(n) | May need to inspect every item |
| Copy all items | O(n) | Each item reference is copied |
| Store n items | O(n) space | Each item needs storage |

**What does O(1) amortized mean?** Most pushes are constant-time. Occasionally, Python may need to allocate more space for the list, so we describe the average cost over a sequence of pushes as amortized O(1).

For most student problems, remember: **push, pop, and peek at the top are O(1); searching through the stack is O(n).**

## 1️⃣4️⃣ Application 1 — Reverse a String

One classic use of a stack is reversing a sequence. If we push each character and then pop every character, the characters come out in reverse order because of LIFO.

Example: `PYTHON` is pushed character by character. The final character, `N`, is popped first.

````python
# ============================================
# 1️⃣4️⃣ Reverse a string using a stack
# ============================================

text = "PYTHON"
char_stack = []

# Push every character
for character in text:
    char_stack.append(character)

reversed_text = ""

# Pop characters in reverse order
while char_stack:
    reversed_text += char_stack.pop()

print("Original:", text)
print("Reversed:", reversed_text)  # NOHTYP
````

### A Python shortcut — slicing

Python can reverse a string directly with `text[::-1]`. That is convenient, but the stack version teaches **why a stack is useful** and prepares you for problems where slicing is not enough.

````python
text = "stack"
print(text[::-1])  # kcats

# Same result using a stack
temporary_stack = list(text)
print("".join(temporary_stack.pop() for _ in range(len(temporary_stack))))
````

## 1️⃣5️⃣ Application 2 — Reverse a List

We can use the same idea to reverse a list. This version creates a **new reversed list** and leaves the original list unchanged.

````python
# ============================================
# 1️⃣5️⃣ Reverse a list using a stack
# ============================================

original_numbers = [10, 20, 30, 40]
number_stack = []

for number in original_numbers:
    number_stack.append(number)

reversed_numbers = []
while number_stack:
    reversed_numbers.append(number_stack.pop())

print("Original:", original_numbers)   # [10, 20, 30, 40]
print("Reversed:", reversed_numbers)   # [40, 30, 20, 10]
````

## 1️⃣6️⃣ Application 3 — Check Balanced Brackets

A common programming task is checking whether brackets are correctly paired and nested.

Examples:
- `"(a + b)"` → balanced
- `"((a + b)"` → not balanced; one opening bracket is still waiting
- `"(a + b))"` → not balanced; a closing bracket appears without a matching opening bracket

### The stack idea

1. When you see an opening bracket such as `(`, `[`, or `{`, push it.
2. When you see a closing bracket, there must be a matching opening bracket on top.
3. Pop the matching opening bracket.
4. At the end, the stack must be empty.

The order matters. For example, `"([)]"` is **not** balanced, even though it has two opening and two closing brackets.

````python
# ============================================
# 1️⃣6️⃣ Balanced brackets — parentheses only
# ============================================

def is_parentheses_balanced(expression):
    stack = []

    for character in expression:
        if character == "(":
            stack.append(character)
        elif character == ")":
            if not stack:          # No opening parenthesis to match
                return False
            stack.pop()

    return len(stack) == 0        # No unmatched opening parenthesis remains

print(is_parentheses_balanced("(a + b)"))    # True
print(is_parentheses_balanced("((a + b)"))   # False
print(is_parentheses_balanced("(a + b))"))   # False
print(is_parentheses_balanced("a + b"))      # True
````

### 16.1 Handle `()`, `[]`, and `{}`

Now we will check three bracket types. When a closing bracket appears, it must match the **most recent unmatched opening bracket** — the one at the top of the stack.

````python
# ============================================
# Balanced (), [], and {}
# ============================================

def are_brackets_balanced(expression):
    stack = []
    matching_open = {
        ")": "(",
        "]": "[",
        "}": "{",
    }
    opening_brackets = set(matching_open.values())

    for character in expression:
        if character in opening_brackets:
            stack.append(character)
        elif character in matching_open:
            if not stack:
                return False
            if stack[-1] != matching_open[character]:
                return False
            stack.pop()

    return not stack

examples = ["([]{})", "([{}])", "([)]", "((]", "hello", "{"]
for example in examples:
    print(f"{example!r:10} -> {are_brackets_balanced(example)}")
````

**Why compare with the top?** Imagine the stack contains `['(', '[']`. The most recent unmatched opening bracket is `[`. A `)` cannot close that `[`, so the expression is invalid. This is why counting the total number of each bracket is not enough.

## 1️⃣7️⃣ Application 4 — Undo Functionality

Text editors and other applications often store previous states so the user can undo an action. A simple model is:

- Before making a change, push the current state onto an undo stack.
- To undo, pop the most recent saved state.

This example uses strings as simplified document states. It is only a teaching model; real editors may store smaller change records to save memory.

````python
# ============================================
# 1️⃣7️⃣ A simple undo stack
# ============================================

document = ""
undo_stack = []

# Before each edit, remember the current state
undo_stack.append(document)
document = "Hello"
print("After edit 1:", document)

undo_stack.append(document)
document = "Hello, Python"
print("After edit 2:", document)

# Undo the most recent edit
if undo_stack:
    document = undo_stack.pop()
print("After undo:", document)  # Hello
````

### 17.1 Browser history with two stacks

A simplified browser can use two stacks:
- **Back stack** stores pages we can return to.
- **Forward stack** stores pages we moved forward from.

When visiting a new page, clear the forward stack because that old forward path is no longer the current path. This is a simplified model, but it demonstrates how stacks can work together.

````python
# ============================================
# 17.1 Simplified browser history
# ============================================

back_stack = []
forward_stack = []
current_page = "home"

# Visit a new page
back_stack.append(current_page)
current_page = "courses"
forward_stack.clear()

back_stack.append(current_page)
current_page = "python"
forward_stack.clear()
print("Current page:", current_page)

# Go back one page
if back_stack:
    forward_stack.append(current_page)
    current_page = back_stack.pop()
print("After Back:", current_page)  # courses

# Go forward one page
if forward_stack:
    back_stack.append(current_page)
    current_page = forward_stack.pop()
print("After Forward:", current_page)  # python
````

## 1️⃣8️⃣ The Call Stack and Recursion

Python also uses a **call stack** while functions call other functions. Each active function call has information Python needs to resume that function, such as its local variables and where it should continue.

For example, when `main()` calls `first()`, and `first()` calls `second()`, `second()` must finish before `first()` continues. This follows the same last-in, first-out idea.

```text
main() calls first()
first() calls second()

Top of call stack: second()  ← finishes first
                    first()
                    main()   ← finishes last
```

Recursion is when a function calls itself. Each recursive call adds another active call until a **base case** stops the process.

````python
# ============================================
# 1️⃣8️⃣ Recursion example — countdown
# ============================================

def countdown(n):
    # Base case: stop calling the function when n reaches 0
    if n == 0:
        print("Done!")
        return

    print(n)
    countdown(n - 1)  # The function calls itself with a smaller value

countdown(3)  # Prints 3, 2, 1, Done!
````

### Recursion special cases

- A **base case** is necessary; otherwise, the function may keep calling itself until Python raises `RecursionError`.
- Each recursive call uses call-stack space.
- For very deep repetition, a loop may be safer than recursion in Python.
- Recursion is not the same thing as manually creating a list-based stack, but both involve a last-in, first-out pattern.

A common beginner bug is forgetting to make progress toward the base case, such as calling `countdown(n)` again instead of `countdown(n - 1)`. Do not run a recursive function that has no valid stopping condition.

## 1️⃣9️⃣ Advanced Application — Evaluate a Postfix Expression

An expression can be written in different forms. In **postfix notation**, an operator comes *after* its operands.

| Infix (usual form) | Postfix |
|---|---|
| `2 + 3` | `2 3 +` |
| `2 + 3 * 4` | `2 3 4 * +` |
| `(2 + 3) * 4` | `2 3 + 4 *` |

To evaluate postfix notation, scan tokens from left to right:
1. If a token is a number, push it.
2. If it is an operator, pop the **right operand first**, then the **left operand**.
3. Apply the operator and push the result.
4. At the end, a valid expression should leave exactly one result.

**Important:** For subtraction and division, operand order matters. If the stack holds left operand `8` and then right operand `2`, the operation is `8 - 2`, not `2 - 8`. We therefore write `right = pop()` first, then `left = pop()`.

````python
# ============================================
# 1️⃣9️⃣ Evaluate a space-separated postfix expression
# Supports +, -, *, and /
# ============================================

def evaluate_postfix(expression):
    stack = []
    operators = {"+", "-", "*", "/"}

    for token in expression.split():
        if token not in operators:
            stack.append(float(token))
            continue

        # The second value popped is the left operand
        if len(stack) < 2:
            raise ValueError("Invalid postfix expression: not enough operands")
        right = stack.pop()
        left = stack.pop()

        if token == "+":
            result = left + right
        elif token == "-":
            result = left - right
        elif token == "*":
            result = left * right
        else:
            if right == 0:
                raise ZeroDivisionError("division by zero in expression")
            result = left / right

        stack.append(result)

    if len(stack) != 1:
        raise ValueError("Invalid postfix expression: extra operands remain")
    return stack[0]

print(evaluate_postfix("2 3 +"))        # 5.0
print(evaluate_postfix("2 3 4 * +"))    # 14.0
print(evaluate_postfix("8 2 -"))        # 6.0
print(evaluate_postfix("8 2 /"))        # 4.0
````

### 19.1 Why do we use a space between postfix tokens?

In this beginner-friendly implementation, `expression.split()` separates tokens at whitespace. Write `"12 3 +"`, not `"123+"`, so Python can clearly identify the number `12`, the number `3`, and the operator `+`.

The function also validates common errors: too few operands, too many values left at the end, and division by zero.

## 2️⃣0️⃣ Advanced Application — Convert Infix to Postfix

In **infix notation**, an operator appears between values, such as `2 + 3 * 4`. To convert it to postfix, we use a stack for operators and parentheses.

This version supports `+`, `-`, `*`, `/`, and parentheses. **Write a space between every token** (for example, `"( 2 + 3 ) * 4"`). It assumes each operand is a single number or variable name and does not handle unary minus or exponentiation.

### Main rules
1. Send operands directly to the output.
2. Push `(` onto the operator stack.
3. When `)` appears, pop operators to the output until `(` is found; discard the parentheses.
4. For an operator, pop operators with greater than or equal precedence before pushing the new operator.
5. At the end, pop all remaining operators to the output.

````python
# ============================================
# 2️⃣0️⃣ Infix to postfix (space-separated tokens)
# ============================================

def infix_to_postfix(expression):
    precedence = {"+": 1, "-": 1, "*": 2, "/": 2}
    operators = set(precedence)
    output = []
    operator_stack = []

    tokens = expression.split()
    if not tokens:
        raise ValueError("Expression cannot be empty")

    for token in tokens:
        if token == "(":
            operator_stack.append(token)
        elif token == ")":
            while operator_stack and operator_stack[-1] != "(":
                output.append(operator_stack.pop())
            if not operator_stack:
                raise ValueError("Mismatched parentheses")
            operator_stack.pop()  # Remove the matching '('
        elif token in operators:
            while (
                operator_stack
                and operator_stack[-1] in operators
                and precedence[operator_stack[-1]] >= precedence[token]
            ):
                output.append(operator_stack.pop())
            operator_stack.append(token)
        else:
            # For this simple version, every non-operator token is an operand.
            output.append(token)

    while operator_stack:
        operator = operator_stack.pop()
        if operator == "(":
            raise ValueError("Mismatched parentheses")
        output.append(operator)

    return " ".join(output)

print(infix_to_postfix("2 + 3 * 4"))       # 2 3 4 * +
print(infix_to_postfix("( 2 + 3 ) * 4"))   # 2 3 + 4 *
print(infix_to_postfix("8 - 6 / 2"))       # 8 6 2 / -
````

### 20.1 Connect conversion and evaluation

The previous function converts the expression, and `evaluate_postfix()` evaluates the result. Together, they show a practical use for stacks in expression processing.

````python
infix_expression = "( 2 + 3 ) * 4"
postfix_expression = infix_to_postfix(infix_expression)
answer = evaluate_postfix(postfix_expression)

print("Postfix:", postfix_expression)
print("Answer:", answer)  # 20.0
````

## 2️⃣1️⃣ Advanced Pattern — The Next Greater Element

For each value in a list, the **Next Greater Element** is the first value to its right that is strictly greater. If no greater value appears to its right, the answer is `-1`.

Example:

```text
Numbers: [2, 1, 2, 4, 3]
Answer:  [4, 2, 4, -1, -1]
```

A **monotonic stack** is a stack kept in a particular order. Here we keep indexes whose next greater element has not yet been found. When a larger number appears, it answers the waiting smaller numbers on top of the stack.

````python
# ============================================
# 2️⃣1️⃣ Next Greater Element — monotonic stack
# ============================================

def next_greater_elements(numbers):
    result = [-1] * len(numbers)  # Default: no greater element found
    index_stack = []              # Stores indexes, not the values themselves

    for current_index, current_value in enumerate(numbers):
        # Resolve previous values that are smaller than the current value
        while (
            index_stack
            and numbers[index_stack[-1]] < current_value
        ):
            previous_index = index_stack.pop()
            result[previous_index] = current_value

        index_stack.append(current_index)

    return result

sample_numbers = [2, 1, 2, 4, 3]
print("Numbers:", sample_numbers)
print("Next greater:", next_greater_elements(sample_numbers))
# [4, 2, 4, -1, -1]
````

### Why store indexes instead of values?

We must write the answer into the correct position in `result`. An index tells us **which element** is waiting for a greater value. It also lets us look up that element with `numbers[index]`.

Each index is pushed once and popped at most once, so this algorithm runs in **O(n)** time, even though it contains a `while` loop inside a `for` loop. The total number of stack operations is linear.

## 2️⃣2️⃣ Advanced Pattern — A Min Stack

A normal stack can tell us its top quickly, but finding its minimum using `min(stack)` takes O(n) time. Sometimes we want `get_min()` in O(1) time.

A **Min Stack** keeps a second stack of minimum values. Whenever a new item is pushed, we also record the minimum so far. When an item is popped, its corresponding minimum record is popped as well.

This implementation stores a minimum at every level, which makes the rule easy to understand. It uses extra O(n) space.

````python
# ============================================
# 2️⃣2️⃣ Min Stack — O(1) get_min()
# ============================================

class MinStack:
    def __init__(self):
        self._values = []
        self._minimums = []

    def push(self, value):
        self._values.append(value)
        if not self._minimums:
            self._minimums.append(value)
        else:
            self._minimums.append(min(value, self._minimums[-1]))

    def pop(self):
        if not self._values:
            raise IndexError("pop from an empty MinStack")
        self._minimums.pop()
        return self._values.pop()

    def top(self):
        if not self._values:
            raise IndexError("top from an empty MinStack")
        return self._values[-1]

    def get_min(self):
        if not self._minimums:
            raise IndexError("minimum from an empty MinStack")
        return self._minimums[-1]

    def is_empty(self):
        return not self._values

minimum_stack = MinStack()
for value in [5, 2, 7, 1, 3]:
    minimum_stack.push(value)
    print(f"Pushed {value}; current minimum = {minimum_stack.get_min()}")

print("Top:", minimum_stack.top())
print("Popped:", minimum_stack.pop())
print("Minimum after pop:", minimum_stack.get_min())  # 1
````

### What happens with duplicate minimum values?

The implementation records the minimum at every stack level, including duplicates. This means that if the minimum value is pushed twice, popping one copy still leaves the other minimum record in place. Test it below.

````python
duplicate_min_stack = MinStack()
for value in [4, 2, 2, 5]:
    duplicate_min_stack.push(value)

print("Minimum:", duplicate_min_stack.get_min())  # 2
duplicate_min_stack.pop()  # Removes 5
print("Minimum:", duplicate_min_stack.get_min())  # 2
duplicate_min_stack.pop()  # Removes one 2
print("Minimum:", duplicate_min_stack.get_min())  # Still 2
````

## 2️⃣3️⃣ Stack Rules, Special Cases, and Common Mistakes

| Mistake | Why it is a problem | Better approach |
|---|---|---|
| Calling `pop()` on an empty stack | Raises `IndexError` | Check emptiness or handle the exception |
| Using `stack[0]` as the top | In this notebook, top is at the right | Use `stack[-1]` to peek |
| Using `pop(0)` for a stack | Removes the left end and shifts items | Use `pop()` at the right end |
| Thinking `peek` removes an item | Peek is read-only | Use `pop()` when you want removal |
| Reversing the order of subtraction/division | Results become incorrect | Pop right operand first, then left operand |
| Forgetting to pop matching brackets | Old opening brackets remain in the stack | Pop after confirming the correct match |
| Assuming a copied variable is a new list | `b = a` creates an alias | Use `a.copy()` for a separate outer list |
| Forgetting a recursion base case | Calls may never stop normally | Define a reachable base case |
| Assuming a stack sorts values | A stack only defines insertion/removal order | Sorting is a separate operation |
| Using stack search as if it were O(1) | An item may be anywhere | Searching is usually O(n) |

**Design principle:** A stack gives direct access to the top, not to every item by position. If you need frequent random access to arbitrary positions, another data structure may fit better.

## ✅ Quick Summary — Python Stack

| Concept | Key point |
|---|---|
| **Definition** | A data structure that follows LIFO — Last In, First Out |
| **Top** | The next item to be removed |
| **Push** | Add an item; for a list stack use `append(item)` |
| **Pop** | Remove and return the top; use `pop()` |
| **Peek** | Read the top without removing; use `stack[-1]` if non-empty |
| **Empty check** | `not stack` or `len(stack) == 0` |
| **Size** | `len(stack)` |
| **Underflow** | Trying to pop or peek when empty |
| **List implementation** | Keep top at the right end |
| **Class implementation** | Package operations into `push`, `pop`, `peek`, and other methods |
| **Deque implementation** | `collections.deque` also supports stack operations |
| **Typical complexity** | Push, pop, peek: O(1) amortized / O(1); search: O(n) |
| **Applications** | Undo, browser history, bracket matching, expression evaluation, recursion |
| **Monotonic stack** | Maintains an order to solve tasks such as Next Greater Element in O(n) |
| **Min Stack** | A second stack can track the minimum for O(1) minimum lookup |

### Final rule to remember

**Push adds to the top. Pop removes from the top. The last item pushed is the first item popped.**

When using a Python list as a stack in this notebook, the top is the **rightmost element**.
