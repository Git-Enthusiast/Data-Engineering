- what happened

## 2. Data → Information → Insight → Decision → Action

A useful mental model:

```text
DATA
 ↓
INFORMATION
 ↓
INSIGHT
 ↓
DECISION
 ↓
ACTION
```

### Example

**Data:** Product A sold 2,500 units.

**Information:** Product A sold more units than any other product.

**Insight:** Product A contributes a large share of revenue and demand is increasing.

**Decision:** Prioritise stock and promotional planning.

**Action:** Increase inventory and allocate marketing budget.

## 3. Finding vs Insight

### Finding
A directly observed result.

> “Revenue declined 12%.”

### Insight
An interpreted result with business meaning.

> “Revenue declined 12%, mainly because two high-value categories fell in the South region.”

### Recommendation
A proposed action supported by the insight.

> “Management should investigate the two categories and review the South-region promotion strategy.”

Students should not jump directly from data to recommendation without showing evidence.

## 4. The Five-Part Story

Use:

```text
1. Finding
2. Evidence
3. Explanation / Interpretation
4. Business Impact
5. Recommendation
```

This becomes the standard structure for student explanations.

## 5. Example Dataset

````python
sales_story = pd.DataFrame({
    "region": ["North", "South", "East", "West"],
    "revenue_last_year": [4200000, 5100000, 3900000, 4700000],
    "revenue_this_year": [4550000, 4380000, 4100000, 4920000]
})

sales_story["change"] = sales_story["revenue_this_year"] - sales_story["revenue_last_year"]
sales_story["change_pct"] = sales_story["change"] / sales_story["revenue_last_year"] * 100

sales_story
````

## 6. Start With the Question

Bad analytical communication often starts with:

> “Here is my chart.”

Strong communication starts with:

> “Which region has the greatest revenue decline, and what does that mean for management?”

The question determines:

- what data is relevant
- what chart is useful
- what comparison is meaningful
- what recommendation is justified.

## 7. Choosing the Right Visual

### Comparison
Use bar charts.

### Trend over time
Use line charts.

### Distribution
Use histogram or box plot.

### Relationship
Use scatter plot.

### Composition
Use carefully chosen stacked or grouped charts when appropriate.

The important skill is not remembering chart syntax. It is **matching the visual to the question**.

````python
plt.figure(figsize=(9, 5))
plt.bar(sales_story["region"], sales_story["change_pct"])
plt.axhline(0, linestyle="--")
plt.ylabel("Change (%)")
plt.title("Revenue Change by Region")
plt.show()
````

## 8. Annotating the Important Point

A chart should guide attention toward the decision-relevant result.

````python
plt.figure(figsize=(10, 5))
bars = plt.bar(sales_story["region"], sales_story["change_pct"])

for bar, value in zip(bars, sales_story["change_pct"]):
    plt.text(
        bar.get_x() + bar.get_width()/2,
        value + (0.5 if value >= 0 else -1.5),
        f"{value:.1f}%",
        ha="center"
    )

plt.axhline(0, linestyle="--")
plt.ylabel("Change (%)")
plt.title("Revenue Change by Region")
plt.show()
````

## 9. A Bad Chart vs Better Chart

A bad chart may:

- contain unnecessary decoration
- use unclear labels
- hide the relevant comparison
- use inappropriate scales
- include too many categories
- make the audience work to understand the result

A better chart:

- has a clear title
- has clear axes
- highlights the relevant comparison
- uses an appropriate chart type
- provides enough context to interpret the number.

## 10. Avoid Misleading Visualisation

Students should recognise:

- truncated axes
- inconsistent scales
- unclear units
- percentages presented without context
- comparing totals when rates are required
- cherry-picking a small period
- visual exaggeration

A visual can be technically correct and still communicate poorly.

````python
comparison = pd.DataFrame({
    "month": ["Jan", "Feb", "Mar", "Apr", "May", "Jun"],
    "sales": [100, 101, 99, 100, 102, 101]
})

fig, ax = plt.subplots(figsize=(10, 5))
ax.plot(comparison["month"], comparison["sales"], marker="o")
ax.set_title("Monthly Sales")
ax.set_ylabel("Sales Index")
plt.show()
````

## 11. Audience Matters

### Analyst
Wants methodology, assumptions, uncertainty and technical detail.

### Business Manager
Wants KPI movement, causes, risks and actions.

### Senior Leadership
Wants the key message, business impact, decision and expected outcome.

The same analysis may need different presentations for each audience.

## 12. Turning Numbers Into Sentences

Students can use this pattern:

> **[Metric] changed by [amount] over [period], mainly because [reason/evidence]. This matters because [business impact]. We recommend [action].**

Example:

> Revenue in the South region fell by 14% year-on-year, driven mainly by lower performance in the Electronics category. This places the region at risk of missing its quarterly target. Management should review category pricing and promotional activity.

## 13. Example — Customer Segment Communication

````python
segments = pd.DataFrame({
    "segment": ["High Value", "Medium Value", "Low Value"],
    "customers": [120, 340, 540],
    "revenue": [1800000, 1500000, 600000]
})

segments["revenue_per_customer"] = segments["revenue"] / segments["customers"]
segments
````

````python
plt.figure(figsize=(9, 5))
plt.bar(segments["segment"], segments["revenue_per_customer"])
plt.ylabel("Revenue per Customer")
plt.title("Revenue per Customer by Segment")
plt.show()
````

### Interpretation exercise

Students should explain:

- Which segment generates the greatest revenue per customer?
- Why might that matter?
- Does the business need more customers, higher spend, or both?
- What action could be justified from this chart?

## 15. Executive Summary Format

A strong one-paragraph executive summary can follow:

1. Objective
2. Main finding
3. Evidence
4. Business implication
5. Recommendation

Example structure:

> The analysis examined regional sales performance. Revenue declined in one region, with the decline concentrated in two product categories. This creates a risk to the quarterly target. Management should review pricing and promotions in those categories and monitor the region weekly.

## 16. Common Communication Errors

- Reporting numbers without meaning.
- Showing a chart without explaining it.
- Giving multiple recommendations without prioritisation.
- Using technical terms for a non-technical audience.
- Claiming causation from correlation.
- Ignoring limitations.
- Providing recommendations that do not follow from evidence.
