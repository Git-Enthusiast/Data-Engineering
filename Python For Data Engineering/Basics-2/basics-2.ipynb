## 1️⃣ What is a Function? — Definition, Syntax & Invocation

A **function** is a self-contained, reusable block of code that performs a specific, well-defined task. Functions are fundamental to writing clean, modular, and maintainable software.

### 🌟 Why Use Functions?
* **Reusability**: Write logic once, use it across multiple parts of your program.
* **Modularity**: Break large, complex programs into smaller, manageable pieces.
* **Maintainability & Readability**: Easier to test, debug, and understand.
* **DRY Principle**: Avoids code redundancy (*Don't Repeat Yourself*).

---

### 🔍 Anatomy of a Python Function
```python
def function_name(parameter1, parameter2):
    """Docstring: Optional description of what the function does."""
    # Function body (indented block)
    result = parameter1 + parameter2
    return result
```

| Component | Keyword / Symbol | Purpose |
|---|---|---|
| **Header Definition** | `def` | Keyword used to declare a function |
| **Function Name** | Identifier (e.g. `greet`) | Name used to call/invoke the function |
| **Parameters** | `(param1, ...)` | Variables that receive input values |
| **Colon** | `:` | Demarcates the start of the indented function body |
| **Docstring** | `""" ... """` | Triple-quoted string describing function purpose & parameters |
| **Body** | Indented lines (4 spaces) | Statements executed when the function is invoked |
| **Return** | `return` *(optional)* | Sends computed result back to the caller |

> 💡 **Key Concept**: Defining a function does **not** run its code. The code only executes when you explicitly **call (invoke)** the function with parentheses `()`.

````python
# ============================================
# 1️⃣ What is a Function? — Definition & Calling
# ============================================

# Defining a function using the 'def' keyword
def greet(name):
    """
    This function takes a name as input
    and prints a greeting message.
    """
    print(f"Hello, {name}! Welcome to Python functions.")

# Calling (invoking) the function with an argument
greet("Rajan")
greet("Data Engineer")

# Inspecting function documentation (docstrings)
print("\nDocstring of greet():")
print(greet.__doc__)
````

## 2️⃣ Parameters vs Arguments

While often used interchangeably in casual conversation, **parameters** and **arguments** have distinct technical meanings:

| Term | Also Known As | Definition | Example Location |
|---|---|---|---|
| **Parameter** | Formal Parameter | Variable defined in the function signature | `def multiply(x, y):` → `x, y` are parameters |
| **Argument** | Actual Argument | Real value passed to the function when calling it | `multiply(4, 5)` → `4, 5` are arguments |

---

### 📌 Argument Types in Python

1. **Positional Arguments**: Arguments matched to parameters based on their position/order.
2. **Keyword Arguments**: Arguments passed by explicitly specifying the parameter name (`param=value`), allowing any order.
3. **Default Parameters**: Parameters assigned a default fallback value in the function header if omitted by caller.

> ⚠️ **Rule for Default Parameters**: Parameters with default values **MUST follow** parameters without default values.
> * ✅ `def power(base, exponent=2):` (Valid)
> * ❌ `def power(base=2, exponent):` (SyntaxError: non-default argument follows default argument)

````python
# ============================================
# 2️⃣ Parameters vs Arguments & Default Values
# ============================================

# 1. Formal Parameters vs Actual Arguments (Positional)
def multiply(x, y):  # x, y are FORMAL PARAMETERS
    """
    This function takes two numbers as input
    and returns their product.
    """
    return x * y

# 4, 5 are ACTUAL ARGUMENTS passed positionally
result = multiply(4, 5)
print(f"The product is: {result}")


# 2. Default Parameter Values
def power(base, exponent=2):
    """
    This function raises the base to the power of exponent.
    If exponent is not provided, it defaults to 2 (square).
    """
    return base ** exponent

# Calling without default argument (uses exponent=2)
squared_value = power(3)
# Calling with custom argument (overrides default)
cubed_value = power(3, 3)

print(f"3 squared is: {squared_value}")
print(f"3 cubed is: {cubed_value}")


# 3. Keyword Arguments (Explicit parameter naming)
kw_result = multiply(y=10, x=5)  # Order doesn't matter when naming keywords!
print(f"Keyword argument product (x=5, y=10): {kw_result}")
````

## 3️⃣ Return Statement & Function Types

Functions can either send data back to the caller using `return` or simply perform actions (such as printing or writing to a database).

### 🔄 Functions with Return vs Without Return
* **With `return`**: Passes computed data back. The returned value can be stored in a variable, used in expressions, or passed to other functions.
* **Without `return`** (or empty `return`): Python automatically returns `None` (an instance of `NoneType`).

---

### ⚙️ Function Reference vs Function Execution
* `func`: A **reference** to the function object itself (`<class 'function'>`).
* `func()`: An **invocation** that runs the function and evaluates to its return value.

---

### 🏷️ Type Annotations (Type Hinting — PEP 484)
Type hints document expected parameter and return types for IDE autocompletion and static analysis (e.g. `mypy`). They do not enforce types at runtime, but are accessible via `__annotations__`.

```python
def sample_function_annotated(x: int) -> str:
    return str(x)
```

````python
# ============================================
# 3️⃣ Return Types, Function References & Annotations
# ============================================

# 1. Function with explicit return
def sample_function():
    return "This is a sample function."

# 2. Function without return (void / print-only)
def print_function():
    print("This is a print function. It does not return any value.")

# Checking return types
print("Return value of sample_function():", sample_function())
print("Type of sample_function() return:", type(sample_function()))

print("\nCalling print_function():")
val = print_function()
print("Returned value of print_function():", val)
print("Type of print_function() return:", type(val))


# 3. Function Reference vs Function Execution
def example():
    return 10

print("\n--- Function Reference vs Execution ---")
print("type(example)   [without ()]:", type(example))     # <class 'function'>
print("type(example()) [with ()]   :", type(example()))   # <class 'int'>


# 4. Type Annotations & __annotations__
def calculate_area(width: float, height: float) -> float:
    """Calculates rectangle area with type hints."""
    return width * height

def no_return_function() -> None:
    print("No return value")

print("\n--- Type Annotations Inspection ---")
print("calculate_area annotations:", calculate_area.__annotations__)
print("no_return_function annotations:", no_return_function.__annotations__)
````

## 4️⃣ Variable-Length Arguments (`*args` & `**kwargs`)

When writing generic functions, you may not know in advance how many arguments a user will pass. Python provides `*args` and `**kwargs` to accept an arbitrary number of arguments.

| Syntax | Name | Data Structure | Used For |
|---|---|---|---|
| `*args` | Non-Keyword Variable Arguments | `tuple` | Arbitrary number of positional arguments |
| `**kwargs` | Keyword Variable Arguments | `dict` | Arbitrary number of keyword/named arguments |

---

### 📋 Argument Order in Function Signatures
When combining different parameter types in a single function, you **must** follow this strict order:
1. Standard Positional Parameters (`a, b`)
2. `*args` (Positional variable arguments)
3. Keyword-Only / Default Parameters (`discount=0.1`)
4. `**kwargs` (Keyword variable arguments)

> 💡 The names `args` and `kwargs` are conventions; the unpacking operators `*` and `**` are what enable the variable-length behavior.

````python
# ============================================
# 4️⃣ Variable-Length Arguments (*args & **kwargs)
# ============================================

# 1. Basic *args and **kwargs inspection
def display_info(*args, **kwargs):
    """
    This function accepts any number of positional and keyword arguments
    and displays them.
    """
    print("Positional arguments (*args as tuple):", args, "Type:", type(args))
    print("Keyword arguments (**kwargs as dict)   :", kwargs, "Type:", type(kwargs))

display_info(1, 2, 3, name="Rajan", age=25, role="Data Engineer")


# 2. Practical Data Engineering Example: Dynamic Pipeline Record Logger
print("\n--- Practical Example: Dynamic Function Processing ---")

def process_batch(batch_id, *records, status="SUCCESS", **metadata):
    """
    Processes a batch of records with optional metadata.
    """
    print(f"Batch ID : {batch_id}")
    print(f"Status   : {status}")
    print(f"Records Count: {len(records)}")
    print(f"Record Items : {records}")
    print(f"Metadata : {metadata}")

process_batch(
    101,
    {"id": 1, "sensor": "temp", "val": 24.5},
    {"id": 2, "sensor": "pressure", "val": 101.3},
    status="COMPLETED",
    source="GCS",
    environment="production"
)
````

## 5️⃣ Nested Functions (Inner Functions) & Execution Lifecycle

In Python, functions are **first-class objects**. This means you can:
* Define a function **inside** another function (Nested / Inner functions).
* Return a function from another function.
* Pass functions as arguments to other functions.

---

### 🛡️ Why Use Nested Functions?
1. **Encapsulation & Helper Logic**: Hide helper functions that are only relevant inside the parent function.
2. **Access to Enclosing Scope**: Inner functions can read variables from the outer function's scope.
3. **Closures & Decorators**: Foundation for Python decorators and factory patterns.

---

### ⏳ Python Execution Lifecycle (Deferred Execution)
> 📌 **Key Rule**: When Python reads a `def` statement, it compiles the function and binds its name to a function object. The body of the function is **not executed** until that function is explicitly called!

````python
# ============================================
# 5️⃣ Nested Functions & Execution Lifecycle
# ============================================

# 1. Defining and executing an inner function
def outer_function(msg):
    """
    This is the outer function that takes a message as input.
    """
    def inner_function():
        """
        This is the inner function that accesses and prints 
        the message from the enclosing outer function scope.
        """
        print(f"[Inner Function Executing]: {msg}")
    
    print("[Outer Function Executing]: Starting inner function call...")
    inner_function()
    print("[Outer Function Executing]: Finished inner function call.")

outer_function("Hello from the outer function scope!")


# 2. Data Cleaning Helper nested inside transformation
print("\n--- Practical Nested Helper Example ---")

def transform_user_record(raw_record):
    """Encapsulates cleaning steps inside inner helper functions."""
    def clean_string(s):
        return s.strip().title() if isinstance(s, str) else s
    
    def validate_age(age):
        return max(0, int(age)) if str(age).isdigit() else None

    return {
        "name": clean_string(raw_record.get("name")),
        "city": clean_string(raw_record.get("city")),
        "age": validate_age(raw_record.get("age"))
    }

raw_data = {"name": "   rajan kumar  ", "city": "bangalore  ", "age": "23"}
cleaned_data = transform_user_record(raw_data)
print("Raw Record    :", raw_data)
print("Cleaned Record:", cleaned_data)
````

## 6️⃣ Built-in Functions & The `builtins` Module

Python includes a rich collection of **built-in functions, exceptions, and constants** that are always directly available in every script without needing any `import` statement.

### 📦 The `builtins` Module
Under the hood, all built-in objects reside in a special module called `builtins`.
* `dir(builtins)`: Lists all built-in names (functions like `len`, `print`, `type`; exceptions like `ValueError`, `TypeError`; constants like `True`, `False`, `None`).
* `help(func)`: Displays the interactive documentation, parameters, and return type for any function.

---

### 📊 Common Built-in Functions Cheat Sheet

| Category | Functions | Purpose |
|---|---|---|
| **Inspection & Meta** | `type()`, `id()`, `dir()`, `help()`, `isinstance()`, `callable()` | Inspect object characteristics and help |
| **Numeric & Math** | `abs()`, `round()`, `pow()`, `bin()`, `hex()`, `oct()`, `sum()`, `min()`, `max()` | Mathematical calculations and base conversions |
| **Collections & Sequences** | `len()`, `sorted()`, `reversed()`, `enumerate()`, `zip()`, `all()`, `any()`, `range()` | Working with lists, tuples, and iterables |
| **Type Conversion** | `int()`, `float()`, `str()`, `bool()`, `list()`, `dict()`, `tuple()`, `set()` | Casting between data types |
| **I/O & Formatting** | `print()`, `input()`, `format()`, `open()` | Console input/output, formatting, file handling |

> 💡 Built-in functions are implemented in optimized C (in CPython), making them significantly faster than custom pure-Python loops.

````python
# ============================================
# 6️⃣ Built-in Functions & The builtins Module
# ============================================

import builtins

# 1. Listing all built-in objects using dir(builtins)
all_builtins = dir(builtins)
print(f"Total built-in objects available: {len(all_builtins)}")
print("\nFirst 15 built-in items:")
print(all_builtins[:15])

# Categorizing built-ins
exceptions = [item for item in all_builtins if item.endswith("Error") or "Exception" in item]
functions_and_others = [item for item in all_builtins if item[0].islower() and not item.startswith("__")]

print(f"\nCount of Built-in Exceptions (e.g. ValueError): {len(exceptions)}")
print(f"Count of Built-in Functions/Types (lowercase): {len(functions_and_others)}")
print("\nSample Built-in Functions:")
print(functions_and_others[:20])


# 2. Interactive documentation using help()
print("\n--- Documentation of len() via help() ---")
# Using __doc__ to display clean inline output
print(len.__doc__)
````

## 7️⃣ Memory & Object Identity with `id()`

In Python, every value is an object stored at a specific location in memory. The built-in `id(obj)` function returns an integer representing the object's **unique memory address**.

---

### 🧠 Integer Caching (Small Integer Optimization)
For performance, CPython pre-allocates and caches small integers in the range **`-5` to `256`**. When multiple variables are assigned a value in this range, they point to the **exact same memory location**!

---

### 🔄 Mutable vs Immutable Object Memory Behavior

| Object Type | Examples | Behavior on Modification | `id()` Result |
|---|---|---|---|
| **Immutable** | `int`, `float`, `str`, `tuple`, `bool` | Cannot be modified in-place; creates a brand new object | `id()` **changes** |
| **Mutable** | `list`, `dict`, `set` | Modified in-place without creating a new container | `id()` **remains same** |

---

### ⚖️ `id()` & `is` (Identity) vs `==` (Value Equality)
* `a == b`: Checks whether `a` and `b` have the **same value/content**.
* `a is b` (or `id(a) == id(b)`): Checks whether `a` and `b` point to the **exact same object in memory**.

````python
# ============================================
# 7️⃣ Object Identity & Memory Analysis with id()
# ============================================

# 1. Small Integer Caching (-5 to 256)
a = 10
b = 10
print(f"id(a): {id(a)}")
print(f"id(b): {id(b)}")
print(f"a is b (Same Memory Address?): {a is b}")  # True


# 2. Distinct Mutable Objects (Lists)
x = [1, 2, 3]
y = [1, 2, 3]
print(f"\nid(x): {id(x)}")
print(f"id(y): {id(y)}")
print(f"x == y (Same values?): {x == y}")         # True (equal content)
print(f"x is y (Same object?): {x is y}")         # False (different objects)
print(f"id(x) == id(y)       : {id(x) == id(y)}")  # False


# 3. Mutable Object Modification (In-Place)
lst = [1, 2]
print(f"\nList before append: {lst}, id: {id(lst)}")
lst.append(3)
print(f"List after append : {lst}, id: {id(lst)}")  # Same ID!


# 4. Immutable Object Modification (Creates New Object)
num = 5
print(f"\nInteger before += 1: {num}, id: {id(num)}")
num += 1
print(f"Integer after += 1 : {num}, id: {id(num)}")  # New ID!


# 5. id() with Functions (Function Object vs Return Value)
def another_function():
    return "Hello"

print(f"\nid(another_function)   [Function Object] : {id(another_function)}")
print(f"id(another_function()) [Returned Value 'Hello']: {id(another_function())}")
````

## 8️⃣ Mathematical & Numeric Built-ins (`bin`, `pow`, `round`, `abs`)

Python provides powerful mathematical built-in functions that eliminate the need to import external libraries for common arithmetic, base conversions, and rounding operations.

---

### 1. `bin(x)` — Binary Representation
Converts an integer to a binary string prefixed with `0b`.
* Works with positive and negative integers (`-10` -> `'-0b1010'`).
* Convert binary strings back to integers using `int(binary_str, 2)`.

---

### 2. `pow(base, exp[, mod])` — Exponentiation & Modular Arithmetic
* **Two arguments**: `pow(base, exp)` computes `base ** exp`.
* **Three arguments**: `pow(base, exp, mod)` computes `(base ** exp) % mod` efficiently (crucial in cryptography, hash partitioning, and data engineering algorithms).

---

### 3. `round(number[, ndigits])` — Rounding
* `round(number)`: Rounds to nearest integer.
* `round(number, ndigits)`: Rounds to `ndigits` decimal places.
* ⚖️ **Banker's Rounding (Round Half to Even)**: When a number is exactly halfway (e.g. `2.5`, `3.5`), Python rounds to the nearest **even** number (`round(2.5) == 2`, `round(3.5) == 4`). This minimizes cumulative statistical bias.

---

### 4. `abs(x)` — Absolute Value
Returns the non-negative magnitude (distance from zero).
* **Integers & Floats**: Strips negative sign (`abs(-10)` -> `10`, `abs(-5.67)` -> `5.67`).
* **Complex Numbers**: Computes modulus $|a + bi| = \sqrt{a^2 + b^2}$ (`abs(3 - 4j)` -> `5.0`).

````python
# ============================================
# 8️⃣ Mathematical Built-ins (bin, pow, round, abs)
# ============================================

# 1. bin() — Binary Conversion
num = 10
neg_num = -10
binary_str = bin(num)
neg_binary_str = bin(neg_num)

print("--- bin() Demonstrations ---")
print(f"Binary of {num}        : {binary_str}")       # 0b1010
print(f"Binary of {neg_num}       : {neg_binary_str}")   # -0b1010
print(f"Binary of expression (5+3): {bin(5 + 3)}")       # 0b1000

# Converting binary string back to integer
reconstructed_int = int(binary_str, 2)
print(f"int('{binary_str}', 2) back to int: {reconstructed_int}")


# 2. pow() — Power and Modular Exponentiation
print("\n--- pow() Demonstrations ---")
base, exponent, modulus = 2, 3, 3
print(f"pow({base}, {exponent})              : {pow(base, exponent)}")           # 8
print(f"pow({base}, {exponent}, {modulus}) (modulus) : {pow(base, exponent, modulus)}")  # (2^3) % 3 = 2
# Efficient large power modulo (data partition hashing)
print(f"pow(7, 256, 13) (large fast mod): {pow(7, 256, 13)}")


# 3. round() — Rounding & Banker's Rounding
print("\n--- round() Demonstrations ---")
pi_approx = 5.6789
print(f"round({pi_approx}, 2) : {round(pi_approx, 2)}")
print(f"round({pi_approx})    : {round(pi_approx)}")
print(f"round(-2.3456, 2)  : {round(-2.3456, 2)}")

# Banker's Rounding (rounds half to nearest EVEN number)
print(f"round(2.5) -> {round(2.5)} (rounds down to even 2)")
print(f"round(3.5) -> {round(3.5)} (rounds up to even 4)")


# 4. abs() — Absolute Values
print("\n--- abs() Demonstrations ---")
print(f"abs(-10)             : {abs(-10)}")
print(f"abs(-5.67)           : {abs(-5.67)}")
# Complex number modulus: sqrt(3^2 + (-4)^2) = sqrt(9 + 16) = 5.0
complex_val = 3 - 4j
print(f"abs({complex_val})       : {abs(complex_val)}")
````

## 9️⃣ Value & String Formatting with `format()`

The built-in `format(value, format_spec)` function formats values according to a format specification string. It is widely used in data pipelines for data normalization, precision control, and generating aligned tabular reports.

---

### 📐 Format Specification Syntax Reference

| Format Specifier | Description | Example Input | Formatted Output |
|---|---|---|---|
| `".3f"` | Fixed-point float with 3 decimals | `format(8/9, ".3f")` | `'0.889'` |
| `",.2f"` | Float with thousands comma separator | `format(1250000.5, ",.2f")` | `'1,250,000.50'` |
| `".1%"` | Percentage with 1 decimal | `format(0.856, ".1%")` | `'85.6%'` |
| `"<30"` | Left-aligned in 30-character field | `format("Hello", "<30")` | `'Hello                         '` |
| `">30"` | Right-aligned in 30-character field | `format("Hello", ">30")` | `'                         Hello'` |
| `"^30"` | Center-aligned in 30-character field | `format("Hello", "^30")` | `'            Hello            '` |
| `"*^30"` | Center-aligned with `*` fill characters | `format("Hello", "*^30")` | `'************Hello*************'` |

````python
# ============================================
# 9️⃣ Value & String Formatting with format()
# ============================================

# 1. Number Precision and Float Formatting
print("--- Number Formatting ---")
print("8/9 formatted to 3 decimal places :", format(8/9, ".3f"))
print("Large number with comma separators:", format(1234567890.456, ",.2f"))
print("Ratio formatted as percentage     :", format(0.94258, ".2%"))
print("Integer formatted as 8-bit binary :", format(42, "08b"))


# 2. String Alignment and Width Formatting
print("\n--- String Alignment (Width 30) ---")
print(format("Hello", "<30") + "| [Left Aligned]")
print(format("Hello", ">30") + "| [Right Aligned]")
print(format("Hello", "^30") + "| [Center Aligned]")
print(format("Hello", "*^30") + "| [Center Aligned with Fill '*']")


# 3. Practical Data Engineering Report: Pipeline Execution Table
print("\n" + "=" * 60)
print(format("DATA PIPELINE SUMMARY REPORT", "^60"))
print("=" * 60)

headers = ["Pipeline Stage", "Records", "Duration", "Success Rate"]
print(f"{format(headers[0], '<22')}{format(headers[1], '>12')}{format(headers[2], '>12')}{format(headers[3], '>14')}")
print("-" * 60)

stages = [
    ("GCS Ingestion", 1542000, "12.4s", 0.9998),
    ("Spark Transformation", 1541800, "45.8s", 0.9995),
    ("BigQuery Loading", 1541800, "8.2s", 1.0000)
]

for stage, recs, duration, rate in stages:
    col_stage = format(stage, "<22")
    col_recs = format(format(recs, ","), ">12")
    col_dur = format(duration, ">12")
    col_rate = format(format(rate, ".2%"), ">14")
    print(f"{col_stage}{col_recs}{col_dur}{col_rate}")

print("=" * 60)
````

* **UDFs in PySpark / Pandas / Dataflow**: Custom transformation functions passed to `.apply()`, `.map()`, or UDF registries require clear parameter signatures, pure return values, and type annotations.
* **Memory & In-Place Performance**: Knowing mutable vs immutable behavior (`id()`) prevents accidental state mutation when processing large batches of records in streaming or distributed environments.
* **Standard Built-ins Over Custom Loops**: Built-ins like `sum()`, `max()`, `abs()`, and `pow()` execute in optimized C, minimizing ETL CPU overhead.
* **Pipeline Monitoring & Logging**: The `format()` function enables structured console tables and alignment across automated pipeline logs.

### 🏆 Golden Rules for Python Functions

> 1. **Single Responsibility Principle (SRP)**: Each function should do **one thing and do it well**.
> 2. **Always Use Docstrings**: Document the purpose, parameters, and return types using docstrings.
> 3. **Leverage Type Hints**: Use Python 3.9+ type annotations (`param: int -> str`) to improve code readability and eliminate bugs early.
> 4. **Avoid Default Mutable Arguments**: Never use mutable objects (`list` or `dict`) as default parameters (`def func(items=[]):` ❌). Instead, use `def func(items=None): items = items or []` ✅.
> 5. **Prefer Pure Functions**: When possible, avoid modifying global state or mutable inputs directly inside functions.
