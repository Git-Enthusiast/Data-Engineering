## 1️⃣ Python Data Types & Dynamic Typing System

Python is a **dynamically typed language**, meaning variable types are determined automatically at runtime based on the assigned value. Variables in Python do not have fixed types; instead, variables are simply **names (labels/references)** that point to objects in memory.

### 📊 Python Built-in Data Type Taxonomy

| Category | Data Type | Mutability | Description / Examples |
|---|---|---|---|
| **Numeric** | `int` | Immutable | Whole numbers of arbitrary precision: `10`, `-500`, `10**100` |
| | `float` | Immutable | Double-precision decimal numbers & exponential notation: `3.14`, `1.5e3` |
| | `complex` | Immutable | Real + Imaginary numbers with `j`: `4 + 5j` |
| **Sequence** | `str` | Immutable | Ordered sequence of Unicode characters: `'Hello'`, `"Data"`, `'''Multi'''` |
| | `list` | Mutable | Ordered, heterogenous collection: `[1, "AI", 3.14, [10, 20]]` |
| | `tuple` | Immutable | Ordered, heterogenous immutable collection: `(1, "ML", 3.14)` |
| | `range` | Immutable | Immutable sequence of arithmetic progressions: `range(0, 10, 2)` |
| **Mapping** | `dict` | Mutable | Key-Value pairs with unique, hashable keys: `{"name": "Rajan", "age": 22}` |
| **Set** | `set` | Mutable | Unordered collection of unique, hashable items: `{1, 2, 3}` |
| | `frozenset` | Immutable | Immutable version of a set: `frozenset([1, 2, 3])` |
| **Boolean** | `bool` | Immutable | Logical truth values (`True` / `False`), subclass of `int` (`True == 1`) |
| **Binary** | `bytes` | Immutable | Raw sequence of bytes: `b"ABC"` |
| | `bytearray` | Mutable | Mutable array of bytes: `bytearray(b"ABC")` |
| **None** | `NoneType` | Immutable | Represents the absence of a value (`None`) |

> 💡 **Note**: Python does **not** have a separate `char` data type. A single character is simply a string of length 1 (`len(c) == 1`).

````python
# ============================================================
# 4️⃣ Type Checking: isinstance() vs type()
# ============================================================

# Best Practice: Prefer isinstance() over type() because:
# 1. isinstance() supports inheritance (e.g. bool is a subclass of int)
# 2. isinstance() accepts a tuple of multiple types for polymorphic checks

val = True

print("--- isinstance() vs type() ---")
print(f"type(val) == bool         : {type(val) == bool}")        # True
print(f"type(val) == int          : {type(val) == int}")         # False (strict exact match)
print(f"isinstance(val, int)      : {isinstance(val, int)}")     # True (accounts for inheritance!)
print(f"isinstance(val, (int, str)): {isinstance(val, (int, str))}") # True (checks union of types)

# Practical Data Pipeline Type Router:
print("\n--- Practical Pipeline Record Type Router ---")
raw_records = [100, 3.1415, "Kafka_Topic_01", False, None, [10, 20, 30], {"batch_id": 101}]

for idx, item in enumerate(raw_records):
    if isinstance(item, bool):
        # Must check bool before int if differentiating, since bool is subclass of int!
        print(f"Record [{idx}] -> Boolean : {item}")
    elif isinstance(item, int):
        print(f"Record [{idx}] -> Integer : {item}")
    elif isinstance(item, float):
        print(f"Record [{idx}] -> Float   : {item:.2f}")
    elif isinstance(item, str):
        print(f"Record [{idx}] -> String  : '{item}'")
    elif isinstance(item, (list, tuple, dict)):
        print(f"Record [{idx}] -> Complex Container ({type(item).__name__}) : {item}")
    elif item is None:
        print(f"Record [{idx}] -> Null/NoneType : {item}")
````

## 2️⃣ Arithmetic Operators & Implicit Type Coercion Rules

Arithmetic operators perform standard mathematical operations on numeric operands.

| Operator | Name | Syntax | Result Type Rule | Example |
|---|---|---|---|---|
| `+` | Addition | `a + b` | `int + int -> int`<br>`int + float -> float` | `5 + 3 -> 8`<br>`5 + 3.0 -> 8.0` |
| `-` | Subtraction | `a - b` | `int - int -> int`<br>`int - float -> float` | `5 - 3 -> 2`<br>`5 - 3.0 -> 2.0` |
| `*` | Multiplication | `a * b` | `int * int -> int`<br>`int * float -> float`<br>`str * int -> str` (repetition) | `5 * 3 -> 15`<br>`5 * 3.0 -> 15.0`<br>`"Hi" * 3 -> "HiHiHi"` |
| `/` | **True Division** | `a / b` | **ALWAYS** returns `float` in Python 3 | `36 / 4 -> 9.0`<br>`5 / 2 -> 2.5` |
| `//` | **Floor Division** | `a // b` | Returns largest integer <= quotient.<br>`int // int -> int`<br>`float` involved -> `float` | `9 // 2 -> 4`<br>`9.0 // 2 -> 4.0`<br>`-9 // 2 -> -5` |
| `%` | **Modulus (Remainder)** | `a % b` | Returns remainder after division.<br>`float` involved -> `float` | `9 % 2 -> 1`<br>`9 % 2.0 -> 1.0`<br>`100 % 25 -> 0` |
| `**` | **Exponentiation** | `a ** b` | `a` raised to the power `b`.<br>`float` power -> `float` | `2 ** 3 -> 8`<br>`9 ** 0.5 -> 3.0` |

### ⚠️ Critical Arithmetic Rules to Remember
1. **True Division (`/`) ALWAYS returns `float`**: Even if numbers divide evenly (`36 / 4` evaluates to `9.0`, not `9`).
2. **Floor Division (`//`) Rounds Down (towards -inf)**:
   - Positive numbers: `9 // 2 == 4` (truncates decimal).
   - Negative numbers: `-9 // 2 == -5` (rounds down to the next smaller integer).
   - Float operands: `9.0 // 2 == 4.0` (floor value formatted as a float).
3. **Float Contagion (Implicit Type Conversion)**: If any operand is a `float`, Python automatically promotes the other operands to `float` and returns a `float`.


````python
# ============================================================
# 2️⃣ Comprehensive Arithmetic Operators & Return Types
# ============================================================

print("--- 1. Addition & Subtraction ---")
print(f"5 + 3   = {5 + 3}   | Type: {type(5 + 3)}")
print(f"5 + 3.0 = {5 + 3.0} | Type: {type(5 + 3.0)} (Implicit float promotion)")
print(f"5 - 3   = {5 - 3}   | Type: {type(5 - 3)}")
print(f"5 - 3.0 = {5 - 3.0} | Type: {type(5 - 3.0)}")

print("\n--- 2. Multiplication & String Repetition ---")
print(f"5 * 3   = {5 * 3}   | Type: {type(5 * 3)}")
print(f"5 * 3.0 = {5 * 3.0} | Type: {type(5 * 3.0)}")
print(f"'Hi' * 3 = {'Hi' * 3} | Type: {type('Hi' * 3)} (Sequence repetition)")

print("\n--- 3. True Division (/) vs Floor Division (//) ---")
print(f"36 / 4    = {36 / 4}    | Type: {type(36 / 4)} (Always float!)")
print(f"5 / 2     = {5 / 2}     | Type: {type(5 / 2)}")
print(f"36 // 4   = {36 // 4}   | Type: {type(36 // 4)} (Integer floor division)")
print(f"5 // 2    = {5 // 2}    | Type: {type(5 // 2)}")
print(f"9.0 // 2  = {9.0 // 2}  | Type: {type(9.0 // 2)} (Float floor division)")
print(f"-9 // 2   = {-9 // 2}   | Type: {type(-9 // 2)} (Rounds down to -5!)")
print(f"-10 // 3  = {-10 // 3}  | Type: {type(-10 // 3)} (Rounds down to -4!)")

print("\n--- 4. Modulus (%) & Exponentiation (**) ---")
print(f"9 % 2     = {9 % 2}     | Type: {type(9 % 2)}")
print(f"9 % 2.0   = {9 % 2.0}   | Type: {type(9 % 2.0)}")
print(f"100 % 25  = {100 % 25}  | Type: {type(100 % 25)} (Divisible -> remainder 0)")
print(f"2 ** 3    = {2 ** 3}    | Type: {type(2 ** 3)}")
print(f"9 ** 0.5  = {9 ** 0.5}  | Type: {type(9 ** 0.5)} (Square root via fractional exponent)")
print(f"23.3 ** 2 = {23.3 ** 2:.2f} | Type: {type(23.3 ** 2)}")
````

## 3️⃣ Comparison (Relational) Operators & Truth Value Testing

Comparison operators compare the values of two operands and always return a boolean (`True` or `False`).

| Operator | Meaning | Example | Evaluates To |
|---|---|---|---|
| `>` | Greater than | `10 > 5` | `True` |
| `<` | Less than | `10 < 5` | `False` |
| `==` | Equal to (value equality) | `10 == 10.0` | `True` (values match despite different types) |
| `!=` | Not equal to | `10 != 5` | `True` |
| `>=` | Greater than or equal to | `10 >= 10` | `True` |
| `<=` | Less than or equal to | `10 <= 5` | `False` |

> 💡 **Comparison Chaining**: Python supports chained comparisons such as `1 < x < 10`, which is evaluated as `(1 < x) and (x < 10)` without evaluating `x` twice.


````python
# ============================================================
# 3️⃣ Comparison Operators Demonstration
# ============================================================

x = 10
y = 5

print("--- Comparison Checks ---")
print(f"x = {x}, y = {y}")
print(f"x > y  : {x > y}  | Type: {type(x > y)}")
print(f"x < y  : {x < y}")
print(f"x == y : {x == y}")
print(f"x != y : {x != y}")
print(f"x >= y : {x >= y}")
print(f"x <= y : {x <= y}")

print("\n--- Cross-Type Equality vs Chained Comparison ---")
# Value equality across int and float
print(f"5 == 5.0      : {5 == 5.0}")       # True (numeric values are identical)
print(f"5 == '5'      : {5 == '5'}")       # False (different types)

# Comparison Chaining
val = 7
print(f"3 < val < 9   : {3 < val < 9}")    # True: equivalent to (3 < 7) and (7 < 9)
print(f"10 < val < 20 : {10 < val < 20}")  # False
````

## 4️⃣ Assignment & Compound Operators — The Deep Dive (`+=` vs `= +`)

Assignment operators store values into variables. **Compound assignment operators** combine an arithmetic operation with assignment in a concise syntax.

### 📋 Standard Compound Operators
| Operator | Equivalent To | Description |
|---|---|---|
| `c += 5` | `c = c + 5` | Addition assignment |
| `c -= 3` | `c = c - 3` | Subtraction assignment |
| `c *= 2` | `c = c * 2` | Multiplication assignment |
| `c /= 4` | `c = c / 4` | True division assignment (becomes float) |
| `c //= 2` | `c = c // 2` | Floor division assignment |
| `c %= 3` | `c = c % 3` | Modulus assignment |
| `c **= 2` | `c = c ** 2` | Exponentiation assignment |

---

### 🔍 Deep Dive: Why `a += b` is NOT always equal to `a = a + b`

This is one of the most critical Python internals and a favorite interview question.

```
                  ┌─────────────────┐
                  │   a += b        │
                  └────────┬────────┘
                           │
             ┌─────────────┴─────────────┐
             ▼                           ▼
     [Mutable Object]            [Immutable Object]
    (e.g., list, dict)          (e.g., int, str, tuple)
             │                           │
    Calls a.__iadd__(b)          Calls a.__add__(b)
    Modifies in-place            Creates NEW object
    Memory ID stays SAME         Memory ID CHANGES
```

* **`a += b`** calls the in-place method **`__iadd__()`**.
* **`a = a + b`** calls the standard method **`__add__()`** and rebinds `a` to a brand-new object.

| Type Category | Data Type | `a = a + b` | `a += b` | Behavior Equivalent? |
|---|---|---|---|---|
| **Immutable** | `int`, `str`, `tuple` | Creates **new** object (`id` changes) | Creates **new** object (`id` changes) | ✅ **Yes** |
| **Mutable** | `list`, `set`, `dict` | Creates **new** object (`id` changes) | Modifies **in-place** (`id` remains same) | ❌ **NO (Different!)** |

> ⚠️ **The Shared Reference (Aliasing) Hazard**:
> When two variables point to the same mutable list (`a = [1, 2]; b = a`), executing `a += [3]` modifies the underlying list in-place, which **silently changes `b` as well**!
> Conversely, `a = a + [3]` creates a new list for `a`, leaving `b` referencing the original unaltered list.


````python
# ============================================================
# 4️⃣ Compound Assignment Operators in Action
# ============================================================

c = 10
print(f"Initial c: {c}")

c += 5   # c = 10 + 5 = 15
print(f"After c += 5  : {c}")

c -= 3   # c = 15 - 3 = 12
print(f"After c -= 3  : {c}")

c *= 2   # c = 12 * 2 = 24
print(f"After c *= 2  : {c}")

c /= 4   # c = 24 / 4 = 6.0 (converted to float)
print(f"After c /= 4  : {c}")

c //= 2  # c = 6.0 // 2 = 3.0
print(f"After c //= 2 : {c}")

c %= 2   # c = 3.0 % 2 = 1.0
print(f"After c %= 2  : {c}")

c **= 3  # c = 1.0 ** 3 = 1.0
print(f"After c **= 3 : {c}")
````

````python
# ============================================================
# 🔬 In-Depth Memory Proof: a += b vs a = a + b
# ============================================================

print("==================================================")
print("1. IMMUTABLE TYPE (INTEGER): Both produce new objects")
print("==================================================")

# Testing a += 5
x1 = 10
print(f"Initial x1: {x1} | id(x1): {id(x1)}")
x1 += 5
print(f"After x1 += 5 : {x1} | id(x1): {id(x1)} (ID Changed -> New Object)")

# Testing x2 = x2 + 5
x2 = 10
print(f"Initial x2: {x2} | id(x2): {id(x2)}")
x2 = x2 + 5
print(f"After x2 = x2 + 5: {x2} | id(x2): {id(x2)} (ID Changed -> New Object)")


print("\n==================================================")
print("2. MUTABLE TYPE (LIST): += is In-Place, + is New Object")
print("==================================================")

# Case A: list1 = list1 + [4, 5] (Calls __add__)
list1 = [1, 2, 3]
print(f"Initial list1 : {list1} | id(list1): {id(list1)}")
list1 = list1 + [4, 5]
print(f"After list1 = list1 + [4, 5]: {list1} | id(list1): {id(list1)} (ID CHANGED -> New list created)")

# Case B: list2 += [4, 5] (Calls __iadd__)
list2 = [1, 2, 3]
print(f"\nInitial list2 : {list2} | id(list2): {id(list2)}")
list2 += [4, 5]
print(f"After list2 += [4, 5]: {list2} | id(list2): {id(list2)} (ID UNCHANGED -> Mutated in-place!)")


print("\n==================================================")
print("3. THE ALIASING SIDE EFFECT (REAL-WORLD PRODUCTION BUG)")
print("==================================================")

# Shared reference with += (In-place mutation leaks to aliases)
a_alias = [1, 2, 3]
b_alias = a_alias  # b_alias points to same list object

print(f"Initial a_alias: {a_alias}, b_alias: {b_alias}")
a_alias += [4]     # Modifies the shared list object in-place!
print(f"After a_alias += [4]:")
print(f"a_alias: {a_alias}")
print(f"b_alias: {b_alias} <-- BUG WARNING: b_alias got modified too!")

# Shared reference with normal + (Safe re-binding)
a_safe = [1, 2, 3]
b_safe = a_safe

a_safe = a_safe + [4] # Creates a new list object for a_safe
print(f"\nAfter a_safe = a_safe + [4]:")
print(f"a_safe: {a_safe}")
print(f"b_safe: {b_safe} <-- b_safe remains unaffected!")
````

## 5️⃣ Logical Operators, Short-Circuit Evaluation & Non-Boolean Returns

Logical operators combine or modify conditional expressions.

| Operator | Syntax | Description |
|---|---|---|
| `and` | `x and y` | Returns `y` if `x` is truthy; otherwise returns `x` immediately (both must be True) |
| `or` | `x or y` | Returns `x` if `x` is truthy; otherwise returns `y` immediately (at least one True) |
| `not` | `not x` | Inverts the truth value: returns `True` if `x` is falsy, `False` if `x` is truthy |

---

### ⚡ Short-Circuit Evaluation Rules
Python uses **short-circuit evaluation**, meaning evaluation stops as soon as the final outcome is determined:
* `True and <expr>` -> Evaluates `<expr>` and returns it.
* `False and <expr>` -> Returns `False` (or the falsy value) **immediately without evaluating `<expr>`**.
* `True or <expr>` -> Returns `True` (or the truthy value) **immediately without evaluating `<expr>`**.
* `False or <expr>` -> Evaluates `<expr>` and returns it.

---

### 💡 Truthy vs Falsy Definitions in Python
Every object in Python has an inherent boolean truth value:

| Falsy Values (Evaluate to `False`) | Truthy Values (Evaluate to `True`) |
|---|---|
| `0`, `0.0`, `0j` (Numeric zeros) | Any non-zero number: `10`, `-5`, `3.14`, `100` |
| `None` (Null value) | Non-empty strings: `"Hello"`, `"0"`, `" "` |
| `False` (Boolean false) | Non-empty collections: `[1]`, `(0,)`, `{"k": "v"}`, `{0}` |
| `""`, `''`, `` (Empty string) | Functions, classes, active objects |
| `[]`, `()`, `{}` , `set()` (Empty collections) | Custom objects implementing non-zero `__bool__` / `__len__` |

---

### 🎯 The "Operand Value Return" Principle
Logical operators in Python **do NOT simply return `True` or `False`** when applied to arbitrary values. **They return the actual operand that determined the result!**

* **`x and y`**: If `x` is falsy -> returns `x`. If `x` is truthy -> returns `y`.
* **`x or y`**: If `x` is truthy -> returns `x`. If `x` is falsy -> returns `y`.
* **Result Type**: The return type dynamically matches whichever operand was selected!


````python
# ============================================================
# 5️⃣ Logical Operators, Truth Tables & Decision Logic
# ============================================================

# Simulating candidate eligibility with multiple logical conditions
knowledge_of_python = True
passout_year = 2023
years_experience = 2

# 1. AND Operator: All conditions must be True
is_eligible_and = (knowledge_of_python is True) and (passout_year >= 2022)
print("--- Logical AND ---")
print(f"Eligible (Python AND >= 2022): {is_eligible_and}")

# 2. OR Operator: At least one condition must be True
is_eligible_or = (years_experience >= 3) or (knowledge_of_python is True)
print("\n--- Logical OR ---")
print(f"Eligible (Exp >= 3 OR Python): {is_eligible_or}")

# 3. NOT Operator: Reverses boolean result
is_not_eligible = not (knowledge_of_python and passout_year >= 2022)
print("\n--- Logical NOT ---")
print(f"NOT Eligible: {is_not_eligible}")
````

````python
# ============================================================
# ⚡ Short-Circuiting & Non-Boolean Operand Returns
# ============================================================

print("--- 1. 'and' Returns the Determining Operand ---")
# If first is truthy, 'and' must check and return the second
print(f"100 and 49      -> Result: {100 and 49} | Type: {type(100 and 49)}")
print(f"19.3 and 23.3   -> Result: {19.3 and 23.3} | Type: {type(19.3 and 23.3)}")
print(f"'Rajan' and True -> Result: {'Rajan' and True} | Type: {type('Rajan' and True)}")
print(f"True and 'Rajan' -> Result: {True and 'Rajan'} | Type: {type(True and 'Rajan')}")

# If first is falsy, 'and' short-circuits and returns the first falsy operand
print(f"0 and 50        -> Result: {0 and 50} (Short-circuited at 0)")
print(f"None and 'Data' -> Result: {None and 'Data'} (Short-circuited at None)")
print(f"[] and [1, 2]   -> Result: {[] and [1, 2]} (Short-circuited at [])")


print("\n--- 2. 'or' Returns the First Truthy Operand ---")
# If first is truthy, 'or' short-circuits and returns it immediately
print(f"100 or 49       -> Result: {100 or 49}")
print(f"'Hello' or ''   -> Result: '{'Hello' or ''}'")

# If first is falsy, 'or' returns the next operand
print(f"0 or 42         -> Result: {0 or 42}")
print(f"0.0 or 23.3     -> Result: {0.0 or 23.3} | Type: {type(0.0 or 23.3)}")
print(f"'' or 'Hello'   -> Result: '{'' or 'Hello'}'")
print(f"[] or [10, 20]  -> Result: {[] or [10, 20]}")


print("\n--- 3. 'not' Always Returns a Strict Boolean ---")
print(f"not 100         -> Result: {not 100}")
print(f"not 0           -> Result: {not 0}")
print(f"not ''          -> Result: {not ''}")
print(f"not 'Data'      -> Result: {not 'Data'}")
print(f"not None        -> Result: {not None}")
````

## 6️⃣ Bitwise Operators, Binary Representation & Two's Complement

Bitwise operators manipulate data directly at the **binary bit level (0s and 1s)**. They are strictly defined for **integers** only.

### 🔢 Types of Bitwise Operators

| Operator | Name | Rule / Truth Table | Binary Example (`5 = 0101`, `3 = 0011`) | Result |
|---|---|---|---|---|
| `&` | **Bitwise AND** | 1 only if BOTH bits are 1 | `0101 & 0011` -> `0001` | `1` |
| `\|` | **Bitwise OR** | 1 if AT LEAST ONE bit is 1 | `0101 \| 0011` -> `0111` | `7` |
| `^` | **Bitwise XOR** | 1 if bits are DIFFERENT, 0 if SAME | `0101 ^ 0011` -> `0110` | `6` |
| `~` | **Bitwise NOT** | Inverts all bits (Two's complement) | `~0101` -> `-(5 + 1)` | `-6` |
| `<<` | **Left Shift** | Shifts bits left by $n$, fills with 0s | `5 << 1` -> `5 * 2^1 = 10` (`1010`) | `10` |
| `>>` | **Right Shift** | Shifts bits right by $n$, discards right | `5 >> 1` -> `5 // 2^1 = 2` (`0010`) | `2` |

---

### 🧠 Understanding Bitwise NOT (`~`) & Two's Complement
Python represents negative integers using **Two's Complement Representation**.
* Mathematical Formula: `~x = -(x + 1)`
* For `x = 5`: `~5 = -(5 + 1) = -6`
* For `x = -6`: `~(-6) = -(-6 + 1) = 5`

---

### ⚡ Shift Operator Formulas
* **Left Shift (`x << n`)**: Multiplies `x` by $2^n$ -> `x * (2 ** n)`
* **Right Shift (`x >> n`)**: Performs integer division of `x` by $2^n$ -> `x // (2 ** n)`

> ❌ **CRITICAL ERROR: Bitwise Operators on Floats**:
> Attempting bitwise operations on floats (e.g. `23.3 ^ 23.3`) raises a **`TypeError`**! Bitwise operations are defined exclusively for integers.
> 
> ⚠️ **`^` is XOR, NOT Exponentiation!**
> * Power calculation: `23.3 ** 2 = 542.89` (Correct)
> * `23.3 ^ 2` -> `TypeError: unsupported operand type(s) for ^: 'float' and 'int'`


````python
# ============================================================
# 6️⃣ Bitwise Operators & Binary Representation
# ============================================================

a = 5   # Binary: 0101
b = 3   # Binary: 0011

print(f"a = {a} (binary: {bin(a)[2:].zfill(4)})")
print(f"b = {b} (binary: {bin(b)[2:].zfill(4)})")

print("\n--- 1. Bitwise AND (&) ---")
# 0101 & 0011 = 0001 (1)
print(f"a & b = {a & b} (binary: {bin(a & b)[2:].zfill(4)})")

print("\n--- 2. Bitwise OR (|) ---")
# 0101 | 0011 = 0111 (7)
print(f"a | b = {a | b} (binary: {bin(a | b)[2:].zfill(4)})")

print("\n--- 3. Bitwise XOR (^) ---")
# 0101 ^ 0011 = 0110 (6)
print(f"a ^ b = {a ^ b} (binary: {bin(a ^ b)[2:].zfill(4)})")
print(f"23 ^ 23 = {23 ^ 23} (XOR with identical value is always 0!)")

print("\n--- 4. Bitwise NOT (~) ---")
# ~x = -(x + 1) -> ~5 = -6
print(f"~a (~5) = {~a} (Formula: -(5 + 1) = -6)")
print(f"~(-6)   = {~(-6)} (Formula: -(-6 + 1) = 5)")

print("\n--- 5. Shift Operators (<< and >>) ---")
# Left Shift: 5 << 1 = 10, 5 << 2 = 20 (x * 2^n)
print(f"5 << 1 = {5 << 1} (5 * 2^1 = 10 | binary: {bin(5 << 1)[2:]})")
print(f"5 << 2 = {5 << 2} (5 * 2^2 = 20 | binary: {bin(5 << 2)[2:]})")

# Right Shift: 5 >> 1 = 2, 5 >> 2 = 1 (x // 2^n)
print(f"5 >> 1 = {5 >> 1} (5 // 2^1 = 2 | binary: {bin(5 >> 1)[2:].zfill(4)})")
print(f"5 >> 2 = {5 >> 2} (5 // 2^2 = 1 | binary: {bin(5 >> 2)[2:].zfill(4)})")
````

## 7️⃣ Special Operators: Identity (`is`, `is not`) & Membership (`in`, `not in`)

Python provides two specialized categories of operators for element containment and memory identity verification.

---

### 🔍 1. Membership Operators (`in`, `not in`)
Membership operators test whether a target element exists within a collection or sequence (`str`, `list`, `tuple`, `set`, `dict`).

| Syntax | Description | Example |
|---|---|---|
| `x in collection` | `True` if `x` is an element of `collection` | `3 in [1, 2, 3] -> True` |
| `x not in collection` | `True` if `x` is NOT in `collection` | `6 not in [1, 2, 3] -> True` |

> ⚠️ **Dictionary Membership Gotcha**:
> By default, `key in my_dict` searches **KEYS ONLY**.
> * `'a' in {'a': 100} -> True`
> * `100 in {'a': 100} -> False` (100 is a value, not a key!)
> * To check values: `100 in my_dict.values() -> True`
> 
> ⚠️ **Self-Membership Trap**:
> `lst = [1, 2, 3]; lst in lst` evaluates to **`False`** because the list container itself is not an element inside itself.

---

### 🧬 2. Identity Operators (`is`, `is not`) vs Value Equality (`==`)

| Operator | Comparison Type | What it checks | Equivalent expression |
|---|---|---|---|
| `==` | **Value Equality** | Do both objects have the same content / data? | `a.__eq__(b)` |
| `is` | **Identity (Reference)** | Do both variables point to the **exact same memory location**? | `id(a) == id(b)` |
| `is not` | **Non-Identity** | Do variables point to different memory locations? | `id(a) != id(b)` |

> 🏆 **Golden Rules for Identity Operators**:
> 1. Use **`==`** for checking value equality (e.g. `count == 10`, `name == "Rajan"`).
> 2. Use **`is`** only for checking identity, specifically singleton checks like **`x is None`** or `x is True`.
> 3. **Never** use `is` to compare numerical values or user-generated strings!


````python
# ============================================================
# 7️⃣ Membership Operators (in, not in)
# ============================================================

# 1. Lists & Sequences
my_list = [10, 20, 30, 40, 50]
print("--- Membership in List ---")
print(f"20 in my_list       : {20 in my_list}")        # True
print(f"99 in my_list       : {99 in my_list}")        # False
print(f"99 not in my_list   : {99 not in my_list}")    # True
print(f"my_list in my_list  : {my_list in my_list}")   # False (List is not an element of itself!)

# 2. Strings
my_str = "Data Engineering"
print("\n--- Membership in String ---")
print(f"'Data' in my_str    : {'Data' in my_str}")
print(f"'z' not in my_str   : {'z' not in my_str}")

# 3. Dictionaries: Keys vs Values Trap
my_dict = {"database": "BigQuery", "engine": "Spark", "rows": 1000}
print("\n--- Membership in Dictionary ---")
print(f"'database' in my_dict         : {'database' in my_dict}")         # True (Checks key)
print(f"'BigQuery' in my_dict         : {'BigQuery' in my_dict}")         # False! (Checks keys by default)
print(f"'BigQuery' in my_dict.values(): {'BigQuery' in my_dict.values()}")# True! (Explicitly checks values)
````

````python
# ============================================================
# 🧬 Identity Operators (is, is not) vs Equality (==)
# ============================================================

# 1. Mutable Objects (Lists)
list_a = [1, 2, 3]
list_b = [1, 2, 3]       # Separate list object with identical content
list_c = list_a          # Alias pointing to same memory address
list_d = list_a.copy()   # Shallow copy -> new memory address

print("--- Identity vs Equality on Lists ---")
print(f"list_a == list_b : {list_a == list_b} (Values are equal)")
print(f"list_a is list_b : {list_a is list_b} (Different objects in memory! id: {id(list_a)} vs {id(list_b)})")
print(f"list_a is list_c : {list_a is list_c} (Same object alias! id: {id(list_a)})")
print(f"list_a is list_d : {list_a is list_d} (copy() creates new object)")
print(f"list_a == list_d : {list_a == list_d} (copy() content matches)")

# 2. Best Practice: Singleton Checks with None
x = None
print("\n--- Singleton Checking ---")
print(f"x is None     : {x is None} (Recommended Pythonic idiom)")
print(f"x is not None : {x is not None}")
````

## 8️⃣ Memory Model, Object Interning & Mutability Traps in Functions

Python employs sophisticated memory optimization techniques for immutable objects, while strictly isolating mutable objects to prevent data corruption.

```
       IMMUTABLE OBJECTS                          MUTABLE OBJECTS
   (int, str, tuple, frozenset)                  (list, dict, set)
 ┌───────────────────────────────┐        ┌───────────────────────────────┐
 │ Python MAY reuse memory       │        │ Python NEVER reuses memory    │
 │ (Integer Caching & Interning) │        │ automatically for literals    │
 └──────────────┬────────────────┘        └──────────────┬────────────────┘
                │                                        │
                ▼                                        ▼
   a = 10 ──┐                               list1 = [1,2] ──> Memory A
            ├──> [Memory 10]                list2 = [1,2] ──> Memory B
   b = 10 ──┘
```

---

### 1. Integer Caching (Small Integer Interning)
* CPython pre-allocates an array of integer objects for small integers in the range **`-5` to `256`**.
* Whenever an integer in this range is created, Python returns a reference to the existing cached object.
* Outside this range, separate objects may be created depending on execution context.

### 2. String Interning
* Identical short string literals (identifiers) are interned and share the same memory address.
* Dynamically created strings (e.g. `"".join(...)`) create new objects in memory unless explicitly interned with `sys.intern()`.

### 3. Function Argument Passing & Mutability Hazards
Python passes arguments by **object reference (Pass-by-Assignment)**:
* **Passing Immutable Objects (Safe)**: Re-assigning or modifying creates a new local object; original variable outside function remains unchanged.
* **Passing Mutable Objects (Dangerous)**: In-place mutations (`.append()`, `.pop()`, `[0] = ...`) modify the caller's original object directly, leading to unexpected side-effects!


````python
# ============================================================
# 8️⃣ Object Interning & Memory Address Demonstration
# ============================================================

# 1. Small Integer Caching (-5 to 256)
int1 = 100
int2 = 100
print("--- Integer Caching ---")
print(f"int1 is int2 : {int1 is int2} | id(int1): {id(int1)}, id(int2): {id(int2)}")

# Modification creates new object
int1 = int1 + 5
print(f"After int1 + 5 -> id(int1): {id(int1)} (New integer 105 created)")

# 2. String Interning
str1 = "data_pipeline"
str2 = "data_pipeline"
str3 = "".join(["data_", "pipeline"])  # Created dynamically at runtime

print("\n--- String Interning ---")
print(f"str1 is str2 : {str1 is str2} (Literal interning)")
print(f"str1 == str3 : {str1 == str3} (Values equal)")
print(f"str1 is str3 : {str1 is str3} (Dynamic string has different memory address!)")

# 3. Mutable Objects (Lists are NEVER automatically reused)
l1 = [1, 2, 3]
l2 = [1, 2, 3]
print("\n--- Mutable Object Memory Isolation ---")
print(f"l1 == l2 : {l1 == l2} (Equal values)")
print(f"l1 is l2 : {l1 is l2} (Separate memory allocations: {id(l1)} vs {id(l2)})")
````

````python
# ============================================================
# ⚠️ Function Call Mutability Trap (Interview Classic)
# ============================================================

# Case 1: Immutable Parameter (SAFE)
def increment_counter(val):
    val = val + 1  # Creates a new local integer
    print(f"Inside function -> val: {val}, id(val): {id(val)}")

counter = 10
print(f"Before function -> counter: {counter}, id(counter): {id(counter)}")
increment_counter(counter)
print(f"After function  -> counter: {counter} (Unaltered!)")


# Case 2: Mutable Parameter (DANGEROUS IN-PLACE MUTATION)
def append_error_log(log_list, error_msg):
    log_list.append(error_msg)  # Modifies caller's list in-place!

app_logs = ["INIT_OK", "AUTH_SUCCESS"]
print(f"\nBefore append -> app_logs: {app_logs}")
append_error_log(app_logs, "DB_TIMEOUT")
print(f"After append  -> app_logs: {app_logs} <-- Modified outside function!")


# Case 3: Defensive Copying Strategy (SAFE PATTERN FOR PIPELINES)
def safe_process_records(records):
    local_copy = records.copy()  # Create shallow copy
    local_copy.append("BATCH_PROCESSED")
    return local_copy

original_records = [1, 2, 3]
processed = safe_process_records(original_records)
print(f"\nDefensive Copy Pattern:")
print(f"Original : {original_records}")
print(f"Processed: {processed}")
````

## 9️⃣ Operator Precedence & Associativity

When an expression contains multiple operators, Python determines the execution order using two rules:
1. **Operator Precedence**: Specifies which operator executes first (higher precedence binds tighter).
2. **Associativity**: Specifies the direction of evaluation (Left -> Right or Right -> Left) when adjacent operators have the **same** precedence.

---

### 🔢 Complete Operator Precedence Hierarchy (Highest to Lowest)

| Rank | Operator Category | Operators | Associativity |
|:---:|---|---|:---:|
| **1 (Highest)** | Parentheses / Grouping | `( )`, `[ ]`, `{ }` | Left-to-Right |
| **2** | Exponentiation | `**` | **Right-to-Left** ⚠️ |
| **3** | Unary Positive, Negative, NOT | `+x`, `-x`, `~x` | Right-to-Left |
| **4** | Multiplication, Division, Floor Div, Modulo | `*`, `/`, `//`, `%` | Left-to-Right |
| **5** | Addition & Subtraction | `+`, `-` | Left-to-Right |
| **6** | Bitwise Shifts | `<<`, `>>` | Left-to-Right |
| **7** | Bitwise AND | `&` | Left-to-Right |
| **8** | Bitwise XOR | `^` | Left-to-Right |
| **9** | Bitwise OR | `\|` | Left-to-Right |
| **10** | Comparisons, Identity & Membership | `==`, `!=`, `<`, `>`, `<=`, `>=`, `is`, `is not`, `in`, `not in` | Left-to-Right |
| **11** | Logical NOT | `not` | Right-to-Left |
| **12** | Logical AND | `and` | Left-to-Right |
| **13** | Logical OR | `or` | Left-to-Right |
| **14** | Walrus Operator (Assignment Expression) | `:=` | Right-to-Left |
| **15 (Lowest)** | Assignment Operators | `=`, `+=`, `-=`, `*=`, etc. | Right-to-Left |

---

### 🔍 Step-by-Step Expression Reduction Examples

#### Example 1: Exponentiation Right-Associativity (`2 ** 3 ** 2`)
* Evaluated as: `2 ** (3 ** 2)` = `2 ** 9` = **`512`**
* *(Left-associative would have produced `(2 ** 3) ** 2 = 8 ** 2 = 64` ❌)*

#### Example 2: Bitwise Shift with Addition (`8 >> 1 + 1`)
* Addition has higher precedence than Shift: `1 + 1 = 2`
* Shift evaluated next: `8 >> 2` = **`2`**

#### Example 3: Logical Precedence (`not False and True or False`)
1. `not False` -> `True`
2. `True and True` -> `True` (`and` executes before `or`)
3. `True or False` -> **`True`**


````python
# ============================================================
# 9️⃣ Precedence & Associativity Step-by-Step Verification
# ============================================================

# 1. Parentheses Override (Highest Precedence)
expr1 = (3 + 4) * 2  # (7) * 2 = 14
print(f"Parentheses: (3 + 4) * 2 = {expr1}")

# 2. Exponentiation is Right-to-Left Associative
expr2 = 2 ** 3 ** 2  # 2 ** (3 ** 2) = 2 ** 9 = 512
print(f"Exponentiation: 2 ** 3 ** 2 = {expr2}")

# 3. Unary Operators (+, -, ~)
val = 5
print(f"Unary (+val, -val, ~val): {+val}, {-val}, {~val}")

# 4. Arithmetic: Mul/Div precedes Add/Sub
expr4 = 3 + 4 * 2    # 3 + (4 * 2) = 3 + 8 = 11
print(f"Mul over Add: 3 + 4 * 2 = {expr4}")

# 5. Addition/Subtraction is Left-to-Right Associative
expr5 = 10 - 4 + 2   # (10 - 4) + 2 = 6 + 2 = 8
print(f"Left-Associative: 10 - 4 + 2 = {expr5}")

# 6. Shift and Arithmetic: Addition precedes Shift
expr6 = 8 >> 1 + 1   # 8 >> (1 + 1) = 8 >> 2 = 2
print(f"Shift with Add: 8 >> 1 + 1 = {expr6}")

# 7. Bitwise Precedence: & precedes ^ precedes |
expr7 = 5 & 3 | 2    # (5 & 3) | 2 = 1 | 2 = 3
print(f"Bitwise & over |: 5 & 3 | 2 = {expr7}")

# 8. Comparison vs Arithmetic: Arithmetic precedes Comparison
expr8 = 10 + 5 > 12  # (10 + 5) > 12 = 15 > 12 = True
print(f"Arithmetic over Comparison: 10 + 5 > 12 = {expr8}")

# 9. Logical Operators: not > and > or
expr9 = not False and True or False  # ((not False) and True) or False = (True and True) or False = True
print(f"Logical Precedence: not False and True or False = {expr9}")
````

## 🔟 Ternary Operator (Conditional Expressions)

The **ternary operator** (also called a *conditional expression*) allows for evaluating and returning a value based on a condition in a single, compact line.

### 📝 Syntax
```python
value_if_true if condition else value_if_false
```

### 🌟 Key Characteristics
* Compact alternative to multi-line `if-else` blocks for simple value assignments.
* Returns an expression, making it usable directly inside list comprehensions, return statements, and function arguments.
* Can be nested for multi-condition branching (though nesting beyond 2 levels is discouraged for readability).


````python
# ============================================================
# 🔟 Ternary Operator Patterns & Examples
# ============================================================

# 1. Basic Value Assignment
age = 18
status = "Eligible to vote" if age >= 18 else "Not eligible to vote"
print(f"Example 1 -> Age: {age} => {status}")

# 2. Inline Parity Check (Even / Odd)
number = 10
parity = "Even" if number % 2 == 0 else "Odd"
print(f"Example 2 -> Number: {number} => {parity}")

# 3. Nested Ternary Operator (Grading System)
marks = 75
grade = "A" if marks >= 80 else ("B" if marks >= 60 else "C")
print(f"Example 3 -> Marks: {marks} => Grade: {grade}")

# 4. Dynamic Function Invocation
def greet_morning(): return "Good morning, Engineer!"
def greet_evening(): return "Good evening, Engineer!"

hour = 15
greeting = greet_morning() if hour < 12 else greet_evening()
print(f"Example 4 -> Hour {hour}:00 => {greeting}")

# 5. List Comprehensions with Ternary Operator
data_points = [1, 2, 3, 4, 5, 6]
classified = ["Even" if n % 2 == 0 else "Odd" for n in data_points]
print(f"Example 5 -> Classified Data: {classified}")

# 6. Inline Temperature Alert
temp = 32
alert = "HOT_ALERT" if temp > 30 else "NORMAL"
print(f"Example 6 -> Temp {temp}°C => Status: {alert}")
````

## 1️⃣1️⃣ 🦭 The Walrus Operator (`:=`) Masterclass — Complete In-Depth Guide

> ⚡ **Requirement:** Python 3.8+ (PEP 572)  
> ✍️ **Author Reference:** Rajan

### What is the Walrus Operator?
The **Walrus Operator (`:=`)** is an **assignment expression** introduced in Python 3.8.

* ✅ **Assign a value to a variable as part of an expression**.
* ✅ **Evaluate, assign, and return** the value all in one step.
* ✅ Extremely useful in **`while` loops, `if` conditions, list comprehensions**, and for **avoiding expensive recomputations**.
* 🦭 **Why the name?** The symbol `:=` looks like the eyes and tusks of a walrus face!

```python
# Syntax
variable := expression
```


---
### 📌 Example 1: Walrus in `while` Loop

**Concept:**
* **Traditional way:** Assign the value first before the loop, then check it in the condition, and re-assign at the end of the body — requires writing the input/read expression **twice**.
* **Walrus way:** Assign and check in a **single expression** inside the `while` condition.

> 💡 This shrinks lines of code and eliminates the need for duplicate pre-loop assignments.


````python
# ❌ WITHOUT Walrus Operator
# Traditional approach: assign first, then check in loop
# Problem: We had to write the input reading TWICE — once before loop and once inside.

# Simulating stream inputs: "rajan", "raj", "quit"
input_stream_1 = iter(["rajan", "raj", "quit"])
def simulated_input(prompt=""):
    val = next(input_stream_1)
    print(f"{prompt}{val}")
    return val

line = simulated_input("Input (type 'quit' to exit): ")          # Step 1: assign before loop
while line != "quit":                                           # Step 2: check condition
    print(f"You typed: {line}")
    line = simulated_input("Input (type 'quit' to exit): ")      # Step 3: re-assign inside loop (duplicate!)
````

````python
# ✅ WITH Walrus Operator
# Assign and check in a single expression inside the while condition
# Benefit: No duplicate input call. Cleaner, shorter, and zero redundancy.

input_stream_walrus_1 = iter(["rajan", "raj", "quit"])
def simulated_input_w(prompt=""):
    val = next(input_stream_walrus_1)
    print(f"{prompt}{val}")
    return val

while (line := simulated_input_w("Input (type 'quit' to exit): ")) != "quit":   # Assign + check in ONE line
    print(f"You typed: {line}")                                                 # 'line' is immediately available!
````

---
### 📌 Example 2: Walrus in `if` Conditions

**Concept:**
* **Traditional way:** Assign a value to a variable first, then separately evaluate it in an `if` condition on the next line.
* **Walrus way:** Assign **and** check at the same time inside the `if` condition.

> 📝 Empty strings `""`, `None`, or `False` evaluate as **Falsy**, so the `if` block is skipped for them.


````python
# ❌ WITHOUT Walrus Operator
# Assign first, then evaluate in the if condition (extra separated line)

users = ["alice", "bob", "charlie", "ed"]

for user in users:
    length = len(user)               # Step 1: assign separately
    if length > 3:                   # Step 2: check in the next line
        print(f"{user} is valid and has length {length}")
    else:
        print(f"{user} is too short")
````

````python
# ✅ WITH Walrus Operator
# Assign and check in the same if condition
# Benefit: Combines assignment and condition check in one concise line.

users = ["alice", "bob", "charlie", "ed"]

for user in users:
    if (length := len(user)) > 3:   # Assign to 'length' AND check > 3 in one line
        print(f"{user} is valid and has length {length}")
    else:
        print(f"{user} is too short")
````

---
### 📌 Example 3: Avoid Recomputation

**Concept:**
* **Traditional way:** Compute an expression inside the `if` condition, then compute it **again** in the body to use the value.
* **Walrus way:** Compute **once**, assign it, and reuse the variable immediately — zero repeated calculation.

> 💡 This is especially critical when the computation is expensive (such as complex math, regular expressions, or database queries).


````python
# ❌ WITHOUT Walrus Operator
# The square is computed TWICE — once in condition, once in print!

numbers = [2, 4, 6, 8, 10]

for num in numbers:
    if num * num > 20:                       # Computed here ❶
        print(f"{num}^2 = {num * num} > 20") # Computed again here ❷ — wasteful!
````

````python
# ✅ WITH Walrus Operator
# Compute once, assign to 'sq', reuse in both condition AND print
# Benefit: Exactly one computation per iteration, stored and reused cleanly.

numbers = [2, 4, 6, 8, 10]

for num in numbers:
    if (sq := num * num) > 20:         # Computed ONCE, stored in 'sq'
        print(f"{num}^2 = {sq} > 20")  # Reuse 'sq' — no recomputation!
````

---
### 📌 Example 4: List Comprehension with Walrus

**Concept:**
* **Traditional way:** You cannot easily reuse the filtered value (e.g. the square) as both the filter condition **and** the output in a single comprehension without recalculating it twice.
* **Walrus way:** Assign the computed value in the `if` filter clause, then directly reuse it in the **output expression**.

### 📚 List Comprehension Syntax Reminder:
```python
[expression  for item in iterable  if condition]
#  ↑ output     ↑ loop variable        ↑ filter
```
* **expression** -> what goes into the new list
* **item** -> variable iterating over the collection
* **condition** -> optional boolean filter


````python
# ❌ WITHOUT Walrus Operator
# Can't reuse the computed square in both condition and output directly
# Workaround: compute it in the output expression as well (double computation!)

values = [1, 5, 8, 10, 12]

squared_over_50 = [x * x for x in values if x * x > 50]  # x*x computed TWICE per element

print(f"Squares > 50: {squared_over_50}")
# Problem: For every element, x*x is evaluated in the 'if' AND in the output — redundant!
````

````python
# ✅ WITH Walrus Operator
# Assign square to 'sq' in the if-clause, reuse 'sq' as the output expression
# Benefit: x*x computed ONCE per element. 'sq' is captured and yielded as output.

values = [1, 5, 8, 10, 12]

#        ↓ output uses 'sq'     ↓ walrus computes AND stores sq, then filters
squared_over_50 = [sq for x in values if (sq := x * x) > 50]

print(f"Squares > 50: {squared_over_50}")
````

---
### 📌 Example 5: Nested Loops & Early Matrix Breakout

**Concept:**
* **Traditional way:** Find the first value satisfying a condition using nested loops, requiring an extra `found` flag variable to break out of both loops and separate assignment.
* **Walrus way:** The value is assigned inline during the condition check (`val := num`), making the intent clearer and removing pre-assignments.


````python
# ❌ WITHOUT Walrus Operator
# Must assign num to a separate variable after checking

matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

found = False
result = None  # need a separate variable to store the found value

for row in matrix:
    for num in row:
        if num > 5:         # check condition
            result = num    # assign separately after the check
            found = True
            break
    if found:
        break

print(f"First number > 5: {result}")
````

````python
# ✅ WITH Walrus Operator
# Assign and check in a single expression — 'val' is immediately available
# Benefit: 'val' is captured via walrus during the condition check — no separate assignment.

matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

found = False

for row in matrix:
    for num in row:
        if (val := num) > 5:      # assign to 'val' AND check in one line
            print(f"First number > 5: {val}")
            found = True
            break
    if found:
        break
````

### 📌 Example 6: Running Total Accumulator Until Limit

**Concept:**
* **Traditional way:** Read input before the loop AND inside the loop — the input statement is written **twice**.
* **Walrus way:** Read input inside the `while` condition — captured once, zero duplication.

> 📝 Each input number is added to a running total. The loop stops when the number entered is $\le 0$ **or** the total exceeds 50.


````python
# ❌ WITHOUT Walrus Operator
# Must read input before the loop AND again at the end of every iteration

# Simulating user input stream
total_stream_without = iter([15, 20, 25, 0])
def read_number_without(prompt=""):
    n = next(total_stream_without)
    print(f"{prompt}{n}")
    return n

total = 0
num = int(read_number_without("Enter number (0 or negative to stop): "))   # read BEFORE loop (duplicate!)

while num > 0:
    total += num
    print(f"Running total: {total}")
    if total > 50:
        print("Total exceeded 50! Loop stopped.")
        break
    num = int(read_number_without("Enter number (0 or negative to stop): "))  # read INSIDE loop (duplicate!)
````

````python
# ✅ WITH Walrus Operator
# Read input directly in the while condition — only written ONCE
# Benefit: input reading is written ONCE. Loop condition is clear and self-contained.

total_stream_with = iter([15, 20, 25, 0])
def read_number_with(prompt=""):
    n = next(total_stream_with)
    print(f"{prompt}{n}")
    return n

total = 0

while (num := int(read_number_with("Enter number (0 or negative to stop): "))) > 0:   # read + check in one line
    total += num
    print(f"Running total: {total}")
    if total > 50:
        print("Total exceeded 50! Loop stopped.")
        break
````

### 📌 Example 7: Filter Stream / User Input (Advanced)

**Concept:**
* **Traditional way:** Read input before and inside the loop (**twice**), then filter inside.
* **Walrus way:** Read input **once** in the `while` condition, filter inside — compact and clean.

> 📝 Only **even numbers** are collected. The loop stops when the user enters `0`.


````python
# ❌ WITHOUT Walrus Operator
# Input must be read before the loop AND repeated at its end — duplication

stream_data_without = iter([4, 7, 10, 15, 18, 0])
def get_stream_item_without(prompt=""):
    val = next(stream_data_without)
    print(f"{prompt}{val}")
    return val

even_numbers = []
print("Enter numbers (0 to stop):")

n = int(get_stream_item_without("Number (0 to stop): "))         # read BEFORE loop
while n != 0:
    if n % 2 == 0:                                               # filter for even numbers
        even_numbers.append(n)
    n = int(get_stream_item_without("Number (0 to stop): "))     # read INSIDE loop (duplicate!)

print(f"Even numbers entered: {even_numbers}")
````

````python
# ✅ WITH Walrus Operator
# Read input in the while condition — assigns 'n' and checks != 0 at once
# Benefit: input reading written ONCE. Loop is self-contained, shorter, and easier to maintain.

stream_data_with = iter([4, 7, 10, 15, 18, 0])
def get_stream_item_with(prompt=""):
    val = next(stream_data_with)
    print(f"{prompt}{val}")
    return val

even_numbers = []
print("Enter numbers (0 to stop):")

while (n := int(get_stream_item_with("Number (0 to stop): "))) != 0:  # assign + check in one line
    if n % 2 == 0:                                                    # filter: keep only even numbers
        even_numbers.append(n)

print(f"Even numbers entered: {even_numbers}")
````

## 1️⃣2️⃣ Data Engineering Connections & Golden Best Practices

In production Data Engineering, understanding operators, mutability, and type behavior is critical for building performant, defect-free data pipelines.

### 🏗️ Data Engineering Applications
1. **Defensive Schema Parsing**: Using `isinstance()` and ternary guards to sanitize dirty JSON payloads before writing to BigQuery / Parquet.
2. **Memory Safety in Batch Transforms**: Avoiding inadvertent in-place list mutations (`+=`) when passing data chunks across pipeline workers.
3. **High-Speed Bitmask Filtering**: Using bitwise operators (`&`, `|`) to parse telemetry status flags at microsecond speeds.
4. **Walrus Streaming Ingestion**: Using `:=` inside chunks generators to pull from Kafka / GCS buckets without memory bloat.

---

### 🏆 Master Summary: Golden Rules Cheat Sheet

| Topic | Golden Rule |
|---|---|
| **Dynamic Typing** | Use `isinstance(obj, (int, float))` for polymorphic type checking; never compare `type(x) == int` when inheritance matters. |
| **Division** | `/` always returns `float`. `//` truncates towards negative infinity (`-9 // 2 == -5`). |
| **In-place Mutation** | `a += b` mutates mutable lists in-place (`__iadd__`); use `a = a + b` or `.copy()` when aliasing must be avoided. |
| **Logical Operators** | `and` / `or` return the determining operand, not just boolean. Use this for fallback values: `val = config_val or default_val`. |
| **Bitwise** | Bitwise operators (`&`, `|`, `^`, `~`) operate only on integers. `~x == -(x + 1)`. `^` is XOR, never power! |
| **Identity vs Equality**| Use `==` for values (`list1 == list2`). Use `is` only for singletons (`x is None`). |
| **Memory Interning** | Python caches integers -5 to 256 and string identifiers; never rely on `is` for numbers outside this range. |
| **Precedence** | `**` is right-associative (`2**3**2 = 512`). Arithmetic > Bitwise > Comparison > Logical. Use `()` for clarity! |
| **Walrus Operator** | Use `:=` in loops, conditions, and comprehensions to avoid duplicated computations and lines of code. |


````python
# ============================================================
# 1️⃣2️⃣ Practical Data Engineering Pipeline Record Cleaner
# ============================================================

def transform_and_validate_batch(raw_batch):
    '''
    Simulates a production ETL pipeline worker that:
    1. Sanitizes nulls / defaults using short-circuit 'or'
    2. Uses walrus ':=' to filter and clean records in a single pass
    3. Employs bitwise flags to classify sensor status
    4. Ensures zero mutation of original input records
    '''
    # Bitmask constants for sensor telemetry
    FLAG_ACTIVE   = 1 << 0  # 0001 (1)
    FLAG_CRITICAL = 1 << 1  # 0010 (2)
    FLAG_VERIFIED = 1 << 2  # 0100 (4)

    clean_records = []

    for record in raw_batch:
        # 1. Fallback defaulting using short-circuit 'or'
        sensor_id = record.get("sensor_id") or "UNKNOWN_SENSOR"
        
        # 2. Defensive type checking
        raw_reading = record.get("reading")
        if not isinstance(raw_reading, (int, float)):
            continue  # Skip corrupt record
        
        # 3. Walrus operator to compute calibrated metric inline
        if (calibrated := round(raw_reading * 1.05, 2)) > 10.0:
            # 4. Bitwise status flag composition
            status_flag = FLAG_ACTIVE | FLAG_VERIFIED
            if calibrated > 50.0:
                status_flag |= FLAG_CRITICAL

            clean_records.append({
                "sensor_id": sensor_id,
                "calibrated_reading": calibrated,
                "status_code": status_flag,
                "is_critical": bool(status_flag & FLAG_CRITICAL)
            })

    return clean_records

# Sample messy incoming pipeline batch
raw_batch_data = [
    {"sensor_id": "SN-101", "reading": 24.5},
    {"sensor_id": "", "reading": 62.1},          # Missing sensor_id -> defaults to UNKNOWN_SENSOR
    {"sensor_id": "SN-103", "reading": "CORRUPT"},# Invalid type -> skipped
    {"sensor_id": "SN-104", "reading": 5.0},     # Below threshold (< 10) -> skipped
    {"sensor_id": "SN-105", "reading": 88.4}     # Critical reading
]

processed_batch = transform_and_validate_batch(raw_batch_data)

print("=== Cleaned & Transformed Batch Output ===")
for rec in processed_batch:
    print(rec)
````
