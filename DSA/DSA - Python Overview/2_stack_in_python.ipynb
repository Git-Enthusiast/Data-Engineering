```python

"""

---------------------------------------------------------------------------
THEORY
---------------------------------------------------------------------------

A stack is a linear data structure that follows the
LIFO principle:

    LIFO = Last In, First Out

This means the most recently added item is the first item removed.

Real-life examples:
- A stack of plates
- Browser history navigation
- Undo functionality in editors
- Call stack in programming languages

Example:
    push(10)
    push(20)
    push(30)

Now if we pop():
    30 comes out first
Then:
    20
Then:
    10

Why stack is useful?
- simple
- fast for insertion/removal at the top
- helps in recursion, parsing, expression evaluation, backtracking

Important stack terms:
- top: the last inserted element
- push: insert at the top
- pop: remove from the top
- peek/top: view the top element without removing
- overflow: trying to push into a full bounded stack
- underflow: trying to pop from an empty stack

Time complexity:
- push: O(1)
- pop: O(1)
- peek: O(1)
- is_empty: O(1)
- size: O(1)

---------------------------------------------------------------------------
1) STACK USING PYTHON LIST
---------------------------------------------------------------------------

A Python list can behave like a stack because:
- append() adds to the end in O(1) amortized time
- pop() removes from the end in O(1)

This is the simplest and most common stack implementation.

---------------------------------------------------------------------------
2) STACK USING collections.deque
---------------------------------------------------------------------------

deque stands for "double-ended queue".
It is efficient for appending and popping from both ends.

For stack behavior:
- append() -> push
- pop()    -> pop

deque is very efficient and often preferred when you need a fast stack/queue.

---------------------------------------------------------------------------
3) STACK USING queue.LifoQueue
---------------------------------------------------------------------------

LifoQueue is a thread-safe stack implementation.
It uses:
- put() -> push
- get() -> pop

Useful in multi-threaded programs.
It also supports a maximum size.

Note:
- It raises exceptions such as queue.Full and queue.Empty
- You can use timeout to avoid blocking forever

---------------------------------------------------------------------------
4) CUSTOM STACK CLASS
---------------------------------------------------------------------------

A custom class is ideal when you want:
- clear method names
- error handling
- max size
- reusable code
- teaching purposes

This file includes a custom class with all common operations.

---------------------------------------------------------------------------
RUN THIS FILE
---------------------------------------------------------------------------

You can run this file directly to see demonstrations.
Interactive menu is included but not executed automatically unless you call it.

"""
```

```python

from collections import deque
from queue import LifoQueue, Full, Empty


# ===========================================================================
# PART A: SIMPLE STACK USING A PYTHON LIST
# ===========================================================================

class ListStack:
    """
    Stack implementation using Python list.

    This class is simple and perfect for beginners.
    """

    def __init__(self, max_size=None):
        """
        Initialize the stack.

        Parameters:
            max_size (int | None): Optional size limit.
                                  If None, stack is unbounded.
        """
        self.items = []
        self.max_size = max_size

    def is_empty(self):
        """Return True if stack has no elements."""
        return len(self.items) == 0

    def is_full(self):
        """
        Return True if the stack is full.

        For an unbounded stack, this always returns False.
        """
        if self.max_size is None:
            return False
        return len(self.items) >= self.max_size

    def size(self):
        """Return the number of elements in the stack."""
        return len(self.items)

    def push(self, item):
        """
        Add an item to the top of the stack.

        Raises:
            OverflowError: if stack has reached max_size.
        """
        if self.is_full():
            raise OverflowError("Stack Overflow: cannot push into a full stack.")
        self.items.append(item)

    def pop(self):
        """
        Remove and return the top item.

        Raises:
            IndexError: if stack is empty.
        """
        if self.is_empty():
            raise IndexError("Stack Underflow: cannot pop from an empty stack.")
        return self.items.pop()

    def peek(self):
        """
        Return the top item without removing it.

        Raises:
            IndexError: if stack is empty.
        """
        if self.is_empty():
            raise IndexError("Cannot peek: stack is empty.")
        return self.items[-1]

    def clear(self):
        """Remove all items."""
        self.items.clear()

    def __repr__(self):
        return f"ListStack(items={self.items}, max_size={self.max_size})"

```

```python

# ===========================================================================
# PART B: STACK USING collections.deque
# ===========================================================================

class DequeStack:
    """
    Stack implementation using collections.deque.

    deque is very efficient for append/pop operations.
    """

    def __init__(self, max_size=None):
        self.items = deque()
        self.max_size = max_size

    def is_empty(self):
        return len(self.items) == 0

    def is_full(self):
        if self.max_size is None:
            return False
        return len(self.items) >= self.max_size

    def size(self):
        return len(self.items)

    def push(self, item):
        if self.is_full():
            raise OverflowError("Stack Overflow: cannot push into a full stack.")
        self.items.append(item)

    def pop(self):
        if self.is_empty():
            raise IndexError("Stack Underflow: cannot pop from an empty stack.")
        return self.items.pop()

    def peek(self):
        if self.is_empty():
            raise IndexError("Cannot peek: stack is empty.")
        return self.items[-1]

    def clear(self):
        self.items.clear()

    def __repr__(self):
        return f"DequeStack(items={list(self.items)}, max_size={self.max_size})"

```

```python

# ===========================================================================
# PART C: STACK USING queue.LifoQueue
# ===========================================================================

class LifoQueueStack:
    """
    Thread-safe stack using queue.LifoQueue.

    Methods:
        put() -> push
        get() -> pop

    This implementation is good when thread safety matters.
    """

    def __init__(self, max_size=0):
        """
        max_size:
            0 means infinite size in queue.LifoQueue
            any positive integer sets a fixed capacity
        """
        self.items = LifoQueue(maxsize=max_size)

    def is_empty(self):
        return self.items.empty()

    def is_full(self):
        return self.items.full()

    def size(self):
        return self.items.qsize()

    def push(self, item, block=True, timeout=None):
        """
        Push an item into the stack.

        Parameters:
            block: if True, waits until space is available
            timeout: max seconds to wait before raising Full
        """
        self.items.put(item, block=block, timeout=timeout)

    def pop(self, block=True, timeout=None):
        """
        Pop the top item from the stack.

        Parameters:
            block: if True, waits until item becomes available
            timeout: max seconds to wait before raising Empty
        """
        return self.items.get(block=block, timeout=timeout)

    def peek(self):
        """
        Peek is not directly supported by LifoQueue.
        We must remove and then put the item back.

        Note: this is not ideal for high-performance situations.
        """
        if self.is_empty():
            raise Empty("Cannot peek: stack is empty.")
        temp = self.pop()
        self.push(temp)
        return temp

    def __repr__(self):
        return f"LifoQueueStack(size={self.size()}, max_size={self.items.maxsize})"

```

```python

# ===========================================================================
# PART D: CUSTOM STACK WITH BETTER ERROR MESSAGES
# ===========================================================================

class Stack:
    """
    A clean custom stack class for learning and reuse.

    Features:
    - push
    - pop
    - peek
    - is_empty
    - is_full
    - size
    - clear
    - max_size support
    """

    def __init__(self, max_size=None):
        self._items = []
        self.max_size = max_size

    def is_empty(self):
        return len(self._items) == 0

    def is_full(self):
        if self.max_size is None:
            return False
        return len(self._items) >= self.max_size

    def size(self):
        return len(self._items)

    def push(self, item):
        if self.is_full():
            raise OverflowError(f"Stack Overflow: max_size={self.max_size}")
        self._items.append(item)

    def pop(self):
        if self.is_empty():
            raise IndexError("Stack Underflow: pop from empty stack")
        return self._items.pop()

    def peek(self):
        if self.is_empty():
            raise IndexError("Cannot peek from empty stack")
        return self._items[-1]

    def clear(self):
        self._items.clear()

    def to_list(self):
        """Return a copy of stack elements from bottom to top."""
        return self._items.copy()

    def __len__(self):
        return self.size()

    def __repr__(self):
        return f"Stack({self._items}, max_size={self.max_size})"

```

```python

# ===========================================================================
# PART E: DEMONSTRATION FUNCTIONS
# ===========================================================================

def demo_basic_operations():
    """
    Demonstrate push, pop, peek, size, empty, and full behavior.
    """
    print("\n=== DEMO: Basic Stack Operations ===")
    s = Stack(max_size=5)

    print("Initial stack:", s)
    print("is_empty:", s.is_empty())
    print("is_full:", s.is_full())
    print("size:", s.size())

    print("\nPushing values: 10, 20, 30")
    s.push(10)
    s.push(20)
    s.push(30)

    print("Stack now:", s)
    print("peek:", s.peek())
    print("size:", s.size())

    popped = s.pop()
    print("popped:", popped)
    print("Stack after pop:", s)

    print("is_empty:", s.is_empty())
    print("is_full:", s.is_full())


def demo_overflow_and_underflow():
    """
    Show what happens when a stack is full or empty.
    """
    print("\n=== DEMO: Overflow and Underflow ===")

    s = Stack(max_size=2)

    try:
        s.pop()
    except Exception as e:
        print("Underflow example:", repr(e))

    try:
        s.push("A")
        s.push("B")
        s.push("C")
    except Exception as e:
        print("Overflow example:", repr(e))


def demo_list_stack():
    print("\n=== DEMO: ListStack ===")
    s = ListStack(max_size=3)
    s.push(1)
    s.push(2)
    print("Stack:", s)
    print("Top:", s.peek())
    print("Pop:", s.pop())
    print("After pop:", s)


def demo_deque_stack():
    print("\n=== DEMO: DequeStack ===")
    s = DequeStack(max_size=3)
    s.push("x")
    s.push("y")
    print("Stack:", s)
    print("Top:", s.peek())
    print("Pop:", s.pop())
    print("After pop:", s)


def demo_lifoqueue_stack():
    print("\n=== DEMO: LifoQueueStack ===")
    s = LifoQueueStack(max_size=3)
    s.push(100)
    s.push(200)
    print("Stack size:", s.size())
    print("Top:", s.peek())
    print("Pop:", s.pop())
    print("After pop size:", s.size())

```

```python

# ===========================================================================
# PART F: COMMON STACK APPLICATIONS
# ===========================================================================

def reverse_string(text):
    """
    Reverse a string using a stack.

    Example:
        "python" -> "nohtyp"
    """
    stack = Stack()
    for ch in text:
        stack.push(ch)

    reversed_chars = []
    while not stack.is_empty():
        reversed_chars.append(stack.pop())

    return "".join(reversed_chars)


def is_balanced_parentheses(expression):
    """
    Check whether parentheses/brackets/braces are balanced.

    Supports:
    (), {}, []

    Examples:
        "{[()]}"   -> True
        "{[(])}"   -> False
    """
    stack = Stack()
    matching = {')': '(', ']': '[', '}': '{'}

    for ch in expression:
        if ch in "([{":
            stack.push(ch)
        elif ch in ")]}":
            if stack.is_empty():
                return False
            if stack.pop() != matching[ch]:
                return False

    return stack.is_empty()


def decimal_to_binary(n):
    """
    Convert a decimal number to binary using a stack.

    Example:
        10 -> "1010"
    """
    if n == 0:
        return "0"

    stack = Stack()
    while n > 0:
        stack.push(str(n % 2))
        n //= 2

    binary = []
    while not stack.is_empty():
        binary.append(stack.pop())

    return "".join(binary)


def next_greater_element(arr):
    """
    Example of a classic stack problem.

    For each element, find the next greater element to its right.
    If none exists, use -1.

    Example:
        [4, 5, 2, 25]
        -> [5, 25, 25, -1]
    """
    result = [-1] * len(arr)
    stack = []  # stores indices

    for i, value in enumerate(arr):
        while stack and arr[stack[-1]] < value:
            idx = stack.pop()
            result[idx] = value
        stack.append(i)

    return result


def undo_redo_demo():
    """
    Very small conceptual demo for undo/redo using stacks.

    In real applications, there are usually two stacks:
    - undo stack
    - redo stack
    """
    undo_stack = Stack()
    redo_stack = Stack()

    # User performs actions
    undo_stack.push("type A")
    undo_stack.push("type B")
    undo_stack.push("delete C")

    # Undo the last action
    last_action = undo_stack.pop()
    redo_stack.push(last_action)

    return {
        "undo_stack_after_undo": undo_stack.to_list(),
        "redo_stack_after_undo": redo_stack.to_list(),
    }

```

```python

# ===========================================================================
# PART G: INTERACTIVE STACK MENU
# ===========================================================================

def interactive_stack_menu():
    """
    Interactive program for stack operations.

    This mirrors the kind of program shown in tutorials:
    - push
    - pop
    - peek
    - display
    - size
    - exit

    Note:
        This function is not called automatically.
        Uncomment the call in main() if you want to use it.
    """
    stack = Stack(max_size=5)

    while True:
        print("\n--- STACK MENU ---")
        print("1. Push")
        print("2. Pop")
        print("3. Peek")
        print("4. Display")
        print("5. Size")
        print("6. Is Empty")
        print("7. Is Full")
        print("8. Clear")
        print("9. Exit")

        choice = input("Enter choice: ").strip()

        try:
            if choice == "1":
                item = input("Enter item to push: ")
                stack.push(item)
                print("Pushed successfully.")

            elif choice == "2":
                print("Popped:", stack.pop())

            elif choice == "3":
                print("Top item:", stack.peek())

            elif choice == "4":
                print("Stack (bottom -> top):", stack.to_list())

            elif choice == "5":
                print("Size:", stack.size())

            elif choice == "6":
                print("Is Empty:", stack.is_empty())

            elif choice == "7":
                print("Is Full:", stack.is_full())

            elif choice == "8":
                stack.clear()
                print("Stack cleared.")

            elif choice == "9":
                print("Exiting menu.")
                break

            else:
                print("Invalid choice. Try again.")

        except Exception as e:
            print("Error:", e)

```

```python

# ===========================================================================
# PART H: EDGE CASE DISCUSSION
# ===========================================================================

EDGE_CASE_NOTES = """
1. Empty stack pop
   - Popping an empty stack is called underflow.
   - Always check is_empty() before pop().

2. Full stack push
   - If the stack has a limit, pushing beyond it causes overflow.
   - Always check is_full() before push().

3. Peek on empty stack
   - Similar to pop from empty stack, but without removal.

4. LifoQueue blocking behavior
   - put() may block if stack is full.
   - get() may block if stack is empty.
   - Use timeout to avoid waiting forever.

5. Type of elements
   - Python stacks can store mixed types:
     numbers, strings, lists, objects, etc.
   - For learning, use consistent types where possible.

6. Recursive thinking
   - The program call stack is different from our custom stack,
     but the same LIFO idea applies.

7. List vs deque
   - list is easiest
   - deque is efficient and excellent for stack usage
   - LifoQueue is thread-safe and can block

8. When not to use a stack
   - When you need random access to any element
   - When ordering should be FIFO instead of LIFO
"""

```
