# 1. Counter

# What is Counter?

`Counter` is a subclass of dictionary used for counting hashable objects.

It stores:

```python
item -> frequency
```

Example:

```python
from collections import Counter

text = "banana"
count = Counter(text)

print(count)
```

Output:

```python
Counter({'a': 3, 'n': 2, 'b': 1})
```

---

# How Counter Works Internally

Normally with dictionary:

```python
freq = {}

for ch in text:
    if ch in freq:
        freq[ch] += 1
    else:
        freq[ch] = 1
```

Counter does this automatically.

---

# Creating Counter

## From String

```python
Counter("hello")
```

Output:

```python
Counter({'l': 2, 'h': 1, 'e': 1, 'o': 1})
```

---

## From List

```python
Counter([1,1,1,2,2,3])
```

Output:

```python
Counter({1:3, 2:2, 3:1})
```

---

## From Dictionary

```python
Counter({'a':5, 'b':2})
```

---

# Most Common Methods in Counter

## elements()

Returns iterator containing elements repeated according to count.

```python
c = Counter(a=2, b=3)

print(list(c.elements()))
```

Output:

```python
['a', 'a', 'b', 'b', 'b']
```

---

## most_common()

Returns most frequent elements.

```python
c = Counter("banana")

print(c.most_common())
```

Output:

```python
[('a', 3), ('n', 2), ('b', 1)]
```

Top 2:

```python
print(c.most_common(2))
```

Output:

```python
[('a', 3), ('n', 2)]
```

---

## update()

Adds counts.

```python
c = Counter("hello")
c.update("world")

print(c)
```

---

## subtract()

Subtract counts.

```python
c1 = Counter(a=5, b=2)
c1.subtract({'a':2})

print(c1)
```

Output:

```python
Counter({'a':3, 'b':2})
```

---

# Counter Operations

```python
c1 = Counter(a=3, b=1)
c2 = Counter(a=1, b=2)
```

Addition:

```python
print(c1 + c2)
```

Subtraction:

```python
print(c1 - c2)
```

Intersection:

```python
print(c1 & c2)
```

Union:

```python
print(c1 | c2)
```

---

# Time Complexity of Counter

| Operation | Complexity |
|---|---|
| access | O(1) |
| update | O(1) average |
| counting n elements | O(n) |

# 2. defaultdict

# What is defaultdict?

`defaultdict` is a dictionary subclass.

It automatically provides default values for missing keys.

Normal dictionary:

```python
d = {}

print(d['a'])
```

Error:

```python
KeyError
```

With defaultdict:

```python
from collections import defaultdict

d = defaultdict(int)

print(d['a'])
```

Output:

```python
0
```

---

# How defaultdict Works

When key does not exist:

- it automatically creates it
- assigns default value

The default value depends on the function passed.

---

# Common Default Factories

| Factory | Default Value |
|---|---|
| int | 0 |
| float | 0.0 |
| str | '' |
| list | [] |
| set | set() |

---

# defaultdict with list

```python
from collections import defaultdict

students = defaultdict(list)

students['python'].append('Rajan')
students['python'].append('Aman')

print(students)
```

Output:

```python
defaultdict(<class 'list'>, {'python': ['Rajan', 'Aman']})
```

---

# Grouping Example

```python
words = ['apple', 'ant', 'banana', 'bat']

result = defaultdict(list)

for word in words:
    result[word[0]].append(word)

print(result)
```

Output:

```python
{'a': ['apple', 'ant'], 'b': ['banana', 'bat']}
```

---

# Time Complexity

Same as dictionary.

| Operation | Complexity |
|---|---|
| insert | O(1) |
| lookup | O(1) |
| delete | O(1) |

---

# Real Use Cases

- grouping data
- graph adjacency lists
- counting
- organizing categories
- dynamic nested dictionaries

# 3. deque

# What is deque?

`deque` means:

```text
Double Ended Queue
```

It allows insertion and deletion from both ends efficiently.

---

# Why deque Instead of List?

Python lists are fast at:

```python
append()
pop()
```

But slow at:

```python
insert(0, x)
pop(0)
```

because elements must shift.

`deque` solves this.

---

# Creating deque

```python
from collections import deque

d = deque([1,2,3])
print(d)
```

---

# Main Methods

## append()

Add to right.

```python
d.append(4)
```

---

## appendleft()

Add to left.

```python
d.appendleft(0)
```

---

## pop()

Remove from right.

```python
d.pop()
```

---

## popleft()

Remove from left.

```python
d.popleft()
```

---

# Rotation

```python
d.rotate(1)
```

Right rotation.

```python
d.rotate(-1)
```

Left rotation.

---

# Limited deque

```python
d = deque(maxlen=3)
```

If full:

- oldest elements automatically removed.

Example:

```python
d.append(1)
d.append(2)
d.append(3)
d.append(4)

print(d)
```

Output:

```python
deque([2,3,4], maxlen=3)
```

---

# Time Complexity

| Operation | Complexity |
|---|---|
| append | O(1) |
| appendleft | O(1) |
| pop | O(1) |
| popleft | O(1) |

---

# Real Use Cases

- queues
- stacks
- sliding window problems
- browser history
- caching
- task scheduling

# 4. namedtuple

# What is namedtuple?

A `namedtuple` is like a tuple but with named fields.

Normal tuple:

```python
student = ('Rajan', 21)
```

Access:

```python
student[0]
```

Hard to understand.

With namedtuple:

```python
from collections import namedtuple

Student = namedtuple('Student', ['name', 'age'])

s = Student('Rajan', 21)

print(s.name)
print(s.age)
```

---

# Advantages

- readable
- immutable
- memory efficient
- cleaner than dictionaries

---

# Methods

## _asdict()

```python
print(s._asdict())
```

---

## _replace()

```python
s2 = s._replace(age=22)
```

---

## _fields

```python
print(s._fields)
```

---

# Time Complexity

Tuple-like access.

| Operation | Complexity |
|---|---|
| access | O(1) |

# 5. OrderedDict

# What is OrderedDict?

Maintains insertion order.

Before Python 3.7:

normal dictionaries did not guarantee order.

So OrderedDict was important.

Now:

Python dictionaries already preserve insertion order.

But OrderedDict still provides extra methods.

---

# Example

```python
from collections import OrderedDict

od = OrderedDict()
od['a'] = 1
od['b'] = 2
od['c'] = 3

print(od)
```

---

# move_to_end()

```python
od.move_to_end('a')
```

---

# popitem()

```python
od.popitem(last=True)
```

---

# Real Use Cases

- LRU cache
- maintaining order-sensitive data
- cache systems

# 6. ChainMap

# What is ChainMap?

Combines multiple dictionaries into one view.

---

# Example

```python
from collections import ChainMap

d1 = {'a':1, 'b':2}
d2 = {'b':20, 'c':30}

cm = ChainMap(d1, d2)

print(cm['a'])
print(cm['b'])
print(cm['c'])
```

Output:

```python
1
2
30
```

---

# Important Rule

Search happens left to right.

---

# Real Use Cases

- configuration management
- nested scopes
- layered settings
- combining environments

# 7. UserDict

# What is UserDict?

Wrapper around dictionary.

Used when creating custom dictionary behavior.

---

# Example

```python
from collections import UserDict

class MyDict(UserDict):
    def __setitem__(self, key, value):
        super().__setitem__(key, value * 2)

md = MyDict()
md['a'] = 10

print(md)
```

Output:

```python
{'a': 20}
```

---

# 8. UserList

Wrapper around list.

Useful for customizing list behavior.

---

# Example

```python
from collections import UserList

class MyList(UserList):
    def append(self, item):
        super().append(item * 2)

ml = MyList()
ml.append(5)

print(ml)
```

Output:

```python
[10]
```

---

# 9. UserString

Wrapper around string.

Used for custom string behavior.

---

# Example

```python
from collections import UserString

class MyString(UserString):
    def upper(self):
        return self.data.upper() + '!!!'

s = MyString('hello')

print(s.upper())
```

Output:

```python
HELLO!!!
```

# Common Interview Questions

# Why is deque faster than list at left insertion?

Because lists use contiguous memory arrays.

Inserting at beginning shifts all elements.

deque uses linked block architecture.

So insertion/removal at both ends is O(1).
